```python
# ============================================================
# PHASE 2B — CREATE THE PAPER'S CHUNKER
# ============================================================

CHUNK_SIZE = 1024
CHUNK_OVERLAP = 0
SEPARATORS = ["\n\n", "\n", " ", ""]

splitter = RecursiveCharacterTextSplitter(
    chunk_size=CHUNK_SIZE,
    chunk_overlap=CHUNK_OVERLAP,
    separators=SEPARATORS
)

print("Chunker configuration:")
print("  chunk_size:", CHUNK_SIZE)
print("  chunk_overlap:", CHUNK_OVERLAP)
print("  separators:", SEPARATORS)
```

```python
# ============================================================
# PHASE 2C — TEST CHUNKING ON ONE TRANSCRIPT
# ============================================================

sample_path = transcript_files[0]

with open(
    sample_path,
    "r",
    encoding="utf-8",
    errors="replace"
) as f:
    sample_text = f.read()

sample_chunks = splitter.split_text(sample_text)

print("Transcript:", os.path.basename(sample_path))
print("Original characters:", len(sample_text))
print("Number of chunks:", len(sample_chunks))

print("\nChunk length statistics:")
chunk_lengths = [len(chunk) for chunk in sample_chunks]

print("Minimum:", min(chunk_lengths))
print("Maximum:", max(chunk_lengths))
print("Average:", sum(chunk_lengths) / len(chunk_lengths))

print("\nFirst 5 chunks:")
for i, chunk in enumerate(sample_chunks[:5]):
    print("\n" + "=" * 80)
    print(f"CHUNK {i}")
    print(f"Characters: {len(chunk)}")
    print("=" * 80)
    print(chunk)
```

```python
# ============================================================
# PHASE 2D — CHUNK ALL 50 TRANSCRIPTS
# ============================================================

all_chunks = []

for transcript_path in transcript_files:

    filename = os.path.basename(transcript_path)

    # Read transcript
    with open(
        transcript_path,
        "r",
        encoding="utf-8",
        errors="replace"
    ) as f:
        text = f.read()

    # Split using the validated paper configuration
    chunks = splitter.split_text(text)

    # Store every chunk with its source metadata
    for chunk_index, chunk_text in enumerate(chunks):

        all_chunks.append({
            "company": None,  # We will derive/validate this separately
            "transcript": filename,
            "chunk_id": f"{Path(filename).stem}_chunk_{chunk_index:04d}",
            "chunk_index": chunk_index,
            "text": chunk_text,
            "char_count": len(chunk_text)
        })

print("Total chunks created:", len(all_chunks))
```

```python
# ============================================================
# PHASE 2E — CHUNK VALIDATION
# ============================================================

chunks_df = pd.DataFrame(all_chunks)

print("Chunk dataframe shape:", chunks_df.shape)

print("\nChunk length statistics:")
print(chunks_df["char_count"].describe())

print("\nMaximum chunk length:", chunks_df["char_count"].max())
print("Minimum chunk length:", chunks_df["char_count"].min())

print("\nChunks exceeding 1024 characters:")
print((chunks_df["char_count"] > 1024).sum())

print("\nTranscripts represented:")
print(chunks_df["transcript"].nunique())
```

```python
# ============================================================
# PHASE 2F — CORPUS METADATA
# 2F-1: Build inventory of all available transcripts
# ============================================================

transcript_inventory = []

for transcript_path in transcript_files:

    filename = os.path.basename(transcript_path)

    transcript_inventory.append({
        "transcript": filename,
        "file_path": transcript_path
    })

transcript_inventory_df = pd.DataFrame(transcript_inventory)

print("Available transcript files:", len(transcript_inventory_df))

display(
    transcript_inventory_df.head(10)
)
```

```python
# ============================================================
# PHASE 2F-2 — PARSE COMPANY METADATA FROM FILENAMES
# ============================================================

import re

def parse_transcript_metadata(filename):
    """
    Convert:
        01_Reliance_Industries_Q1_FY24.txt

    into:
        company = Reliance Industries
        transcript = original filename
    """

    stem = os.path.splitext(filename)[0]

    # Remove the numeric prefix, e.g. "01_"
    stem = re.sub(r"^\d+_", "", stem)

    # Remove the quarter/year suffix
    company_name = re.sub(r"_Q1_FY24$", "", stem)

    # Convert underscores to spaces
    company_name = company_name.replace("_", " ")

    return company_name


transcript_inventory_df["company"] = (
    transcript_inventory_df["transcript"]
    .apply(parse_transcript_metadata)
)

print("Transcript metadata created:")
display(transcript_inventory_df.head(15))
```

```python
# ============================================================
# PHASE 2F-3 — CHECK COMPANY/TICKER COVERAGE
# ============================================================

# Build a unique company → ticker reference from ground truth.
company_ticker_reference = (
    ground_truth[["company", "ticker"]]
    .drop_duplicates()
    .sort_values("company")
    .reset_index(drop=True)
)

print("Unique companies with ticker information in ground truth:",
      len(company_ticker_reference))

print("\nCompanies in ground truth:")
display(company_ticker_reference)
```

```python
# ============================================================
# PHASE 2F-3B — FIND THE 3 COMPANIES WITHOUT TICKERS
# ============================================================

corpus_companies = set(
    transcript_inventory_df["company"]
)

known_companies = set(
    company_ticker_reference["company"]
)

companies_without_ticker = sorted(
    corpus_companies - known_companies
)

print(
    "Corpus companies without ticker information in ground truth:",
    len(companies_without_ticker)
)

for company in companies_without_ticker:
    print(" -", company)
```

```python
# ============================================================
# PHASE 2F-4 — COMPLETE CORPUS COMPANY/TICKER METADATA
# ============================================================

# Explicit ticker mapping for the 3 transcripts that have
# no questions in ground_truth_400.csv.
#
# These are corpus metadata only; they are not being inferred
# from the 400-question dataset.

manual_ticker_mapping = {
    "Power Grid Corporation": "POWERGRID",
    "Reliance Industries": "RELIANCE",
    "Tata Motors": "TATAMOTORS",
}

# Start with the 42 company/ticker pairs available in ground truth.
company_ticker_map = dict(
    zip(
        company_ticker_reference["company"],
        company_ticker_reference["ticker"]
    )
)

# Add the 3 missing corpus companies explicitly.
company_ticker_map.update(manual_ticker_mapping)

# Attach ticker to the 45-transcript inventory.
transcript_inventory_df["ticker"] = (
    transcript_inventory_df["company"]
    .map(company_ticker_map)
)

print("Total corpus transcripts:", len(transcript_inventory_df))
print(
    "Transcripts with missing company:",
    transcript_inventory_df["company"].isna().sum()
)
print(
    "Transcripts with missing ticker:",
    transcript_inventory_df["ticker"].isna().sum()
)

print("\nComplete corpus metadata:")
display(
    transcript_inventory_df[
        ["transcript", "company", "ticker"]
    ]
)
```

```python
# ============================================================
# PHASE 2F-5 — FINAL CORPUS METADATA VALIDATION
# ============================================================

print("Total transcripts:", len(transcript_inventory_df))
print(
    "Unique transcripts:",
    transcript_inventory_df["transcript"].nunique()
)
print(
    "Unique companies:",
    transcript_inventory_df["company"].nunique()
)
print(
    "Unique tickers:",
    transcript_inventory_df["ticker"].nunique()
)

print("\nMissing values:")
print(
    transcript_inventory_df[
        ["transcript", "company", "ticker"]
    ].isna().sum()
)

print("\nDuplicate transcript entries:")
print(
    transcript_inventory_df[
        transcript_inventory_df["transcript"].duplicated(keep=False)
    ]
)
```

```python
# ============================================================
# PHASE 2F-6 — ATTACH METADATA TO ALL CHUNKS
# ============================================================

# Start from the clean chunks produced in Phase 2D.
chunks_df = pd.DataFrame(all_chunks).copy()

# Remove the placeholder company column created during chunking.
chunks_df = chunks_df.drop(columns=["company"])

# Keep only the metadata needed for the corpus.
metadata_for_chunks = transcript_inventory_df[
    ["transcript", "company", "ticker"]
].copy()

# Attach company and ticker using the transcript filename.
chunks_df = chunks_df.merge(
    metadata_for_chunks,
    on="transcript",
    how="left",
    validate="many_to_one"
)

print("Final chunk dataframe shape:", chunks_df.shape)

print("\nColumns:")
print(chunks_df.columns.tolist())

print("\nMissing metadata:")
print(
    chunks_df[
        ["transcript", "company", "ticker"]
    ].isna().sum()
)

print("\nFirst 5 chunks:")
display(chunks_df.head())
```

```python
# ============================================================
# PHASE 2F-7 — FINAL CHUNK METADATA INTEGRITY CHECK
# ============================================================

print("Total chunks:", len(chunks_df))
print("Unique transcripts:", chunks_df["transcript"].nunique())
print("Unique companies:", chunks_df["company"].nunique())
print("Unique tickers:", chunks_df["ticker"].nunique())

print("\nMissing values:")
print(
    chunks_df[
        ["transcript", "company", "ticker"]
    ].isna().sum()
)

# Verify that every transcript has exactly one company/ticker.
metadata_consistency = (
    chunks_df
    .groupby("transcript")
    .agg(
        companies=("company", "nunique"),
        tickers=("ticker", "nunique")
    )
    .reset_index()
)

print("\nTranscripts with inconsistent metadata:")
display(
    metadata_consistency[
        (metadata_consistency["companies"] != 1) |
        (metadata_consistency["tickers"] != 1)
    ]
)
```

```python
# ================================================================
# PHASE 3.4 — GENERATING QUESTION EMBEDDINGS
# ================================================================

import os
import numpy as np
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer

print("=" * 70)
print("PHASE 3.4 — GENERATING QUESTION EMBEDDINGS")
print("=" * 70)

# ------------------------------------------------
# Paths
# ------------------------------------------------

DATA_DIR = "/kaggle/input/datasets/jasapologize/financial-replication-data"

GROUND_TRUTH_PATH = os.path.join(
    DATA_DIR,
    "ground_truth_400.csv"
)

QUESTION_EMBEDDINGS_PATH = (
    "/kaggle/working/question_embeddings.npy"
)

QUESTION_METADATA_PATH = (
    "/kaggle/working/question_embedding_metadata.csv"
)

EMBEDDING_MODEL = "sentence-transformers/all-mpnet-base-v2"

# ------------------------------------------------
# Load ground truth
# ------------------------------------------------

ground_truth_df = pd.read_csv(GROUND_TRUTH_PATH)

print(f"Ground truth shape: {ground_truth_df.shape}")

assert len(ground_truth_df) == 400
assert "question_id" in ground_truth_df.columns
assert "question" in ground_truth_df.columns
assert "company" in ground_truth_df.columns
assert "ticker" in ground_truth_df.columns

questions = ground_truth_df["question"].astype(str).tolist()

print(f"Questions:           {len(questions)}")

# ------------------------------------------------
# Load embedding model
# ------------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Embedding model:     {EMBEDDING_MODEL}")
print(f"Device:              {device}")

question_embedder = SentenceTransformer(
    EMBEDDING_MODEL,
    device=device
)

embedding_dimension = (
    question_embedder.get_sentence_embedding_dimension()
)

print(f"Embedding dimension: {embedding_dimension}")

# ------------------------------------------------
# Generate embeddings
# ------------------------------------------------

print("\nGenerating question embeddings...")

question_embeddings = question_embedder.encode(
    questions,
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True
).astype(np.float32)

print(f"Raw shape: {question_embeddings.shape}")

# ------------------------------------------------
# Normalize for cosine similarity
# ------------------------------------------------

norms = np.linalg.norm(
    question_embeddings,
    axis=1,
    keepdims=True
)

question_embeddings = (
    question_embeddings / (norms + 1e-8)
)

# ------------------------------------------------
# Save
# ------------------------------------------------

np.save(
    QUESTION_EMBEDDINGS_PATH,
    question_embeddings
)

ground_truth_df[
    ["question_id", "company", "ticker"]
].to_csv(
    QUESTION_METADATA_PATH,
    index=False
)

# ------------------------------------------------
# Validation
# ------------------------------------------------

saved_embeddings = np.load(
    QUESTION_EMBEDDINGS_PATH
)

assert saved_embeddings.shape == (400, 768)
assert np.isfinite(saved_embeddings).all()

norms_check = np.linalg.norm(
    saved_embeddings,
    axis=1
)

assert np.allclose(
    norms_check,
    1.0,
    atol=1e-5
)

metadata_check = pd.read_csv(
    QUESTION_METADATA_PATH
)

assert (
    metadata_check["question_id"].tolist()
    == ground_truth_df["question_id"].tolist()
)

print("\n" + "=" * 70)
print("PHASE 3.4 CHECK PASSED")
print("=" * 70)

print("✓ 400 questions embedded")
print(f"✓ Shape: {saved_embeddings.shape}")
print("✓ Dimension: 768")
print("✓ No NaN/Inf values")
print("✓ Embeddings normalized")
print("✓ Question ordering verified")
print("✓ question_embeddings.npy saved")
print("=" * 70)
```

```python
# ================================================================
# PHASE 4.2 — MMR CONFIGURATION + FUNCTION
# ================================================================

import numpy as np

print("=" * 70)
print("PHASE 4.2 — MMR CONFIGURATION")
print("=" * 70)

# ------------------------------------------------
# Retrieval parameters
# ------------------------------------------------

FETCH_K = 20       # Paper: retrieve 20 candidates
FINAL_K = 4        # Paper: use 4 chunks in final context
MMR_LAMBDA = 0.5   # Implementation assumption
                   # Paper does not specify lambda

print(f"Candidate chunks (fetch_k): {FETCH_K}")
print(f"Final chunks (k):           {FINAL_K}")
print(f"MMR lambda:                 {MMR_LAMBDA}")
print("MMR lambda = 0.5 is an implementation assumption.")
print("The paper specifies 20 candidates and 4 final chunks.")

# ------------------------------------------------
# MMR function
# ------------------------------------------------

def maximal_marginal_relevance(
    query_embedding,
    candidate_embeddings,
    candidate_indices,
    lambda_mult=MMR_LAMBDA,
    k=FINAL_K
):
    """
    Select k chunks using Maximal Marginal Relevance.

    Relevance:
        similarity between query and candidate

    Diversity:
        similarity between candidate and already-selected chunks

    MMR:
        lambda * relevance
        - (1 - lambda) * diversity
    """

    candidate_indices = np.asarray(candidate_indices)

    if len(candidate_indices) == 0:
        return []

    k = min(k, len(candidate_indices))

    # ------------------------------------------------
    # Query-candidate relevance
    # ------------------------------------------------

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32
    )

    candidate_embeddings = np.asarray(
        candidate_embeddings,
        dtype=np.float32
    )

    relevance_scores = (
        candidate_embeddings @ query_embedding
    )

    # ------------------------------------------------
    # Candidate-candidate similarity
    # ------------------------------------------------

    candidate_similarity = (
        candidate_embeddings @ candidate_embeddings.T
    )

    # ------------------------------------------------
    # First selection = highest relevance
    # ------------------------------------------------

    selected_positions = [
        int(np.argmax(relevance_scores))
    ]

    # ------------------------------------------------
    # Iteratively select remaining chunks
    # ------------------------------------------------

    while len(selected_positions) < k:

        best_position = None
        best_score = -np.inf

        for position in range(len(candidate_indices)):

            if position in selected_positions:
                continue

            relevance = relevance_scores[position]

            # Similarity to the most similar
            # already-selected candidate
            diversity_penalty = max(
                candidate_similarity[
                    position,
                    selected_positions
                ]
            )

            mmr_score = (
                lambda_mult * relevance
                - (1 - lambda_mult) * diversity_penalty
            )

            if mmr_score > best_score:
                best_score = mmr_score
                best_position = position

        selected_positions.append(best_position)

    # Convert local candidate positions
    # back to global chunk indices.
    selected_global_indices = [
        int(candidate_indices[position])
        for position in selected_positions
    ]

    return selected_global_indices


# ------------------------------------------------
# Smoke test
# ------------------------------------------------

test_question_embedding = question_embeddings[0]

# Use the first 20 chunks simply to test the function.
# This is NOT the actual retrieval.
test_candidate_indices = np.arange(
    min(FETCH_K, len(chunks_df))
)

test_candidate_embeddings = (
    chunk_embeddings[test_candidate_indices]
)

test_selected = maximal_marginal_relevance(
    query_embedding=test_question_embedding,
    candidate_embeddings=test_candidate_embeddings,
    candidate_indices=test_candidate_indices,
    lambda_mult=MMR_LAMBDA,
    k=FINAL_K
)

# ------------------------------------------------
# Validation
# ------------------------------------------------

assert len(test_selected) == FINAL_K
assert len(set(test_selected)) == FINAL_K

for idx in test_selected:
    assert 0 <= idx < len(chunks_df)

print("\n" + "=" * 70)
print("PHASE 4.2 CHECK PASSED")
print("=" * 70)

print(f"✓ MMR function defined")
print(f"✓ fetch_k = {FETCH_K}")
print(f"✓ final_k = {FINAL_K}")
print(f"✓ MMR lambda = {MMR_LAMBDA}")
print(f"✓ Smoke test selected {len(test_selected)} unique chunks")
print("=" * 70)
```

```python
# ================================================================
# PHASE 4.3 — RETRIEVAL: WITHOUT_METADATA vs WITH_METADATA
# ================================================================

import os
import numpy as np
import pandas as pd

print("=" * 70)
print("PHASE 4.3 — RETRIEVAL: WITHOUT_METADATA vs WITH_METADATA")
print("=" * 70)

# ----------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------

CHUNKS_PATH = "/kaggle/working/chunks.csv"
CHUNK_METADATA_PATH = "/kaggle/working/chunk_metadata.csv"
CHUNK_EMBEDDINGS_PATH = "/kaggle/working/chunk_embeddings.npy"
QUESTION_EMBEDDINGS_PATH = "/kaggle/working/question_embeddings.npy"
GROUND_TRUTH_PATH = (
    "/kaggle/input/datasets/jasapologize/"
    "financial-replication-data/ground_truth_400.csv"
)

WITHOUT_METADATA_PATH = "/kaggle/working/without_metadata_retrieval.csv"
WITH_METADATA_PATH = "/kaggle/working/with_metadata_retrieval.csv"

# ----------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------

chunks_df = pd.read_csv(CHUNKS_PATH)
chunk_metadata_df = pd.read_csv(CHUNK_METADATA_PATH)
chunk_embeddings = np.load(CHUNK_EMBEDDINGS_PATH)
question_embeddings = np.load(QUESTION_EMBEDDINGS_PATH)
ground_truth_df = pd.read_csv(GROUND_TRUTH_PATH)

print(f"Chunks:              {chunks_df.shape}")
print(f"Chunk metadata:      {chunk_metadata_df.shape}")
print(f"Chunk embeddings:    {chunk_embeddings.shape}")
print(f"Question embeddings: {question_embeddings.shape}")
print(f"Ground truth:        {ground_truth_df.shape}")

# ----------------------------------------------------------------
# VALIDATION
# ----------------------------------------------------------------

assert len(chunks_df) == len(chunk_embeddings)
assert len(chunks_df) == len(chunk_metadata_df)
assert len(ground_truth_df) == len(question_embeddings)

assert chunk_embeddings.shape[1] == question_embeddings.shape[1]
assert "question_id" in ground_truth_df.columns
assert "company" in ground_truth_df.columns
assert "ticker" in ground_truth_df.columns
assert "chunk_id" in chunks_df.columns
assert "company" in chunks_df.columns
assert "ticker" in chunks_df.columns

assert chunks_df["chunk_id"].is_unique
assert ground_truth_df["question_id"].is_unique

# ----------------------------------------------------------------
# NORMALIZE EMBEDDINGS
# ----------------------------------------------------------------
# After normalization, dot product = cosine similarity.

chunk_norms = np.linalg.norm(
    chunk_embeddings, axis=1, keepdims=True
)

question_norms = np.linalg.norm(
    question_embeddings, axis=1, keepdims=True
)

chunk_embeddings_normalized = (
    chunk_embeddings / (chunk_norms + 1e-8)
)

question_embeddings_normalized = (
    question_embeddings / (question_norms + 1e-8)
)

assert np.isfinite(chunk_embeddings_normalized).all()
assert np.isfinite(question_embeddings_normalized).all()

# ----------------------------------------------------------------
# RETRIEVE TOP-20 BY COSINE SIMILARITY
# ----------------------------------------------------------------

def retrieve_candidates(
    query_embedding,
    candidate_indices,
    fetch_k=20
):
    """
    Retrieve the top fetch_k chunks using cosine similarity.

    candidate_indices:
        Chunks allowed to participate in retrieval.
        For WITHOUT_METADATA: all chunks.
        For WITH_METADATA: only chunks belonging to
        the question's company.
    """

    candidate_indices = np.asarray(
        candidate_indices,
        dtype=np.int64
    )

    candidate_vectors = chunk_embeddings_normalized[
        candidate_indices
    ]

    # Because both vectors are normalized:
    # dot product = cosine similarity.
    similarities = candidate_vectors @ query_embedding

    top_n = min(fetch_k, len(candidate_indices))

    top_positions = np.argsort(
        -similarities
    )[:top_n]

    top_indices = candidate_indices[top_positions]
    top_scores = similarities[top_positions]

    return top_indices, top_scores


# ----------------------------------------------------------------
# RUN BOTH RETRIEVAL CONDITIONS
# ----------------------------------------------------------------

FETCH_K = 20
FINAL_K = 4

without_metadata_rows = []
with_metadata_rows = []

all_chunk_indices = np.arange(len(chunks_df))

print("\nStarting retrieval...")
print(f"Questions: {len(ground_truth_df)}")
print(f"Candidate chunks: {FETCH_K}")
print(f"Final chunks: {FINAL_K}")

# ----------------------------------------------------------------
# QUESTION-BY-QUESTION RETRIEVAL
# ----------------------------------------------------------------

for q_idx, row in ground_truth_df.iterrows():

    question_id = row["question_id"]
    company = row["company"]
    ticker = row["ticker"]

    query_embedding = question_embeddings_normalized[q_idx]

    # ============================================================
    # CONDITION 1 — WITHOUT_METADATA
    # ============================================================

    # No company restriction.
    # All transcript chunks are eligible.

    candidate_indices = all_chunk_indices

    top20_indices, top20_scores = retrieve_candidates(
        query_embedding=query_embedding,
        candidate_indices=candidate_indices,
        fetch_k=FETCH_K
    )

    top20_embeddings = chunk_embeddings_normalized[
        top20_indices
    ]

    # MMR selects the final 4 from the 20 candidates.
    selected_positions = maximal_marginal_relevance(
        query_embedding=query_embedding,
        candidate_embeddings=top20_embeddings,
        candidate_indices=np.arange(len(top20_indices)),
        lambda_mult=MMR_LAMBDA,
        k=FINAL_K
    )

    selected_indices = top20_indices[selected_positions]

    # Save final 4
    for rank, chunk_idx in enumerate(selected_indices, start=1):

        chunk_row = chunks_df.iloc[chunk_idx]

        # Similarity of this final chunk to the question.
        # This is NOT the MMR score.
        similarity_score = float(
            chunk_embeddings_normalized[chunk_idx]
            @ query_embedding
        )

        without_metadata_rows.append({
            "question_id": question_id,
            "company": company,
            "ticker": ticker,
            "condition": "WITHOUT_METADATA",
            "rank": rank,
            "chunk_id": chunk_row["chunk_id"],
            "chunk_index": chunk_row["chunk_index"],
            "transcript": chunk_row["transcript"],
            "retrieved_company": chunk_row["company"],
            "retrieved_ticker": chunk_row["ticker"],
            "similarity_score": similarity_score,
            "text": chunk_row["text"]
        })

    # ============================================================
    # CONDITION 2 — WITH_METADATA
    # ============================================================

    # Metadata filter:
    # only chunks belonging to the company associated
    # with the question are allowed into retrieval.

    metadata_mask = (
        chunks_df["company"] == company
    )

    metadata_candidate_indices = np.flatnonzero(
        metadata_mask.to_numpy()
    )

    assert len(metadata_candidate_indices) > 0, (
        f"No chunks found for company: {company}"
    )

    # Similarity search is performed ONLY inside the
    # metadata-filtered company documents.
    top20_indices, top20_scores = retrieve_candidates(
        query_embedding=query_embedding,
        candidate_indices=metadata_candidate_indices,
        fetch_k=FETCH_K
    )

    top20_embeddings = chunk_embeddings_normalized[
        top20_indices
    ]

    # MMR selects the final 4 from those 20.
    selected_positions = maximal_marginal_relevance(
        query_embedding=query_embedding,
        candidate_embeddings=top20_embeddings,
        candidate_indices=np.arange(len(top20_indices)),
        lambda_mult=MMR_LAMBDA,
        k=FINAL_K
    )

    selected_indices = top20_indices[selected_positions]

    # Critical validation:
    # metadata condition must not retrieve another company.
    retrieved_companies = chunks_df.iloc[
        selected_indices
    ]["company"].unique()

    assert len(retrieved_companies) == 1
    assert retrieved_companies[0] == company

    # Save final 4
    for rank, chunk_idx in enumerate(selected_indices, start=1):

        chunk_row = chunks_df.iloc[chunk_idx]

        similarity_score = float(
            chunk_embeddings_normalized[chunk_idx]
            @ query_embedding
        )

        with_metadata_rows.append({
            "question_id": question_id,
            "company": company,
            "ticker": ticker,
            "condition": "WITH_METADATA",
            "rank": rank,
            "chunk_id": chunk_row["chunk_id"],
            "chunk_index": chunk_row["chunk_index"],
            "transcript": chunk_row["transcript"],
            "retrieved_company": chunk_row["company"],
            "retrieved_ticker": chunk_row["ticker"],
            "similarity_score": similarity_score,
            "text": chunk_row["text"]
        })


# ----------------------------------------------------------------
# CREATE DATAFRAMES
# ----------------------------------------------------------------

without_metadata_df = pd.DataFrame(
    without_metadata_rows
)

with_metadata_df = pd.DataFrame(
    with_metadata_rows
)

# ----------------------------------------------------------------
# SAVE RESULTS
# ----------------------------------------------------------------

without_metadata_df.to_csv(
    WITHOUT_METADATA_PATH,
    index=False
)

with_metadata_df.to_csv(
    WITH_METADATA_PATH,
    index=False
)

# ----------------------------------------------------------------
# VALIDATION
# ----------------------------------------------------------------

expected_rows = len(ground_truth_df) * FINAL_K

assert len(without_metadata_df) == expected_rows
assert len(with_metadata_df) == expected_rows

assert (
    without_metadata_df.groupby("question_id")
    .size()
    .eq(FINAL_K)
    .all()
)

assert (
    with_metadata_df.groupby("question_id")
    .size()
    .eq(FINAL_K)
    .all()
)

assert set(without_metadata_df["rank"]) == {1, 2, 3, 4}
assert set(with_metadata_df["rank"]) == {1, 2, 3, 4}

# WITH_METADATA must have zero cross-company retrieval.
with_cross_company = (
    with_metadata_df["retrieved_company"]
    != with_metadata_df["company"]
).sum()

assert with_cross_company == 0

# WITHOUT_METADATA leakage analysis.
without_cross_company = (
    without_metadata_df["retrieved_company"]
    != without_metadata_df["company"]
).sum()

# ----------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 4.3 CHECK PASSED")
print("=" * 70)

print(f"✓ Questions processed:              {len(ground_truth_df)}")
print(f"✓ WITHOUT_METADATA rows:            {len(without_metadata_df)}")
print(f"✓ WITH_METADATA rows:               {len(with_metadata_df)}")
print(f"✓ Candidates per question:          {FETCH_K}")
print(f"✓ Final chunks per question:        {FINAL_K}")
print(f"✓ MMR lambda:                       {MMR_LAMBDA}")

print("\nRetrieval pipeline:")
print("  WITHOUT_METADATA:")
print("    All chunks → similarity top-20 → MMR → final 4")

print("\n  WITH_METADATA:")
print("    Company filter → similarity top-20 → MMR → final 4")

print("\nMetadata validation:")
print(f"  WITH_METADATA cross-company chunks: {with_cross_company}")

print("\nWITHOUT_METADATA leakage:")
print(f"  Cross-company chunks:               {without_cross_company}")

print("\nSaved:")
print(f"  ✓ {WITHOUT_METADATA_PATH}")
print(f"  ✓ {WITH_METADATA_PATH}")

print("=" * 70)
```

```python
# ================================================================
# PHASE 4.4 — RETRIEVAL VALIDATION & ANALYSIS
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("PHASE 4.4 — RETRIEVAL VALIDATION & ANALYSIS")
print("=" * 70)

# ----------------------------------------------------------------
# LOAD RETRIEVAL RESULTS
# ----------------------------------------------------------------

WITHOUT_METADATA_PATH = "/kaggle/working/without_metadata_retrieval.csv"
WITH_METADATA_PATH = "/kaggle/working/with_metadata_retrieval.csv"

without_df = pd.read_csv(WITHOUT_METADATA_PATH)
with_df = pd.read_csv(WITH_METADATA_PATH)

print(f"WITHOUT_METADATA: {without_df.shape}")
print(f"WITH_METADATA:    {with_df.shape}")

# ----------------------------------------------------------------
# VALIDATE BASIC STRUCTURE
# ----------------------------------------------------------------

assert len(without_df) == 1600
assert len(with_df) == 1600

assert without_df["question_id"].nunique() == 400
assert with_df["question_id"].nunique() == 400

assert set(without_df["rank"]) == {1, 2, 3, 4}
assert set(with_df["rank"]) == {1, 2, 3, 4}

# ----------------------------------------------------------------
# CORRECT COMPANY / TICKER CHECK
# ----------------------------------------------------------------

without_df["correct_company"] = (
    without_df["retrieved_company"]
    == without_df["company"]
)

with_df["correct_company"] = (
    with_df["retrieved_company"]
    == with_df["company"]
)

without_df["correct_ticker"] = (
    without_df["retrieved_ticker"]
    == without_df["ticker"]
)

with_df["correct_ticker"] = (
    with_df["retrieved_ticker"]
    == with_df["ticker"]
)

# ----------------------------------------------------------------
# CHUNK-LEVEL RETRIEVAL ACCURACY
# ----------------------------------------------------------------

without_company_accuracy = (
    without_df["correct_company"].mean()
)

with_company_accuracy = (
    with_df["correct_company"].mean()
)

without_ticker_accuracy = (
    without_df["correct_ticker"].mean()
)

with_ticker_accuracy = (
    with_df["correct_ticker"].mean()
)

# ----------------------------------------------------------------
# QUESTION-LEVEL RETRIEVAL RECALL
# ----------------------------------------------------------------
# Did at least ONE of the final 4 chunks come from
# the correct company?

without_question_recall = (
    without_df
    .groupby("question_id")["correct_company"]
    .any()
    .mean()
)

with_question_recall = (
    with_df
    .groupby("question_id")["correct_company"]
    .any()
    .mean()
)

# ----------------------------------------------------------------
# RANK-1 ACCURACY
# ----------------------------------------------------------------

without_rank1 = (
    without_df[without_df["rank"] == 1]["correct_company"].mean()
)

with_rank1 = (
    with_df[with_df["rank"] == 1]["correct_company"].mean()
)

# ----------------------------------------------------------------
# RANK-WISE ACCURACY
# ----------------------------------------------------------------

rank_results = []

for rank in [1, 2, 3, 4]:

    without_rank_accuracy = (
        without_df[without_df["rank"] == rank]
        ["correct_company"]
        .mean()
    )

    with_rank_accuracy = (
        with_df[with_df["rank"] == rank]
        ["correct_company"]
        .mean()
    )

    rank_results.append({
        "rank": rank,
        "without_metadata": without_rank_accuracy,
        "with_metadata": with_rank_accuracy
    })

rank_df = pd.DataFrame(rank_results)

# ----------------------------------------------------------------
# CROSS-COMPANY LEAKAGE
# ----------------------------------------------------------------

without_leakage_chunks = (
    (~without_df["correct_company"]).sum()
)

with_leakage_chunks = (
    (~with_df["correct_company"]).sum()
)

without_leakage_rate = (
    without_leakage_chunks / len(without_df)
)

with_leakage_rate = (
    with_leakage_chunks / len(with_df)
)

# ----------------------------------------------------------------
# QUESTIONS WITH AT LEAST ONE WRONG-COMPANY CHUNK
# ----------------------------------------------------------------

without_question_leakage = (
    without_df
    .groupby("question_id")["correct_company"]
    .apply(lambda x: (~x).any())
    .sum()
)

with_question_leakage = (
    with_df
    .groupby("question_id")["correct_company"]
    .apply(lambda x: (~x).any())
    .sum()
)

# ----------------------------------------------------------------
# QUESTIONS WITH ALL FOUR CORRECT
# ----------------------------------------------------------------

without_all_four_correct = (
    without_df
    .groupby("question_id")["correct_company"]
    .all()
    .sum()
)

with_all_four_correct = (
    with_df
    .groupby("question_id")["correct_company"]
    .all()
    .sum()
)

# ----------------------------------------------------------------
# SAVE RANK-WISE ANALYSIS
# ----------------------------------------------------------------

RANK_RESULTS_PATH = "/kaggle/working/retrieval_rank_analysis.csv"

rank_df.to_csv(
    RANK_RESULTS_PATH,
    index=False
)

# ----------------------------------------------------------------
# PRINT RESULTS
# ----------------------------------------------------------------

print("\n" + "=" * 70)
print("RETRIEVAL VALIDATION RESULTS")
print("=" * 70)

print("\n1. CHUNK-LEVEL CORRECT COMPANY")

print(
    f"WITHOUT_METADATA: "
    f"{without_company_accuracy:.4f} "
    f"({without_company_accuracy * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_company_accuracy:.4f} "
    f"({with_company_accuracy * 100:.2f}%)"
)

print("\n2. CHUNK-LEVEL CORRECT TICKER")

print(
    f"WITHOUT_METADATA: "
    f"{without_ticker_accuracy:.4f} "
    f"({without_ticker_accuracy * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_ticker_accuracy:.4f} "
    f"({with_ticker_accuracy * 100:.2f}%)"
)

print("\n3. QUESTION-LEVEL RECALL")
print("At least one correct-company chunk in final 4:")

print(
    f"WITHOUT_METADATA: "
    f"{without_question_recall:.4f} "
    f"({without_question_recall * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_question_recall:.4f} "
    f"({with_question_recall * 100:.2f}%)"
)

print("\n4. RANK-1 CORRECT COMPANY")

print(
    f"WITHOUT_METADATA: "
    f"{without_rank1:.4f} "
    f"({without_rank1 * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_rank1:.4f} "
    f"({with_rank1 * 100:.2f}%)"
)

print("\n5. CROSS-COMPANY LEAKAGE")

print(
    f"WITHOUT_METADATA: "
    f"{without_leakage_chunks} / {len(without_df)} "
    f"({without_leakage_rate * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_leakage_chunks} / {len(with_df)} "
    f"({with_leakage_rate * 100:.2f}%)"
)

print("\n6. QUESTIONS WITH CROSS-COMPANY LEAKAGE")

print(
    f"WITHOUT_METADATA: "
    f"{without_question_leakage} / 400"
)

print(
    f"WITH_METADATA:    "
    f"{with_question_leakage} / 400"
)

print("\n7. QUESTIONS WHERE ALL 4 CHUNKS ARE CORRECT COMPANY")

print(
    f"WITHOUT_METADATA: "
    f"{without_all_four_correct} / 400"
)

print(
    f"WITH_METADATA:    "
    f"{with_all_four_correct} / 400"
)

print("\n8. RANK-WISE CORRECT COMPANY")

print(rank_df.to_string(index=False))

print("\n" + "=" * 70)
print("PHASE 4.4 CHECK PASSED")
print("=" * 70)

print("✓ 1,600 rows validated per condition")
print("✓ 400 questions validated per condition")
print("✓ Rank 1–4 validated")
print("✓ Company-level retrieval accuracy calculated")
print("✓ Question-level recall calculated")
print("✓ Cross-company leakage calculated")
print("✓ Rank-wise analysis saved")
print(f"✓ Saved: {RANK_RESULTS_PATH}")

print("=" * 70)
```

```python
# ============================================================
# PHASE 5.2 — BUILD 800 LLAMA GENERATION PROMPTS
# ============================================================

import os
import pandas as pd

print("=" * 70)
print("PHASE 5.2 — BUILD LLAMA GENERATION PROMPTS")
print("=" * 70)

# ------------------------------------------------------------
# FILE PATHS
# ------------------------------------------------------------

GT_FILE = (
    "/kaggle/input/datasets/jasapologize/"
    "financial-replication-data/ground_truth_400.csv"
)

possible_without = [
    "/kaggle/working/without_metadata_retrieval.csv",
    "/kaggle/working/retrieval_without_metadata.csv",
]

possible_with = [
    "/kaggle/working/with_metadata_retrieval.csv",
    "/kaggle/working/retrieval_with_metadata.csv",
]

WITHOUT_FILE = next(
    (p for p in possible_without if os.path.exists(p)),
    None
)

WITH_FILE = next(
    (p for p in possible_with if os.path.exists(p)),
    None
)

print("WITHOUT metadata file:", WITHOUT_FILE)
print("WITH metadata file:   ", WITH_FILE)

assert WITHOUT_FILE is not None, (
    "WITHOUT metadata retrieval file not found."
)

assert WITH_FILE is not None, (
    "WITH metadata retrieval file not found."
)

assert os.path.exists(GT_FILE), (
    f"Ground truth file not found: {GT_FILE}"
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

without_df = pd.read_csv(WITHOUT_FILE)
with_df = pd.read_csv(WITH_FILE)
gt_df = pd.read_csv(GT_FILE)

print("\nLoaded rows:")
print("WITHOUT metadata:", len(without_df))
print("WITH metadata:   ", len(with_df))
print("Ground truth:    ", len(gt_df))

# ------------------------------------------------------------
# SHOW COLUMNS
# ------------------------------------------------------------

print("\nWITHOUT columns:")
print(without_df.columns.tolist())

print("\nWITH columns:")
print(with_df.columns.tolist())

# ------------------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------------------

def detect_column(df, candidates, description):

    for column in candidates:
        if column in df.columns:
            return column

    raise ValueError(
        f"\nCould not find {description}.\n"
        f"Tried: {candidates}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# WITHOUT
WO_QCOL = detect_column(
    without_df,
    ["question_id", "question"],
    "question ID"
)

WO_RANKCOL = detect_column(
    without_df,
    ["rank", "mmr_rank", "retrieval_rank"],
    "retrieval rank"
)

WO_TEXTCOL = detect_column(
    without_df,
    ["chunk_text", "text", "chunk"],
    "chunk text"
)

# WITH
W_QCOL = detect_column(
    with_df,
    ["question_id", "question"],
    "question ID"
)

W_RANKCOL = detect_column(
    with_df,
    ["rank", "mmr_rank", "retrieval_rank"],
    "retrieval rank"
)

W_TEXTCOL = detect_column(
    with_df,
    ["chunk_text", "text", "chunk"],
    "chunk text"
)

print("\nDetected columns:")
print("WITHOUT question:", WO_QCOL)
print("WITHOUT rank:    ", WO_RANKCOL)
print("WITHOUT text:    ", WO_TEXTCOL)
print("WITH question:   ", W_QCOL)
print("WITH rank:       ", W_RANKCOL)
print("WITH text:       ", W_TEXTCOL)

# ------------------------------------------------------------
# GROUND-TRUTH QUESTION MAP
# ------------------------------------------------------------

required_gt_columns = [
    "question_id",
    "company",
    "question"
]

for col in required_gt_columns:
    assert col in gt_df.columns, (
        f"Missing ground-truth column: {col}"
    )

question_map = (
    gt_df[
        ["question_id", "company", "question"]
    ]
    .drop_duplicates("question_id")
    .set_index("question_id")
)

assert len(question_map) == 400, (
    f"Expected 400 unique questions, "
    f"found {len(question_map)}"
)

# ------------------------------------------------------------
# RETRIEVAL INTEGRITY CHECK
# ------------------------------------------------------------

def validate_retrieval(
    df,
    qcol,
    rankcol,
    condition
):

    unique_questions = df[qcol].nunique()

    print(f"\n{condition}")
    print("-" * 50)
    print("Unique questions:", unique_questions)

    assert unique_questions == 400

    counts = df.groupby(qcol)[rankcol].nunique()

    print(
        "Questions with exactly 4 chunks:",
        (counts == 4).sum(),
        "/",
        len(counts)
    )

    assert (counts == 4).all(), (
        f"{condition}: every question must have exactly 4 chunks."
    )

    ranks = sorted(df[rankcol].dropna().unique())

    print("Ranks found:", ranks)

validate_retrieval(
    without_df,
    WO_QCOL,
    WO_RANKCOL,
    "WITHOUT METADATA"
)

validate_retrieval(
    with_df,
    W_QCOL,
    W_RANKCOL,
    "WITH METADATA"
)

# ------------------------------------------------------------
# SORT RETRIEVED CHUNKS
# ------------------------------------------------------------

without_df = without_df.sort_values(
    [WO_QCOL, WO_RANKCOL]
).reset_index(drop=True)

with_df = with_df.sort_values(
    [W_QCOL, W_RANKCOL]
).reset_index(drop=True)

# ------------------------------------------------------------
# GENERATION PROMPT
#
# IMPORTANT:
# SAME prompt template for both conditions.
#
# The ONLY experimental difference is the retrieved context.
# ------------------------------------------------------------

PROMPT_TEMPLATE = """You are answering a question based on a financial earnings call transcript.

Use only the information provided in the context.

Provide a concise but complete answer to the question.
Do not answer with only "Yes" or "No". If the answer is yes or no,
explain the relevant details from the context.

{contexts}

Question:
{question}

Answer:"""

# ------------------------------------------------------------
# BUILD PROMPTS
# ------------------------------------------------------------

def build_prompts(
    retrieval_df,
    qcol,
    rankcol,
    textcol,
    condition
):

    rows = []

    for question_id, group in retrieval_df.groupby(
        qcol,
        sort=False
    ):

        group = group.sort_values(rankcol)

        # Exactly 4 chunks per question
        assert len(group) == 4

        # Question information comes from the GT file
        assert question_id in question_map.index

        question = question_map.loc[
            question_id,
            "question"
        ]

        company = question_map.loc[
            question_id,
            "company"
        ]

        contexts = []

        for i, (_, row) in enumerate(
            group.iterrows(),
            start=1
        ):

            chunk_text = str(
                row[textcol]
            ).strip()

            contexts.append(
                f"Context {i}:\n{chunk_text}"
            )

        context_block = "\n\n".join(contexts)

        prompt = PROMPT_TEMPLATE.format(
            contexts=context_block,
            question=question
        )

        rows.append({
            "question_id": question_id,
            "company": company,
            "question": question,
            "condition": condition,
            "prompt": prompt,
            "n_contexts": 4
        })

    return pd.DataFrame(rows)

# ------------------------------------------------------------
# CREATE 400 + 400
# ------------------------------------------------------------

llama_without = build_prompts(
    without_df,
    WO_QCOL,
    WO_RANKCOL,
    WO_TEXTCOL,
    "WITHOUT_METADATA"
)

llama_with = build_prompts(
    with_df,
    W_QCOL,
    W_RANKCOL,
    W_TEXTCOL,
    "WITH_METADATA"
)

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

assert len(llama_without) == 400
assert len(llama_with) == 400

assert llama_without["question_id"].nunique() == 400
assert llama_with["question_id"].nunique() == 400

assert set(
    llama_without["question_id"]
) == set(
    llama_with["question_id"]
)

# ------------------------------------------------------------
# COMBINE
# ------------------------------------------------------------

llama_prompts = pd.concat(
    [
        llama_without,
        llama_with
    ],
    ignore_index=True
)

assert len(llama_prompts) == 800

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

llama_prompts.to_csv(
    PROMPT_FILE,
    index=False
)

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("✅ PHASE 5.2 COMPLETE")
print("=" * 70)

print("WITHOUT metadata prompts:", len(llama_without))
print("WITH metadata prompts:   ", len(llama_with))
print("TOTAL prompts:            ", len(llama_prompts))

print("\nSaved:")
print(PROMPT_FILE)

# ------------------------------------------------------------
# Q008 PREVIEW
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ADANIENT_Q008 PROMPT PREVIEW")
print("=" * 70)

q008 = llama_prompts[
    llama_prompts["question_id"] == "ADANIENT_Q008"
]

for _, row in q008.iterrows():

    print("\n" + "-" * 70)
    print(row["condition"])
    print("-" * 70)
    print(row["prompt"])
```

```python
# ============================================================
# PHASE 5.3 — LLAMA PROMPT TOKEN VALIDATION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("PHASE 5.3 — LLAMA PROMPT TOKEN VALIDATION")
print("=" * 70)

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

prompts_df = pd.read_csv(PROMPT_FILE)

assert len(prompts_df) == 800

# ------------------------------------------------------------
# TOKENIZE ALL PROMPTS
# ------------------------------------------------------------

token_counts = []

for prompt in prompts_df["prompt"]:

    tokens = tokenizer(
        str(prompt),
        add_special_tokens=True,
        return_attention_mask=False
    )["input_ids"]

    token_counts.append(len(tokens))

prompts_df["input_tokens"] = token_counts

# ------------------------------------------------------------
# TOKEN STATISTICS
# ------------------------------------------------------------

print("\nToken statistics:")
print(
    prompts_df["input_tokens"].describe()
)

print("\nMaximum input tokens:",
      prompts_df["input_tokens"].max())

print("Minimum input tokens:",
      prompts_df["input_tokens"].min())

print("Mean input tokens:",
      prompts_df["input_tokens"].mean())

print("Median input tokens:",
      prompts_df["input_tokens"].median())

# ------------------------------------------------------------
# CONTEXT LIMIT CHECK
# ------------------------------------------------------------

CONTEXT_LIMIT = 4096

over_limit = prompts_df[
    prompts_df["input_tokens"] > CONTEXT_LIMIT
]

at_limit = prompts_df[
    prompts_df["input_tokens"] == CONTEXT_LIMIT
]

within_limit = prompts_df[
    prompts_df["input_tokens"] <= CONTEXT_LIMIT
]

print("\n" + "=" * 70)
print("CONTEXT CHECK")
print("=" * 70)

print(
    "Within 4096:",
    len(within_limit),
    "/ 800"
)

print(
    "Exactly 4096:",
    len(at_limit),
    "/ 800"
)

print(
    "Over 4096:",
    len(over_limit),
    "/ 800"
)

# ------------------------------------------------------------
# OUTPUT CAPACITY
#
# IMPORTANT:
# Llama has a 4096-token TOTAL context.
# Therefore available generation space is:
#
# 4096 - input_tokens
#
# We record this rather than blindly assuming 1024 tokens
# can always be generated.
# ------------------------------------------------------------

prompts_df["available_generation_tokens"] = (
    CONTEXT_LIMIT -
    prompts_df["input_tokens"]
)

# ------------------------------------------------------------
# CHECK WHETHER 1024 NEW TOKENS FIT
# ------------------------------------------------------------

prompts_df["can_fit_1024_output"] = (
    prompts_df["input_tokens"] + 1024
    <= CONTEXT_LIMIT
)

print(
    "\nCan fit full 1024-token generation:",
    prompts_df["can_fit_1024_output"].sum(),
    "/ 800"
)

print(
    "Cannot fit full 1024-token generation:",
    (~prompts_df["can_fit_1024_output"]).sum(),
    "/ 800"
)

# ------------------------------------------------------------
# CONDITION-WISE STATS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CONDITION-WISE TOKEN STATISTICS")
print("=" * 70)

for condition, group in prompts_df.groupby("condition"):

    print("\n", condition)
    print("-" * 50)

    print("Questions:", len(group))
    print("Mean input:",
          round(group["input_tokens"].mean(), 2))
    print("Median input:",
          int(group["input_tokens"].median()))
    print("Min input:",
          group["input_tokens"].min())
    print("Max input:",
          group["input_tokens"].max())

    print(
        "Can fit 1024 output:",
        group["can_fit_1024_output"].sum(),
        "/",
        len(group)
    )

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

TOKEN_FILE = "/kaggle/working/llama_prompt_token_validation.csv"

prompts_df.to_csv(
    TOKEN_FILE,
    index=False
)

# ------------------------------------------------------------
# SHOW LONGEST PROMPTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("5 LONGEST PROMPTS")
print("=" * 70)

print(
    prompts_df[
        [
            "question_id",
            "condition",
            "input_tokens",
            "available_generation_tokens"
        ]
    ]
    .sort_values(
        "input_tokens",
        ascending=False
    )
    .head(5)
    .to_string(index=False)
)

print("\nSaved:")
print(TOKEN_FILE)

print("\n" + "=" * 70)
print("✅ PHASE 5.3 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# PHASE 5.8 — LLAMA ANSWER-BOUNDARY HARD STOP
# Prevent continuation into subsequent transcript Q&A
# ================================================================

import gc
import random
import numpy as np
import torch
import pandas as pd
import re

from transformers import StoppingCriteria, StoppingCriteriaList


print("=" * 70)
print("PHASE 5.8 — LLAMA ANSWER-BOUNDARY HARD STOP")
print("=" * 70)


# ------------------------------------------------
# 1. REPRODUCIBILITY
# ------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# ------------------------------------------------
# 2. LOAD PROMPTS
# ------------------------------------------------

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

prompts_df = pd.read_csv(PROMPT_FILE)

q008 = prompts_df[
    prompts_df["question_id"].astype(str).str.upper()
    == "ADANIENT_Q008"
].copy()

if len(q008) != 2:
    raise ValueError(
        f"Expected exactly 2 Q008 prompts, found {len(q008)}"
    )


# ------------------------------------------------
# 3. ROBUST TEXT-BASED STOPPING
# ------------------------------------------------

class TranscriptBoundaryStop(StoppingCriteria):

    def __init__(self, tokenizer, prompt_length):
        super().__init__()

        self.tokenizer = tokenizer
        self.prompt_length = prompt_length

        # Patterns indicating that the model has started
        # generating the next transcript/Q&A section.
        self.patterns = [
            r"\nQuestion\s*:",
            r"\nContext\s+\d+\s*:",
            r"\nModerator\s*:",
        ]

    def __call__(self, input_ids, scores, **kwargs):

        # Only inspect generated tokens,
        # NOT the original prompt.
        generated_ids = input_ids[
            0,
            self.prompt_length:
        ]

        if generated_ids.numel() == 0:
            return False

        # Decode generated portion
        generated_text = self.tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )

        # Only need to inspect the tail.
        # This keeps the check reasonably cheap.
        tail = generated_text[-300:]

        for pattern in self.patterns:

            if re.search(pattern, tail, flags=re.IGNORECASE):
                return True

        return False


# ------------------------------------------------
# 4. GENERATION FUNCTION
# ------------------------------------------------

def generate_llama_with_boundary_stop(prompt):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        add_special_tokens=True
    )

    inputs = {
        k: v.to(model.device)
        for k, v in inputs.items()
    }

    input_tokens = inputs["input_ids"].shape[1]

    stopping_criteria = StoppingCriteriaList([
        TranscriptBoundaryStop(
            tokenizer=tokenizer,
            prompt_length=input_tokens
        )
    ])

    print(f"Input tokens: {input_tokens}")
    print("Maximum new tokens: 256")
    print("Boundary stopping: ENABLED")

    with torch.no_grad():

        outputs = model.generate(
            **inputs,

            max_new_tokens=256,

            do_sample=True,
            temperature=0.7,
            top_p=0.9,

            # Slightly stronger repetition control
            repetition_penalty=1.15,

            stopping_criteria=stopping_criteria,

            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.eos_token_id,

            use_cache=True
        )

    generated_ids = outputs[0][input_tokens:]

    answer = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    )

    # ------------------------------------------------
    # REMOVE ANY TRANSCRIPT BOUNDARY
    # ------------------------------------------------

    boundary_patterns = [
        r"\nQuestion\s*:.*",
        r"\nContext\s+\d+\s*:.*",
        r"\nModerator\s*:.*",
    ]

    for pattern in boundary_patterns:

        match = re.search(
            pattern,
            answer,
            flags=re.IGNORECASE | re.DOTALL
        )

        if match:
            answer = answer[:match.start()].rstrip()

    return answer, len(generated_ids)


# ------------------------------------------------
# 5. RUN BOTH CONDITIONS
# ------------------------------------------------

results = []

for condition in ["WITHOUT_METADATA", "WITH_METADATA"]:

    # Same seed for each condition
    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)

    row = q008[
        q008["condition"].astype(str).str.upper()
        == condition
    ]

    prompt = row.iloc[0]["prompt"]

    print("\n")
    print("=" * 70)
    print(f"Q008 — {condition}")
    print("=" * 70)

    print("\nQUESTION:")
    print(
        "Also, has there been any progress on technology "
        "sharing for electrolyzer?"
    )

    print("\nGENERATING...")

    answer, generated_tokens = generate_llama_with_boundary_stop(
        prompt
    )

    print("\n" + "-" * 70)
    print("MODEL OUTPUT")
    print("-" * 70)

    print(answer)

    print("\n" + "-" * 70)
    print(f"Generated tokens before cleanup: {generated_tokens}")
    print("-" * 70)

    results.append({
        "question_id": "ADANIENT_Q008",
        "condition": condition,
        "generated_tokens": generated_tokens,
        "answer": answer
    })

    gc.collect()
    torch.cuda.empty_cache()


# ------------------------------------------------
# 6. SAVE
# ------------------------------------------------

results_df = pd.DataFrame(results)

OUTPUT_FILE = "/kaggle/working/llama_q008_boundary_stop.csv"

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n")
print("=" * 70)
print("PHASE 5.8 COMPLETE")
print("=" * 70)

print(f"\nSaved:")
print(OUTPUT_FILE)

print("\nSummary:")
print(
    results_df[
        [
            "question_id",
            "condition",
            "generated_tokens"
        ]
    ].to_string(index=False)
)
```

```python
# ================================================================
# PHASE 5.8B — GENERATION CONFIGURATION SWEEP
# ================================================================
#
# 4 questions × 2 retrieval conditions × 7 configurations
# = 56 generations
#
# Purpose:
# Select a generation configuration BEFORE the final 800 run.
#
# IMPORTANT:
# This is a development/sensitivity test.
# The winning configuration will then be FROZEN for Phase 5.9.
# ================================================================

import os
import re
import gc
import random
import numpy as np
import pandas as pd
import torch

from transformers import StoppingCriteria, StoppingCriteriaList


# ================================================================
# CONFIGURATIONS A-G
# ================================================================

CONFIGS = {
    "A": {
        "temperature": 0.2,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "B": {
        "temperature": 0.4,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "C": {
        "temperature": 0.5,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "D": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "E": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.20,
    },

    "F": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.25,
    },

    "G": {
        "temperature": 0.5,
        "top_p": 0.95,
        "repetition_penalty": 1.20,
    },
}


MAX_NEW_TOKENS = 1024
SEED = 42

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

OUTPUT_FILE = "/kaggle/working/llama_config_sweep_56.csv"

N_QUESTIONS = 4


# ================================================================
# LOAD PROMPTS
# ================================================================

df = pd.read_csv(PROMPT_FILE)

print("=" * 70)
print("PHASE 5.8B — CONFIGURATION SWEEP")
print("=" * 70)

print("\nTotal prompts:", len(df))

print("\nColumns:")
print(df.columns.tolist())


# ================================================================
# FIND 4 REPRESENTATIVE QUESTIONS
# ================================================================
#
# We select 4 questions from the WITH_METADATA set and then
# retrieve the corresponding WITHOUT_METADATA prompts.
#
# We avoid using only Q008 so the configuration isn't tuned
# specifically to the electrolyzer question.
# ================================================================

with_df = df[
    df["condition"] == "WITH_METADATA"
].copy()

# Add prompt length as a rough complexity indicator
with_df["question_length"] = (
    with_df["question"]
    .astype(str)
    .str.len()
)

with_df["question_words"] = (
    with_df["question"]
    .astype(str)
    .str.split()
    .str.len()
)

# ------------------------------------------------
# Choose four questions across different lengths
# ------------------------------------------------

with_df = with_df.sort_values(
    "question_words"
).reset_index(drop=True)

positions = np.linspace(
    0,
    len(with_df) - 1,
    N_QUESTIONS
).astype(int)

selected_question_ids = (
    with_df.iloc[positions]["question_id"]
    .tolist()
)

# Always include Q008 because we already extensively
# diagnosed its behavior and it is useful as a known reference.
if "ADANIENT_Q008" not in selected_question_ids:

    # Replace the closest-length question
    q008_row = with_df[
        with_df["question_id"] == "ADANIENT_Q008"
    ]

    if len(q008_row) > 0:

        selected_question_ids[-1] = "ADANIENT_Q008"


print("\nSelected questions:")

for qid in selected_question_ids:

    qrow = with_df[
        with_df["question_id"] == qid
    ].iloc[0]

    print(
        f"  {qid} | "
        f"{qrow['company']} | "
        f"{qrow['question_words']} words"
    )

    print(
        f"      {qrow['question']}"
    )


# ================================================================
# GET BOTH CONDITIONS
# ================================================================

test_df = df[
    df["question_id"].isin(
        selected_question_ids
    )
].copy()

# Make sure we have exactly:
# 4 questions × 2 conditions = 8 prompts

print("\nSelected prompts:", len(test_df))

print(
    test_df[
        ["question_id", "condition"]
    ].to_string(index=False)
)

assert len(test_df) == 8, (
    f"Expected 8 prompts, got {len(test_df)}"
)


# ================================================================
# BOUNDARY STOPPING
# ================================================================

class TranscriptBoundaryStop(StoppingCriteria):

    def __init__(self, tokenizer, prompt_length):

        self.tokenizer = tokenizer
        self.prompt_length = prompt_length

        self.patterns = [
            re.compile(
                r"\n\s*Question\s*:",
                re.IGNORECASE
            ),

            re.compile(
                r"\n\s*Context\s+\d+\s*:",
                re.IGNORECASE
            ),

            re.compile(
                r"\n\s*Moderator\s*:",
                re.IGNORECASE
            ),
        ]

    def __call__(
        self,
        input_ids,
        scores,
        **kwargs
    ):

        generated_ids = input_ids[
            0,
            self.prompt_length:
        ]

        if generated_ids.numel() == 0:
            return False

        generated_text = self.tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )

        recent_text = generated_text[-500:]

        for pattern in self.patterns:

            if pattern.search(recent_text):
                return True

        return False


# ================================================================
# CLEAN ANSWER
# ================================================================

def clean_answer(text):

    text = text.strip()

    patterns = [
        r"\n\s*Question\s*:",
        r"\n\s*Context\s+\d+\s*:",
        r"\n\s*Moderator\s*:",
    ]

    positions = []

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            re.IGNORECASE
        )

        if match:
            positions.append(
                match.start()
            )

    if positions:
        text = text[
            :min(positions)
        ]

    return text.strip()


# ================================================================
# GENERATION FUNCTION
# ================================================================

def generate_with_config(
    prompt,
    config
):

    # ------------------------------------------------
    # Reset seed BEFORE EACH GENERATION
    # ------------------------------------------------
    #
    # This makes configurations more comparable.
    # ------------------------------------------------

    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)

    # ------------------------------------------------
    # Tokenize
    # ------------------------------------------------

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs[
        "input_ids"
    ].to(model.device)

    attention_mask = inputs[
        "attention_mask"
    ].to(model.device)

    prompt_length = input_ids.shape[1]

    # ------------------------------------------------
    # Boundary stopper
    # ------------------------------------------------

    stopping = StoppingCriteriaList([
        TranscriptBoundaryStop(
            tokenizer=tokenizer,
            prompt_length=prompt_length
        )
    ])

    # ------------------------------------------------
    # Generate
    # ------------------------------------------------

    with torch.no_grad():

        outputs = model.generate(

            input_ids=input_ids,

            attention_mask=attention_mask,

            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=True,

            temperature=config[
                "temperature"
            ],

            top_p=config[
                "top_p"
            ],

            repetition_penalty=config[
                "repetition_penalty"
            ],

            stopping_criteria=stopping,

            eos_token_id=tokenizer.eos_token_id,

            pad_token_id=tokenizer.pad_token_id,

            use_cache=True,
        )

    # ------------------------------------------------
    # Extract generated tokens
    # ------------------------------------------------

    generated_ids = outputs[
        0,
        prompt_length:
    ]

    generated_tokens = len(
        generated_ids
    )

    # ------------------------------------------------
    # Decode
    # ------------------------------------------------

    raw_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    )

    cleaned_text = clean_answer(
        raw_text
    )

    return (
        cleaned_text,
        generated_tokens,
        raw_text
    )


# ================================================================
# SIMPLE AUTOMATIC DIAGNOSTICS
# ================================================================

def diagnostic_metrics(
    answer,
    question,
    expected_company
):

    answer_lower = answer.lower()

    # ------------------------------------------------
    # Boundary markers
    # ------------------------------------------------

    continuation_markers = [
        "question:",
        "context 5:",
        "context 6:",
        "context 7:",
        "context 8:",
        "moderator:",
    ]

    continuation = any(
        marker in answer_lower
        for marker in continuation_markers
    )

    # ------------------------------------------------
    # Repeated sentence detection
    # ------------------------------------------------

    sentences = [
        s.strip()
        for s in re.split(
            r"[.!?]+",
            answer
        )
        if s.strip()
    ]

    repeated_sentence = False

    if len(sentences) >= 3:

        counts = pd.Series(
            sentences
        ).value_counts()

        repeated_sentence = (
            counts.max() >= 3
        )

    # ------------------------------------------------
    # Very long generation
    # ------------------------------------------------

    word_count = len(
        answer.split()
    )

    long_generation = (
        word_count > 300
    )

    # ------------------------------------------------
    # Empty / degenerate answer
    # ------------------------------------------------

    empty = (
        len(answer.strip()) == 0
    )

    return {
        "word_count": word_count,
        "continuation_detected": continuation,
        "repeated_sentence": repeated_sentence,
        "long_generation_gt300": long_generation,
        "empty_answer": empty,
    }


# ================================================================
# RUN 56 GENERATIONS
# ================================================================

results = []

total_runs = (
    len(test_df)
    * len(CONFIGS)
)

run_number = 0


for _, row in test_df.iterrows():

    question_id = row["question_id"]
    condition = row["condition"]

    prompt = row["prompt"]

    question = row["question"]

    company = row.get(
        "company",
        ""
    )

    for config_name, config in CONFIGS.items():

        run_number += 1

        print("\n")
        print("=" * 70)

        print(
            f"RUN {run_number}/{total_runs}"
        )

        print(
            f"{question_id} | "
            f"{condition} | "
            f"CONFIG {config_name}"
        )

        print(
            f"T={config['temperature']} "
            f"TopP={config['top_p']} "
            f"RP={config['repetition_penalty']}"
        )

        print("=" * 70)

        try:

            answer, generated_tokens, raw_text = (
                generate_with_config(
                    prompt,
                    config
                )
            )

            metrics = diagnostic_metrics(
                answer,
                question,
                company
            )

            print("\nANSWER:")
            print("-" * 70)
            print(answer)
            print("-" * 70)

            print(
                f"Generated tokens: "
                f"{generated_tokens}"
            )

            print(
                f"Words: "
                f"{metrics['word_count']}"
            )

            print(
                f"Continuation detected: "
                f"{metrics['continuation_detected']}"
            )

            print(
                f"Repeated sentence: "
                f"{metrics['repeated_sentence']}"
            )

            results.append({

                "question_id":
                    question_id,

                "condition":
                    condition,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    config[
                        "temperature"
                    ],

                "top_p":
                    config[
                        "top_p"
                    ],

                "repetition_penalty":
                    config[
                        "repetition_penalty"
                    ],

                "question":
                    question,

                "answer":
                    answer,

                "raw_answer":
                    raw_text,

                "generated_tokens":
                    generated_tokens,

                "word_count":
                    metrics[
                        "word_count"
                    ],

                "continuation_detected":
                    metrics[
                        "continuation_detected"
                    ],

                "repeated_sentence":
                    metrics[
                        "repeated_sentence"
                    ],

                "long_generation_gt300":
                    metrics[
                        "long_generation_gt300"
                    ],

                "empty_answer":
                    metrics[
                        "empty_answer"
                    ],

                "status":
                    "success"
            })


        except Exception as e:

            print(
                "\nERROR:",
                type(e).__name__,
                str(e)
            )

            results.append({

                "question_id":
                    question_id,

                "condition":
                    condition,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    config[
                        "temperature"
                    ],

                "top_p":
                    config[
                        "top_p"
                    ],

                "repetition_penalty":
                    config[
                        "repetition_penalty"
                    ],

                "question":
                    question,

                "answer":
                    "",

                "raw_answer":
                    "",

                "generated_tokens":
                    0,

                "word_count":
                    0,

                "continuation_detected":
                    False,

                "repeated_sentence":
                    False,

                "long_generation_gt300":
                    False,

                "empty_answer":
                    True,

                "status":
                    "error",

                "error":
                    f"{type(e).__name__}: {e}"
            })


        # ------------------------------------------------
        # Save after EVERY generation
        # ------------------------------------------------

        pd.DataFrame(results).to_csv(
            OUTPUT_FILE,
            index=False
        )

        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()


# ================================================================
# FINAL SUMMARY
# ================================================================

results_df = pd.DataFrame(results)

print("\n")
print("=" * 70)
print("CONFIGURATION SWEEP COMPLETE")
print("=" * 70)

print(
    f"\nTotal generations: "
    f"{len(results_df)}"
)

print(
    "\nExpected generations:",
    total_runs
)

print("\nStatus:")
print(
    results_df[
        "status"
    ].value_counts()
)

print("\nSaved:")
print(OUTPUT_FILE)


# ================================================================
# CONFIGURATION SUMMARY
# ================================================================

successful = results_df[
    results_df["status"] == "success"
].copy()

if len(successful) > 0:

    summary = (
        successful
        .groupby("config")
        .agg(

            mean_tokens=(
                "generated_tokens",
                "mean"
            ),

            mean_words=(
                "word_count",
                "mean"
            ),

            continuation_rate=(
                "continuation_detected",
                "mean"
            ),

            repetition_rate=(
                "repeated_sentence",
                "mean"
            ),

            long_generation_rate=(
                "long_generation_gt300",
                "mean"
            ),

            empty_rate=(
                "empty_answer",
                "mean"
            ),
        )
        .reset_index()
    )

    # Convert rates to percentages

    for col in [
        "continuation_rate",
        "repetition_rate",
        "long_generation_rate",
        "empty_rate"
    ]:

        summary[col] *= 100

    print("\n")
    print("=" * 70)
    print("CONFIGURATION SUMMARY")
    print("=" * 70)

    print(
        summary.to_string(
            index=False
        )
    )

    summary.to_csv(
        "/kaggle/working/"
        "llama_config_sweep_summary.csv",
        index=False
    )

    print(
        "\nSummary saved:"
    )

    print(
        "/kaggle/working/"
        "llama_config_sweep_summary.csv"
    )
```

```python
# ================================================================
# PHASE 5.8C — FINAL SMALL CONFIGURATION SWEEP
# ================================================================
# 3 questions × 2 conditions × 4 configurations = 24 generations
#
# Configurations:
# A: T=0.2, top_p=0.90, RP=1.15
# B: T=0.4, top_p=0.90, RP=1.15
# C: T=0.5, top_p=0.90, RP=1.15
# D: T=0.7, top_p=0.90, RP=1.15
#
# Maximum output for sweep: 384 tokens
# ================================================================

import os
import re
import gc
import random
import numpy as np
import pandas as pd
import torch

from transformers import StoppingCriteria, StoppingCriteriaList


# ================================================================
# CONFIG
# ================================================================

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

OUTPUT_FILE = (
    "/kaggle/working/"
    "llama_config_sweep_24.csv"
)

MAX_NEW_TOKENS = 384

SEED = 42

CONFIGS = {

    "A": {
        "temperature": 0.2,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "B": {
        "temperature": 0.4,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "C": {
        "temperature": 0.5,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "D": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
}


# ================================================================
# EXACT 3 QUESTIONS
# ================================================================

SELECTED_QUESTIONS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
]


# ================================================================
# LOAD PROMPTS
# ================================================================

df = pd.read_csv(PROMPT_FILE)

print("=" * 70)
print("PHASE 5.8C — 24-RUN CONFIGURATION SWEEP")
print("=" * 70)

print("\nPrompt file:", PROMPT_FILE)
print("Total prompts:", len(df))

print("\nSelected questions:")

for qid in SELECTED_QUESTIONS:

    rows = df[
        df["question_id"] == qid
    ]

    if len(rows) == 0:
        raise ValueError(
            f"Question not found: {qid}"
        )

    print(
        f"\n{qid}"
    )

    print(
        rows[
            [
                "company",
                "question",
                "condition"
            ]
        ].drop_duplicates().to_string(
            index=False
        )
    )


# ================================================================
# SELECT BOTH CONDITIONS
# ================================================================

test_df = df[
    df["question_id"].isin(
        SELECTED_QUESTIONS
    )
].copy()

# Exactly 3 × 2 = 6 prompts
assert len(test_df) == 6, (
    f"Expected 6 prompts, found {len(test_df)}"
)

print("\nSelected prompts:")
print(
    test_df[
        [
            "question_id",
            "condition"
        ]
    ].sort_values(
        ["question_id", "condition"]
    ).to_string(index=False)
)


# ================================================================
# BOUNDARY STOPPING
# ================================================================

class TranscriptBoundaryStop(StoppingCriteria):

    def __init__(
        self,
        tokenizer,
        prompt_length
    ):

        self.tokenizer = tokenizer
        self.prompt_length = prompt_length

        self.patterns = [

            # Question markers
            re.compile(
                r"\n\s*Question\s*:?",
                re.IGNORECASE
            ),

            # Context markers
            re.compile(
                r"\n\s*Context\s+\d+\s*:?",
                re.IGNORECASE
            ),

            # Moderator
            re.compile(
                r"\n\s*Moderator\s*:?",
                re.IGNORECASE
            ),

            # Common transcript speaker pattern
            re.compile(
                r"\n\s*(Operator|Analyst|CFO|CEO|"
                r"Managing Director|Chief Executive Officer)"
                r"\s*:?",
                re.IGNORECASE
            ),
        ]


    def __call__(
        self,
        input_ids,
        scores,
        **kwargs
    ):

        generated_ids = input_ids[
            0,
            self.prompt_length:
        ]

        if generated_ids.numel() == 0:
            return False

        generated_text = self.tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )

        recent_text = generated_text[-500:]

        for pattern in self.patterns:

            if pattern.search(recent_text):
                return True

        return False


# ================================================================
# CLEAN ANSWER
# ================================================================

def clean_answer(text):

    text = text.strip()

    patterns = [

        r"\n\s*Question\s*:?",
        r"\n\s*Context\s+\d+\s*:?",
        r"\n\s*Moderator\s*:?",
    ]

    positions = []

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            re.IGNORECASE
        )

        if match:
            positions.append(
                match.start()
            )

    if positions:

        text = text[
            :min(positions)
        ]

    return text.strip()


# ================================================================
# GENERATION
# ================================================================

def generate_answer(
    prompt,
    config
):

    # Same seed for every configuration
    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)


    # ------------------------------------------------
    # TOKENIZE
    # ------------------------------------------------

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs[
        "input_ids"
    ].to(model.device)

    attention_mask = inputs[
        "attention_mask"
    ].to(model.device)

    prompt_length = input_ids.shape[1]


    # ------------------------------------------------
    # STOPPING
    # ------------------------------------------------

    stopping = StoppingCriteriaList([
        TranscriptBoundaryStop(
            tokenizer=tokenizer,
            prompt_length=prompt_length
        )
    ])


    # ------------------------------------------------
    # GENERATE
    # ------------------------------------------------

    with torch.no_grad():

        outputs = model.generate(

            input_ids=input_ids,

            attention_mask=attention_mask,

            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=True,

            temperature=config[
                "temperature"
            ],

            top_p=config[
                "top_p"
            ],

            repetition_penalty=config[
                "repetition_penalty"
            ],

            stopping_criteria=stopping,

            eos_token_id=tokenizer.eos_token_id,

            pad_token_id=tokenizer.pad_token_id,

            use_cache=True,
        )


    # ------------------------------------------------
    # GENERATED TOKENS
    # ------------------------------------------------

    generated_ids = outputs[
        0,
        prompt_length:
    ]

    generated_tokens = len(
        generated_ids
    )


    # ------------------------------------------------
    # DECODE
    # ------------------------------------------------

    raw_answer = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    )

    answer = clean_answer(
        raw_answer
    )


    return (
        answer,
        raw_answer,
        generated_tokens
    )


# ================================================================
# DIAGNOSTICS
# ================================================================

def get_diagnostics(answer):

    answer_lower = answer.lower()

    # ------------------------------------------------
    # Transcript continuation markers
    # ------------------------------------------------

    continuation_patterns = [

        r"\bquestion\s*:",

        r"\bcontext\s+\d+\s*:",

        r"\bmoderator\s*:?",

        r"\boperator\s*:?",

        r"\banalyst\s*:?",

        r"\bcfo\s*:?",

        r"\bceo\s*:?",

        r"\bmanaging director\s*:?",
    ]

    continuation = any(
        re.search(
            p,
            answer_lower
        )
        for p in continuation_patterns
    )


    # ------------------------------------------------
    # Repeated text
    # ------------------------------------------------

    words = answer_lower.split()

    repeated_5gram = False

    if len(words) >= 15:

        fivegrams = [
            " ".join(
                words[i:i+5]
            )
            for i in range(
                len(words) - 4
            )
        ]

        counts = pd.Series(
            fivegrams
        ).value_counts()

        if len(counts) > 0:

            repeated_5gram = (
                counts.iloc[0] >= 4
            )


    # ------------------------------------------------
    # Excessive length
    # ------------------------------------------------

    excessive_length = (
        len(words) > 250
    )


    return {
        "word_count": len(words),

        "continuation_detected":
            continuation,

        "repeated_5gram":
            repeated_5gram,

        "excessive_length":
            excessive_length,
    }


# ================================================================
# RUN 24 GENERATIONS
# ================================================================

results = []

total_runs = (
    len(test_df)
    * len(CONFIGS)
)

run_number = 0


for _, row in test_df.iterrows():

    question_id = row[
        "question_id"
    ]

    condition = row[
        "condition"
    ]

    question = row[
        "question"
    ]

    company = row.get(
        "company",
        ""
    )

    prompt = row[
        "prompt"
    ]


    for config_name, config in CONFIGS.items():

        run_number += 1

        print("\n")
        print("=" * 70)

        print(
            f"RUN {run_number}/{total_runs}"
        )

        print(
            f"{question_id} | "
            f"{condition} | "
            f"CONFIG {config_name}"
        )

        print(
            f"T={config['temperature']} | "
            f"TopP={config['top_p']} | "
            f"RP={config['repetition_penalty']} | "
            f"MaxTokens={MAX_NEW_TOKENS}"
        )

        print("=" * 70)


        try:

            answer, raw_answer, generated_tokens = (
                generate_answer(
                    prompt,
                    config
                )
            )

            diagnostics = get_diagnostics(
                answer
            )


            print("\nANSWER:")
            print("-" * 70)
            print(answer)
            print("-" * 70)

            print(
                "Generated tokens:",
                generated_tokens
            )

            print(
                "Words:",
                diagnostics[
                    "word_count"
                ]
            )

            print(
                "Continuation:",
                diagnostics[
                    "continuation_detected"
                ]
            )

            print(
                "Repeated 5-gram:",
                diagnostics[
                    "repeated_5gram"
                ]
            )


            results.append({

                "question_id":
                    question_id,

                "condition":
                    condition,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    config[
                        "temperature"
                    ],

                "top_p":
                    config[
                        "top_p"
                    ],

                "repetition_penalty":
                    config[
                        "repetition_penalty"
                    ],

                "question":
                    question,

                "answer":
                    answer,

                "raw_answer":
                    raw_answer,

                "generated_tokens":
                    generated_tokens,

                "word_count":
                    diagnostics[
                        "word_count"
                    ],

                "continuation_detected":
                    diagnostics[
                        "continuation_detected"
                    ],

                "repeated_5gram":
                    diagnostics[
                        "repeated_5gram"
                    ],

                "excessive_length":
                    diagnostics[
                        "excessive_length"
                    ],

                "status":
                    "success"
            })


        except Exception as e:

            print(
                "ERROR:",
                type(e).__name__,
                str(e)
            )

            results.append({

                "question_id":
                    question_id,

                "condition":
                    condition,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    config[
                        "temperature"
                    ],

                "top_p":
                    config[
                        "top_p"
                    ],

                "repetition_penalty":
                    config[
                        "repetition_penalty"
                    ],

                "question":
                    question,

                "answer":
                    "",

                "raw_answer":
                    "",

                "generated_tokens":
                    0,

                "word_count":
                    0,

                "continuation_detected":
                    False,

                "repeated_5gram":
                    False,

                "excessive_length":
                    False,

                "status":
                    "error",

                "error":
                    f"{type(e).__name__}: {e}"
            })


        # ------------------------------------------------
        # SAVE AFTER EVERY RUN
        # ------------------------------------------------

        pd.DataFrame(
            results
        ).to_csv(
            OUTPUT_FILE,
            index=False
        )


        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()


# ================================================================
# FINAL SUMMARY
# ================================================================

results_df = pd.DataFrame(
    results
)

print("\n")
print("=" * 70)
print("24-RUN SWEEP COMPLETE")
print("=" * 70)

print(
    "\nTotal runs:",
    len(results_df)
)

print(
    "\nStatus:"
)

print(
    results_df[
        "status"
    ].value_counts()
)

print(
    "\nSaved:"
)

print(
    OUTPUT_FILE
)


# ================================================================
# CONFIG SUMMARY
# ================================================================

successful = results_df[
    results_df["status"] == "success"
]

if len(successful) > 0:

    summary = (
        successful
        .groupby("config")
        .agg(

            mean_tokens=(
                "generated_tokens",
                "mean"
            ),

            mean_words=(
                "word_count",
                "mean"
            ),

            continuation_count=(
                "continuation_detected",
                "sum"
            ),

            repetition_count=(
                "repeated_5gram",
                "sum"
            ),

            excessive_length_count=(
                "excessive_length",
                "sum"
            ),
        )
        .reset_index()
    )

    print("\n")
    print("=" * 70)
    print("CONFIGURATION SUMMARY")
    print("=" * 70)

    print(
        summary.to_string(
            index=False
        )
    )

    summary.to_csv(
        "/kaggle/working/"
        "llama_config_sweep_24_summary.csv",
        index=False
    )
```

```python
# ============================================================
# CORRECTED 12-RUN CONFIGURATION VALIDATION
# NO ONLINE STOPPING CRITERION
# POST-PROCESS GENERATED TEXT ONLY
# ============================================================

import os
import re
import gc
import random
import numpy as np
import pandas as pd
import torch


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# 2. CONFIGURATIONS
# ============================================================

CONFIGS = {
    "A": {
        "temperature": 0.2,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "B": {
        "temperature": 0.4,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "C": {
        "temperature": 0.5,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    "D": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
}


# ============================================================
# 3. LOAD PROMPTS
# ============================================================

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

prompts_df = pd.read_csv(PROMPT_FILE)

print("Prompt dataframe shape:", prompts_df.shape)
print("Columns:", prompts_df.columns.tolist())


# ============================================================
# 4. SELECT 3 REPRESENTATIVE QUESTIONS
# ============================================================

TEST_QUESTIONS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
]

test_df = prompts_df[
    (prompts_df["question_id"].isin(TEST_QUESTIONS)) &
    (prompts_df["condition"] == "WITH_METADATA")
].copy()


print("\nSelected rows:")
print(
    test_df[
        ["question_id", "company", "condition"]
    ].to_string(index=False)
)


# ============================================================
# 5. VALIDATE SELECTION
# ============================================================

for qid in TEST_QUESTIONS:

    rows = test_df[
        test_df["question_id"] == qid
    ]

    if len(rows) != 1:
        raise ValueError(
            f"{qid}: expected exactly 1 row, "
            f"found {len(rows)}"
        )

print("\n✓ Correctly selected 3 WITH_METADATA prompts.")


# ============================================================
# 6. POST-PROCESSING ONLY
#
# IMPORTANT:
# We DO NOT stop generation based on the prompt.
# We only inspect the newly generated answer.
# ============================================================

CONTINUATION_PATTERNS = [

    # New question
    re.compile(
        r"\n\s*Question\s*:?",
        re.IGNORECASE
    ),

    # New context block
    re.compile(
        r"\n\s*Context\s+\d+\s*:?",
        re.IGNORECASE
    ),

    # Moderator continuation
    re.compile(
        r"\n\s*Moderator\s*:?",
        re.IGNORECASE
    ),
]


def clean_generated_answer(text):

    text = text.strip()

    earliest = None

    for pattern in CONTINUATION_PATTERNS:

        match = pattern.search(text)

        if match:

            if earliest is None:
                earliest = match.start()

            else:
                earliest = min(
                    earliest,
                    match.start()
                )

    if earliest is not None:

        cleaned = text[:earliest].strip()

        return cleaned, True

    return text, False


# ============================================================
# 7. GENERATION FUNCTION
# ============================================================

def generate_answer(prompt, config_name):

    cfg = CONFIGS[config_name]

    set_seed(SEED)

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    prompt_length = inputs["input_ids"].shape[1]

    with torch.no_grad():

        outputs = model.generate(

            **inputs,

            # DEVELOPMENT CAP ONLY
            max_new_tokens=384,

            do_sample=True,

            temperature=cfg["temperature"],

            top_p=cfg["top_p"],

            repetition_penalty=cfg["repetition_penalty"],

            pad_token_id=tokenizer.eos_token_id,

            eos_token_id=tokenizer.eos_token_id,
        )

    # --------------------------------------------------------
    # IMPORTANT:
    # Remove the original prompt tokens.
    # Only decode NEW tokens.
    # --------------------------------------------------------

    generated_ids = outputs[0][prompt_length:]

    raw_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    )

    raw_text = raw_text.strip()

    generated_tokens = len(generated_ids)

    # --------------------------------------------------------
    # Post-process continuation
    # --------------------------------------------------------

    cleaned_text, continuation_detected = (
        clean_generated_answer(raw_text)
    )

    return (
        raw_text,
        cleaned_text,
        generated_tokens,
        continuation_detected,
    )


# ============================================================
# 8. RUN 12 EXPERIMENTS
# ============================================================

results = []

total_runs = len(TEST_QUESTIONS) * len(CONFIGS)

run_number = 0


for qid in TEST_QUESTIONS:

    row = test_df[
        test_df["question_id"] == qid
    ].iloc[0]

    question = str(row["question"])
    company = str(row["company"])
    prompt = str(row["prompt"])


    print("\n")
    print("=" * 100)
    print(f"QUESTION: {qid}")
    print(f"COMPANY : {company}")
    print(f"QUESTION: {question}")
    print("=" * 100)


    for config_name in ["A", "B", "C", "D"]:

        run_number += 1

        cfg = CONFIGS[config_name]

        print("\n")
        print("-" * 100)

        print(
            f"[{run_number}/{total_runs}] "
            f"CONFIG {config_name}"
        )

        print(
            f"T={cfg['temperature']} | "
            f"TopP={cfg['top_p']} | "
            f"Rep={cfg['repetition_penalty']}"
        )

        print("-" * 100)


        try:

            (
                raw_answer,
                cleaned_answer,
                n_tokens,
                continuation_detected,
            ) = generate_answer(
                prompt,
                config_name
            )


            # ------------------------------------------------
            # Repetition diagnostic
            # ------------------------------------------------

            words = cleaned_answer.lower().split()

            if len(words) >= 10:

                unique_words = len(set(words))

                repetition_ratio = (
                    1 -
                    unique_words / len(words)
                )

            else:

                repetition_ratio = 0.0


            # ------------------------------------------------
            # Save result
            # ------------------------------------------------

            results.append({

                "question_id":
                    qid,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    cfg["temperature"],

                "top_p":
                    cfg["top_p"],

                "repetition_penalty":
                    cfg["repetition_penalty"],

                "generated_tokens":
                    n_tokens,

                "continuation_detected":
                    continuation_detected,

                "repetition_ratio":
                    repetition_ratio,

                "raw_answer":
                    raw_answer,

                "cleaned_answer":
                    cleaned_answer,
            })


            # ------------------------------------------------
            # Print
            # ------------------------------------------------

            print("\nRAW OUTPUT:")
            print(raw_answer)

            print("\nCLEANED OUTPUT:")
            print(cleaned_answer)

            print(
                "\nGenerated tokens:",
                n_tokens
            )

            print(
                "Continuation detected:",
                continuation_detected
            )

            print(
                "Repetition ratio:",
                round(repetition_ratio, 3)
            )


        except Exception as e:

            print("\nERROR:", repr(e))

            results.append({

                "question_id":
                    qid,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    cfg["temperature"],

                "top_p":
                    cfg["top_p"],

                "repetition_penalty":
                    cfg["repetition_penalty"],

                "generated_tokens":
                    -1,

                "continuation_detected":
                    True,

                "repetition_ratio":
                    1.0,

                "raw_answer":
                    f"ERROR: {repr(e)}",

                "cleaned_answer":
                    f"ERROR: {repr(e)}",
            })


        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()


# ============================================================
# 9. SAVE
# ============================================================

results_df = pd.DataFrame(results)

OUTPUT_FILE = (
    "/kaggle/working/"
    "llama_12run_config_validation_FIXED.csv"
)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n")
print("=" * 100)
print("SAVED:")
print(OUTPUT_FILE)
print("=" * 100)


# ============================================================
# 10. SUMMARY
# ============================================================

summary = (
    results_df
    .groupby("config")
    .agg(

        runs=("config", "size"),

        mean_tokens=(
            "generated_tokens",
            "mean"
        ),

        max_tokens=(
            "generated_tokens",
            "max"
        ),

        continuation_count=(
            "continuation_detected",
            "sum"
        ),

        mean_repetition=(
            "repetition_ratio",
            "mean"
        ),
    )
    .reset_index()
)


print("\nCONFIGURATION SUMMARY")
print("=" * 100)

print(
    summary.to_string(
        index=False
    )
)


# ============================================================
# 11. FINAL OUTPUTS SIDE-BY-SIDE
# ============================================================

for cfg in ["A", "B", "C", "D"]:

    print("\n")
    print("#" * 100)
    print(f"CONFIG {cfg}")
    print("#" * 100)

    cfg_df = results_df[
        results_df["config"] == cfg
    ]

    for _, r in cfg_df.iterrows():

        print("\n" + "-" * 90)

        print(
            f"{r['question_id']} | "
            f"{r['generated_tokens']} tokens"
        )

        print("-" * 90)

        print(
            r["cleaned_answer"]
        )
```

```python
# ============================================================
# LLAMA-2 BASE — FINAL FAST CONFIGURATION CHECK
#
# 3 questions × 4 configs = 12 runs
# WITH_METADATA only
# 128 generated tokens maximum
#
# IMPORTANT:
# NO stopping criteria
# NO answer post-processing
# We only compare raw generation behavior.
# ============================================================

import gc
import random
import numpy as np
import pandas as pd
import torch


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# 2. CONFIGURATIONS
# ============================================================

CONFIGS = {
    "A": {
        "temperature": 0.2,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
    "B": {
        "temperature": 0.4,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
    "C": {
        "temperature": 0.5,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
    "D": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
}


# ============================================================
# 3. LOAD PROMPTS
# ============================================================

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

prompts_df = pd.read_csv(PROMPT_FILE)

TEST_QUESTIONS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
]

test_df = prompts_df[
    prompts_df["question_id"].isin(TEST_QUESTIONS)
    & (prompts_df["condition"] == "WITH_METADATA")
].copy()


# ============================================================
# 4. VERIFY DATA
# ============================================================

print("Selected questions:")
print(
    test_df[
        ["question_id", "company", "condition"]
    ].to_string(index=False)
)

assert len(test_df) == 3, (
    f"Expected 3 prompts, found {len(test_df)}"
)

assert test_df["question_id"].nunique() == 3

print("\n✓ 3 WITH_METADATA prompts loaded.")


# ============================================================
# 5. GENERATION FUNCTION
# ============================================================

def generate_answer(prompt, config_name):

    cfg = CONFIGS[config_name]

    # Same seed before EVERY run so differences are
    # attributable to the generation configuration.
    set_seed(SEED)

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    prompt_tokens = inputs["input_ids"].shape[1]

    with torch.inference_mode():

        outputs = model.generate(

            **inputs,

            # SHORT DEVELOPMENT TEST
            max_new_tokens=128,

            do_sample=True,

            temperature=cfg["temperature"],
            top_p=cfg["top_p"],
            repetition_penalty=cfg["repetition_penalty"],

            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )

    # Only decode newly generated tokens.
    generated_ids = outputs[0][prompt_tokens:]

    answer = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()

    return answer, len(generated_ids)


# ============================================================
# 6. RUN 12 EXPERIMENTS
# ============================================================

results = []

run_number = 0
total_runs = 12


for _, row in test_df.iterrows():

    qid = row["question_id"]
    company = row["company"]
    question = row["question"]
    prompt = row["prompt"]

    print("\n")
    print("=" * 100)
    print(f"QUESTION: {qid}")
    print(f"COMPANY : {company}")
    print(f"QUESTION: {question}")
    print("=" * 100)


    for config_name in ["A", "B", "C", "D"]:

        run_number += 1

        cfg = CONFIGS[config_name]

        print("\n" + "-" * 100)

        print(
            f"[{run_number}/{total_runs}] "
            f"CONFIG {config_name}"
        )

        print(
            f"temperature={cfg['temperature']} | "
            f"top_p={cfg['top_p']} | "
            f"repetition_penalty={cfg['repetition_penalty']}"
        )

        print("-" * 100)


        try:

            answer, n_tokens = generate_answer(
                prompt,
                config_name
            )

            results.append({
                "question_id": qid,
                "company": company,
                "config": config_name,
                "temperature": cfg["temperature"],
                "top_p": cfg["top_p"],
                "repetition_penalty": cfg["repetition_penalty"],
                "generated_tokens": n_tokens,
                "answer": answer,
            })

            print("\nOUTPUT:")
            print(answer)

            print(
                f"\nGenerated tokens: {n_tokens}"
            )


        except Exception as e:

            print("\nERROR:", repr(e))

            results.append({
                "question_id": qid,
                "company": company,
                "config": config_name,
                "temperature": cfg["temperature"],
                "top_p": cfg["top_p"],
                "repetition_penalty": cfg["repetition_penalty"],
                "generated_tokens": -1,
                "answer": f"ERROR: {repr(e)}",
            })


        # Free memory between runs
        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()


# ============================================================
# 7. SAVE RESULTS
# ============================================================

results_df = pd.DataFrame(results)

OUTPUT_FILE = (
    "/kaggle/working/"
    "llama_config_test_12x128.csv"
)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# 8. SUMMARY
# ============================================================

summary = (
    results_df
    .groupby("config")
    .agg(
        runs=("config", "size"),
        mean_tokens=("generated_tokens", "mean"),
        max_tokens=("generated_tokens", "max"),
    )
    .reset_index()
)

print("\n")
print("=" * 100)
print("CONFIGURATION SUMMARY")
print("=" * 100)

print(summary.to_string(index=False))

print("\nSaved to:")
print(OUTPUT_FILE)


# ============================================================
# 9. PRINT ALL RESULTS AGAIN IN COMPACT FORM
# ============================================================

print("\n")
print("=" * 100)
print("ALL OUTPUTS")
print("=" * 100)

for _, r in results_df.iterrows():

    print(
        f"\n[{r['config']}] "
        f"{r['question_id']} "
        f"({r['generated_tokens']} tokens)"
    )

    print(r["answer"])
```

```python
# ============================================================
# FINAL LLAMA-2 GENERATION-CONFIG TEST
#
# 4 QUESTIONS × 3 CONFIGS = 12 RUNS
#
# CONFIGS:
#   A = T 0.2 / Top-p 0.90 / Rep 1.15
#   C = T 0.5 / Top-p 0.90 / Rep 1.15
#   D = T 0.7 / Top-p 0.90 / Rep 1.15
#
# WITH_METADATA ONLY
# max_new_tokens = 256
#
# IMPORTANT:
# Stopping criterion examines ONLY newly generated tokens.
# It NEVER examines the original prompt.
# ============================================================

import gc
import random
import re
import numpy as np
import pandas as pd
import torch

from transformers import StoppingCriteria, StoppingCriteriaList


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

def set_seed(seed=42):

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# 2. THREE FINAL CONFIGURATIONS
# ============================================================

CONFIGS = {

    # Low-temperature / stable
    "A": {
        "temperature": 0.2,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    # Middle
    "C": {
        "temperature": 0.5,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },

    # Higher sampling
    "D": {
        "temperature": 0.7,
        "top_p": 0.90,
        "repetition_penalty": 1.15,
    },
}


# ============================================================
# 3. LOAD PROMPTS
# ============================================================

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

prompts_df = pd.read_csv(PROMPT_FILE)

print("Prompt dataframe:", prompts_df.shape)
print("Columns:", prompts_df.columns.tolist())


# ============================================================
# 4. SELECT 4 QUESTIONS
#
# Three are fixed because they have already been used in
# previous diagnostics.
#
# The fourth is selected reproducibly from the remaining
# questions using seed 42.
# ============================================================

FIXED_QUESTIONS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
]


# Get all WITH_METADATA questions
available_questions = (
    prompts_df[
        prompts_df["condition"] == "WITH_METADATA"
    ]["question_id"]
    .drop_duplicates()
    .tolist()
)


# Remaining questions
remaining_questions = [
    q for q in available_questions
    if q not in FIXED_QUESTIONS
]


# Reproducibly select one additional question
rng = np.random.default_rng(SEED)

extra_question = rng.choice(
    remaining_questions
)

TEST_QUESTIONS = (
    FIXED_QUESTIONS
    + [extra_question]
)


print("\nSelected 4 questions:")
print(TEST_QUESTIONS)


# ============================================================
# 5. BUILD TEST DATAFRAME
# ============================================================

test_df = prompts_df[
    prompts_df["question_id"].isin(TEST_QUESTIONS)
    & (prompts_df["condition"] == "WITH_METADATA")
].copy()


# Preserve desired order
test_df["question_order"] = (
    test_df["question_id"]
    .map({
        q: i
        for i, q in enumerate(TEST_QUESTIONS)
    })
)

test_df = (
    test_df
    .sort_values("question_order")
    .drop(columns="question_order")
    .reset_index(drop=True)
)


# ============================================================
# 6. VALIDATION
# ============================================================

assert len(test_df) == 4, (
    f"Expected 4 rows, found {len(test_df)}"
)

assert (
    test_df["question_id"].nunique() == 4
)

print("\nSelected prompts:")
print(
    test_df[
        [
            "question_id",
            "company",
            "condition"
        ]
    ].to_string(index=False)
)


# ============================================================
# 7. CORRECT GENERATION-ONLY STOPPER
# ============================================================

class GeneratedTextStopper(StoppingCriteria):

    def __init__(
        self,
        tokenizer,
        prompt_length
    ):

        super().__init__()

        self.tokenizer = tokenizer
        self.prompt_length = prompt_length

        # ONLY markers that indicate transcript continuation.
        #
        # DO NOT include CFO / CEO / Analyst / etc.
        # Those can legitimately occur in an answer.

        self.patterns = [

            re.compile(
                r"\n\s*Question\s*:?",
                re.IGNORECASE
            ),

            re.compile(
                r"\n\s*Context\s+\d+\s*:?",
                re.IGNORECASE
            ),

            re.compile(
                r"\n\s*Moderator\s*:?",
                re.IGNORECASE
            ),
        ]


    def __call__(
        self,
        input_ids,
        scores,
        **kwargs
    ):

        # ====================================================
        # CRITICAL FIX
        #
        # Ignore the original prompt completely.
        #
        # Only inspect tokens generated AFTER prompt_length.
        # ====================================================

        generated_ids = input_ids[
            0,
            self.prompt_length:
        ]

        if generated_ids.numel() == 0:
            return False


        generated_text = self.tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )


        # Avoid stopping on tiny partial generations.
        if len(generated_text.strip()) < 10:
            return False


        for pattern in self.patterns:

            if pattern.search(
                generated_text
            ):
                return True


        return False


# ============================================================
# 8. CLEAN ONLY A DETECTED CONTINUATION
# ============================================================

def clean_answer(text):

    text = text.strip()

    cut_positions = []

    patterns = [

        re.compile(
            r"\n\s*Question\s*:?",
            re.IGNORECASE
        ),

        re.compile(
            r"\n\s*Context\s+\d+\s*:?",
            re.IGNORECASE
        ),

        re.compile(
            r"\n\s*Moderator\s*:?",
            re.IGNORECASE
        ),
    ]


    for pattern in patterns:

        match = pattern.search(text)

        if match:
            cut_positions.append(
                match.start()
            )


    if cut_positions:

        text = text[
            :min(cut_positions)
        ].strip()


    return text


# ============================================================
# 9. GENERATION FUNCTION
# ============================================================

def generate_answer(
    prompt,
    config_name
):

    cfg = CONFIGS[config_name]

    set_seed(SEED)


    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)


    prompt_length = (
        inputs["input_ids"].shape[1]
    )


    stopper = GeneratedTextStopper(
        tokenizer=tokenizer,
        prompt_length=prompt_length
    )


    stopping_criteria = StoppingCriteriaList([
        stopper
    ])


    with torch.inference_mode():

        outputs = model.generate(

            **inputs,

            # Development/configuration-test ceiling.
            # NOT the paper's final 1024 setting.
            max_new_tokens=256,

            do_sample=True,

            temperature=cfg["temperature"],

            top_p=cfg["top_p"],

            repetition_penalty=
                cfg["repetition_penalty"],

            stopping_criteria=
                stopping_criteria,

            pad_token_id=
                tokenizer.eos_token_id,

            eos_token_id=
                tokenizer.eos_token_id,
        )


    # --------------------------------------------------------
    # ONLY NEWLY GENERATED TOKENS
    # --------------------------------------------------------

    generated_ids = outputs[
        0,
        prompt_length:
    ]


    raw_answer = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()


    cleaned_answer = clean_answer(
        raw_answer
    )


    generated_tokens = len(
        generated_ids
    )


    return (
        raw_answer,
        cleaned_answer,
        generated_tokens
    )


# ============================================================
# 10. RUN 12 EXPERIMENTS
# ============================================================

results = []

run_number = 0
total_runs = 12


for _, row in test_df.iterrows():

    qid = row["question_id"]
    company = row["company"]
    question = row["question"]
    prompt = row["prompt"]


    print("\n\n")
    print("=" * 100)
    print(f"QUESTION: {qid}")
    print(f"COMPANY : {company}")
    print(f"QUESTION: {question}")
    print("=" * 100)


    for config_name in [
        "A",
        "C",
        "D"
    ]:

        run_number += 1

        cfg = CONFIGS[
            config_name
        ]


        print("\n")
        print("-" * 100)

        print(
            f"[{run_number}/{total_runs}] "
            f"CONFIG {config_name}"
        )

        print(
            f"T={cfg['temperature']} | "
            f"TopP={cfg['top_p']} | "
            f"Rep={cfg['repetition_penalty']}"
        )

        print("-" * 100)


        try:

            (
                raw_answer,
                cleaned_answer,
                n_tokens
            ) = generate_answer(
                prompt,
                config_name
            )


            results.append({

                "question_id":
                    qid,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    cfg["temperature"],

                "top_p":
                    cfg["top_p"],

                "repetition_penalty":
                    cfg["repetition_penalty"],

                "generated_tokens":
                    n_tokens,

                "raw_answer":
                    raw_answer,

                "cleaned_answer":
                    cleaned_answer,
            })


            print("\nRAW OUTPUT:")
            print(raw_answer)

            print(
                "\nGenerated tokens:",
                n_tokens
            )

            print("\nCLEANED ANSWER:")
            print(cleaned_answer)


        except Exception as e:

            print(
                "\nERROR:",
                repr(e)
            )

            results.append({

                "question_id":
                    qid,

                "company":
                    company,

                "config":
                    config_name,

                "temperature":
                    cfg["temperature"],

                "top_p":
                    cfg["top_p"],

                "repetition_penalty":
                    cfg["repetition_penalty"],

                "generated_tokens":
                    -1,

                "raw_answer":
                    f"ERROR: {repr(e)}",

                "cleaned_answer":
                    f"ERROR: {repr(e)}",
            })


        # Free memory
        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()


# ============================================================
# 11. SAVE RESULTS
# ============================================================

results_df = pd.DataFrame(
    results
)

OUTPUT_FILE = (
    "/kaggle/working/"
    "llama_FINAL_12run_config_test.csv"
)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# 12. SUMMARY
# ============================================================

summary = (
    results_df
    .groupby("config")
    .agg(

        runs=(
            "config",
            "size"
        ),

        mean_tokens=(
            "generated_tokens",
            "mean"
        ),

        median_tokens=(
            "generated_tokens",
            "median"
        ),

        max_tokens=(
            "generated_tokens",
            "max"
        ),

    )
    .reset_index()
)


print("\n\n")
print("=" * 100)
print("FINAL CONFIGURATION TEST SUMMARY")
print("=" * 100)

print(
    summary.to_string(
        index=False
    )
)


print("\nSaved:")
print(OUTPUT_FILE)


# ============================================================
# 13. COMPACT OUTPUT TABLE
# ============================================================

print("\n")
print("=" * 100)
print("RESULTS")
print("=" * 100)

for _, r in results_df.iterrows():

    print(
        f"\n{r['config']} | "
        f"{r['question_id']} | "
        f"{r['generated_tokens']} tokens"
    )

    print(
        r["cleaned_answer"]
    )
```

```python
# ============================================================
# FINAL PROMPT VALIDATION — 5 QUESTIONS
#
# Generation config is FROZEN:
#   temperature       = 0.2
#   top_p             = 0.90
#   repetition_penalty = 1.15
#
# Prompt is the ONLY thing being changed.
#
# WITH_METADATA only
# 256 max_new_tokens for validation
# ============================================================

import gc
import random
import re
import numpy as np
import pandas as pd
import torch


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# ============================================================
# 2. FROZEN GENERATION CONFIG
# ============================================================

TEMPERATURE = 0.2
TOP_P = 0.90
REPETITION_PENALTY = 1.15

MAX_NEW_TOKENS = 256


# ============================================================
# 3. OPTIMIZED PROMPT
# ============================================================

PROMPT_TEMPLATE = """The following are excerpts from an earnings call transcript, followed by a question about the call. Answer the question using only information stated in the excerpts. Begin with the direct answer (for a yes/no question, begin with Yes or No), then give the relevant details from the excerpts, combining related facts from different excerpts and keeping names, numbers and dates as written. Do not add anything the excerpts do not state. If the excerpts do not answer the question, write: Not stated in the excerpts. Write the answer as one short paragraph of neutral, third-person prose.

### Transcript excerpts

{contexts}

### Question

{question}

### Answer (one short paragraph, based only on the excerpts above)
"""


# ============================================================
# 4. LOAD ORIGINAL PROMPTS
# ============================================================

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"

df = pd.read_csv(PROMPT_FILE)

print("Original dataframe:", df.shape)
print("Columns:", df.columns.tolist())


# ============================================================
# 5. SELECT 5 QUESTIONS
#
# Three previously used:
#   ADANIENT_Q008
#   ADANIPORTS_Q011
#   DRREDDY_Q019
#
# Two additional questions are selected reproducibly.
# ============================================================

FIXED_QUESTIONS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
]


available_questions = (
    df[
        df["condition"] == "WITH_METADATA"
    ]["question_id"]
    .drop_duplicates()
    .tolist()
)


remaining_questions = [
    q for q in available_questions
    if q not in FIXED_QUESTIONS
]


rng = np.random.default_rng(SEED)

extra_questions = rng.choice(
    remaining_questions,
    size=2,
    replace=False
).tolist()


TEST_QUESTIONS = (
    FIXED_QUESTIONS
    + extra_questions
)


print("\n5 validation questions:")
for q in TEST_QUESTIONS:
    print(" ", q)


# ============================================================
# 6. GET RETRIEVED CHUNKS FROM EXISTING PROMPT CSV
#
# The existing CSV contains the 4 retrieved chunks INSIDE
# the original prompt.
#
# We extract them from the existing prompt rather than
# changing retrieval itself.
# ============================================================

test_df = df[
    df["question_id"].isin(TEST_QUESTIONS)
    & (df["condition"] == "WITH_METADATA")
].copy()


# One prompt per question
test_df = (
    test_df
    .drop_duplicates(
        subset=["question_id"]
    )
    .copy()
)


# Preserve chosen order
order_map = {
    q: i
    for i, q in enumerate(TEST_QUESTIONS)
}

test_df["order"] = (
    test_df["question_id"]
    .map(order_map)
)

test_df = (
    test_df
    .sort_values("order")
    .drop(columns="order")
    .reset_index(drop=True)
)


if len(test_df) != 5:
    raise ValueError(
        f"Expected 5 questions, found {len(test_df)}"
    )


# ============================================================
# 7. EXTRACT THE FOUR CONTEXT CHUNKS
# ============================================================

def extract_contexts_from_old_prompt(prompt):

    """
    Existing prompt structure is approximately:

    ...
    {context text}

    Question:
    {question}

    Answer:

    We extract everything between the context region
    and the Question: marker.

    Then split the four retrieved chunks using the
    existing Context N: markers.
    """

    prompt = str(prompt)

    # Find question boundary
    question_match = re.search(
        r"\n\s*Question\s*:",
        prompt,
        flags=re.IGNORECASE
    )

    if question_match is None:
        raise ValueError(
            "Could not find 'Question:' in existing prompt."
        )


    context_region = prompt[
        :question_match.start()
    ]


    # Find where actual context begins.
    # Existing prompts use Context 1:, Context 2:, etc.
    first_context = re.search(
        r"Context\s+1\s*:",
        context_region,
        flags=re.IGNORECASE
    )

    if first_context is None:

        # Fallback:
        # use everything before Question:
        context_text = context_region.strip()

        return [context_text]


    context_region = context_region[
        first_context.start():
    ]


    # Split at Context 1:, Context 2:, ...
    parts = re.split(
        r"\n\s*Context\s+\d+\s*:\s*",
        context_region,
        flags=re.IGNORECASE
    )


    chunks = [
        p.strip()
        for p in parts
        if p.strip()
    ]


    return chunks


# ============================================================
# 8. BUILD NEW PROMPTS
# ============================================================

new_prompts = []

for _, row in test_df.iterrows():

    old_prompt = row["prompt"]

    chunks = extract_contexts_from_old_prompt(
        old_prompt
    )

    # We expect exactly 4 retrieved chunks.
    # If the extraction gives something different,
    # print it rather than silently changing it.

    print("\n")
    print("=" * 90)
    print(
        row["question_id"],
        "|",
        row["company"]
    )
    print(
        "Extracted chunks:",
        len(chunks)
    )


    contexts = "\n\n---\n\n".join(
        chunk.strip()
        for chunk in chunks
    )


    new_prompt = PROMPT_TEMPLATE.format(
        contexts=contexts,
        question=str(row["question"])
    )


    new_prompts.append({
        "question_id":
            row["question_id"],

        "company":
            row["company"],

        "question":
            row["question"],

        "n_contexts":
            len(chunks),

        "prompt":
            new_prompt,
    })


validation_df = pd.DataFrame(
    new_prompts
)


# ============================================================
# 9. PROMPT TOKEN VALIDATION
# ============================================================

print("\n")
print("=" * 90)
print("PROMPT TOKEN VALIDATION")
print("=" * 90)


token_counts = []

for _, row in validation_df.iterrows():

    encoded = tokenizer(
        row["prompt"],
        add_special_tokens=True
    )

    n = len(
        encoded["input_ids"]
    )

    token_counts.append(n)

    print(
        f"{row['question_id']}: "
        f"{n} tokens"
    )


validation_df[
    "prompt_tokens"
] = token_counts


print(
    "\nMean:",
    validation_df["prompt_tokens"].mean()
)

print(
    "Max:",
    validation_df["prompt_tokens"].max()
)


# ============================================================
# 10. GENERATE
# ============================================================

def generate(prompt):

    # Reset seed before each question.
    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)


    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)


    prompt_length = (
        inputs["input_ids"].shape[1]
    )


    with torch.inference_mode():

        outputs = model.generate(

            **inputs,

            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=True,

            temperature=TEMPERATURE,

            top_p=TOP_P,

            repetition_penalty=
                REPETITION_PENALTY,

            pad_token_id=
                tokenizer.eos_token_id,

            eos_token_id=
                tokenizer.eos_token_id,
        )


    generated_ids = outputs[
        0,
        prompt_length:
    ]


    raw = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()


    return raw, len(generated_ids)


# ============================================================
# 11. POST-PROCESS ONLY OBVIOUS CONTINUATION
# ============================================================

def clean_answer(raw):

    text = raw.strip()

    # IMPORTANT:
    # Do NOT cut on '\n\n'.
    # A legitimate answer can contain multiple paragraphs.
    #
    # Only remove obvious structural continuation.

    patterns = [

        r"\n\s*Context\s+\d+\s*:?",
        r"\n\s*Question\s*:?",
        r"\n\s*Moderator\s*:?",
    ]


    positions = []

    for pattern in patterns:

        m = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if m:
            positions.append(
                m.start()
            )


    if positions:

        text = text[
            :min(positions)
        ].strip()


    return text


# ============================================================
# 12. RUN 5 QUESTIONS
# ============================================================

results = []


for _, row in validation_df.iterrows():

    print("\n")
    print("=" * 100)
    print(
        f"{row['question_id']} | "
        f"{row['company']}"
    )

    print(
        "QUESTION:",
        row["question"]
    )

    print("=" * 100)


    raw, n_tokens = generate(
        row["prompt"]
    )


    cleaned = clean_answer(
        raw
    )


    results.append({

        "question_id":
            row["question_id"],

        "company":
            row["company"],

        "question":
            row["question"],

        "prompt_tokens":
            row["prompt_tokens"],

        "generated_tokens":
            n_tokens,

        "raw_output":
            raw,

        "cleaned_answer":
            cleaned,
    })


    print("\nRAW OUTPUT:")
    print(raw)

    print(
        "\nGenerated tokens:",
        n_tokens
    )

    print("\nCLEANED ANSWER:")
    print(cleaned)


    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ============================================================
# 13. SAVE VALIDATION
# ============================================================

results_df = pd.DataFrame(
    results
)


OUTPUT_FILE = (
    "/kaggle/working/"
    "optimized_prompt_5question_validation.csv"
)


results_df.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n")
print("=" * 100)
print("VALIDATION COMPLETE")
print("=" * 100)

print(
    results_df[
        [
            "question_id",
            "prompt_tokens",
            "generated_tokens"
        ]
    ].to_string(index=False)
)

print("\nSaved:")
print(OUTPUT_FILE)
```

```python
# ============================================================
# FINAL PROMPT + RETRIEVAL CHECK — 5 QUESTION VALIDATION
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import torch

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------
MODEL_ID = "meta-llama/Llama-2-7b-hf"

TEMPERATURE = 0.2
TOP_P = 0.90
REPETITION_PENALTY = 1.15
MAX_NEW_TOKENS = 256

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"
RETRIEVAL_FILE = "/kaggle/working/with_metadata_retrieval.csv"

OUTPUT_FILE = "/kaggle/working/final_prompt_5question_validation.csv"
CHUNKS_FILE = "/kaggle/working/final_prompt_5question_retrieved_chunks.csv"


# ------------------------------------------------------------
# QUESTIONS
# ------------------------------------------------------------
TEST_IDS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
    "AXISBANK_Q005",
    "SUNPHARMA_Q007",
]


# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------
prompts_df = pd.read_csv(PROMPT_FILE)
retrieval_df = pd.read_csv(RETRIEVAL_FILE)

print("Prompts:", prompts_df.shape)
print("Retrieval:", retrieval_df.shape)

print("\nRetrieval columns:")
print(retrieval_df.columns.tolist())


# ------------------------------------------------------------
# BUILD CONTEXT DIRECTLY FROM RETRIEVAL CSV
# IMPORTANT:
# Do NOT parse the old prompt.
# ------------------------------------------------------------
def get_contexts(question_id):
    rows = retrieval_df[
        retrieval_df["question_id"].astype(str) == str(question_id)
    ].copy()

    rows = rows.sort_values("rank")

    if len(rows) != 4:
        raise ValueError(
            f"{question_id}: expected 4 retrieved chunks, found {len(rows)}"
        )

    contexts = rows["text"].astype(str).tolist()

    return rows, contexts


# ------------------------------------------------------------
# FINAL SIMPLIFIED PROMPT
# ------------------------------------------------------------
def build_final_prompt(question, contexts):

    context_text = "\n\n---\n\n".join(contexts)

    prompt = f"""The following are excerpts from an earnings call transcript.

Answer the question using only information contained in the excerpts.
Give the direct answer first, followed by only the relevant supporting details.
Keep names, numbers, dates, and facts exactly as stated in the excerpts.
Do not introduce information that is not present in the excerpts.
Output only the answer.

### Transcript excerpts

{context_text}

### Question

{question}

### Answer
"""

    return prompt


# ------------------------------------------------------------
# TOKENIZER / MODEL
# Assumes tokenizer + model are already loaded.
# ------------------------------------------------------------
print("\nChecking model/tokenizer...")

assert tokenizer is not None
assert model is not None

device = next(model.parameters()).device

print("Device:", device)
print("Model:", MODEL_ID)


# ------------------------------------------------------------
# GENERATION
# ------------------------------------------------------------
def generate_answer(prompt):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs["input_ids"].to(device)
    attention_mask = inputs["attention_mask"].to(device)

    prompt_len = input_ids.shape[1]

    with torch.no_grad():
        outputs = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,

            # IMPORTANT:
            # Do NOT pass max_length.
            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=True,
            temperature=TEMPERATURE,
            top_p=TOP_P,
            repetition_penalty=REPETITION_PENALTY,

            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )

    generated_ids = outputs[0][prompt_len:]

    raw_answer = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()

    return raw_answer, len(generated_ids)


# ------------------------------------------------------------
# LIGHT POST-PROCESSING
# Only remove obvious transcript continuation.
# DO NOT cut on blank lines.
# ------------------------------------------------------------
def clean_answer(text):

    text = text.strip()

    # Obvious continuation markers
    patterns = [
        r"\n\s*###\s*(?:Transcript excerpts|Question|Answer|Explanation)\b",
        r"\n\s*Question\s*:",
        r"\n\s*Context\s+\d+\s*:",
        r"\n\s*Moderator\s*:",
    ]

    cut_positions = []

    for pattern in patterns:
        m = re.search(pattern, text, flags=re.IGNORECASE)
        if m:
            cut_positions.append(m.start())

    if cut_positions:
        text = text[:min(cut_positions)].strip()

    return text


# ------------------------------------------------------------
# RUN 5 QUESTIONS
# ------------------------------------------------------------
results = []
chunk_records = []

print("\n")
print("=" * 110)
print("5-QUESTION FINAL PROMPT VALIDATION")
print("=" * 110)

for qid in TEST_IDS:

    # ----------------------------------------
    # Get question
    # ----------------------------------------
    qrows = prompts_df[
        prompts_df["question_id"].astype(str) == str(qid)
    ]

    if len(qrows) == 0:
        print(f"\nERROR: {qid} not found in prompt dataframe")
        continue

    qrow = qrows.iloc[0]

    question = str(qrow["question"])
    company = str(qrow["company"])

    # ----------------------------------------
    # Get exact 4 retrieved chunks
    # ----------------------------------------
    rrows, contexts = get_contexts(qid)

    # ----------------------------------------
    # Build prompt
    # ----------------------------------------
    prompt = build_final_prompt(
        question=question,
        contexts=contexts
    )

    # ----------------------------------------
    # Token validation
    # ----------------------------------------
    prompt_tokens = len(
        tokenizer(
            prompt,
            add_special_tokens=True
        )["input_ids"]
    )

    # ----------------------------------------
    # Print retrieval evidence
    # ----------------------------------------
    print("\n")
    print("=" * 110)
    print(f"{qid} | {company}")
    print("=" * 110)

    print("\nQUESTION:")
    print(question)

    print("\nRETRIEVED CHUNKS:")
    print("-" * 110)

    for rank, context in enumerate(contexts, start=1):

        row = rrows.iloc[rank - 1]

        similarity = row.get("similarity_score", np.nan)
        retrieved_company = row.get("retrieved_company", "")

        print(f"\n--- CHUNK {rank} ---")
        print(
            f"Company: {retrieved_company} | "
            f"Similarity: {similarity}"
        )
        print(context)

        chunk_records.append({
            "question_id": qid,
            "company": company,
            "question": question,
            "rank": rank,
            "retrieved_company": retrieved_company,
            "similarity_score": similarity,
            "text": context,
        })

    # ----------------------------------------
    # Prompt token count
    # ----------------------------------------
    print("\nPROMPT TOKENS:", prompt_tokens)

    if prompt_tokens + MAX_NEW_TOKENS > 4096:
        print(
            "WARNING: prompt + generation exceeds Llama context window!"
        )
    else:
        print("Context window check: OK")

    # ----------------------------------------
    # Generate
    # ----------------------------------------
    print("\nGENERATING...")

    raw_answer, generated_tokens = generate_answer(prompt)

    cleaned_answer = clean_answer(raw_answer)

    # ----------------------------------------
    # Display result
    # ----------------------------------------
    print("\nRAW OUTPUT:")
    print(raw_answer)

    print("\nCLEANED ANSWER:")
    print(cleaned_answer)

    print("\nGenerated tokens:", generated_tokens)

    # ----------------------------------------
    # Save result
    # ----------------------------------------
    results.append({
        "question_id": qid,
        "company": company,
        "question": question,
        "prompt_tokens": prompt_tokens,
        "generated_tokens": generated_tokens,
        "raw_answer": raw_answer,
        "cleaned_answer": cleaned_answer,
        "n_contexts": len(contexts),
    })


# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------
results_df = pd.DataFrame(results)
chunks_df = pd.DataFrame(chunk_records)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

chunks_df.to_csv(
    CHUNKS_FILE,
    index=False
)


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------
print("\n")
print("=" * 110)
print("VALIDATION SUMMARY")
print("=" * 110)

print(
    results_df[
        [
            "question_id",
            "prompt_tokens",
            "generated_tokens",
            "cleaned_answer"
        ]
    ].to_string(index=False)
)

print("\nSaved:")
print(OUTPUT_FILE)
print(CHUNKS_FILE)
```

```python
# ============================================================
# FINAL GENERATION TEST — 5 QUESTIONS
# WITH GENERATION-ONLY TRANSCRIPT BOUNDARY STOPPING
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import torch

from transformers import StoppingCriteria, StoppingCriteriaList


# ============================================================
# CONFIG
# ============================================================

TEMPERATURE = 0.2
TOP_P = 0.90
REPETITION_PENALTY = 1.15

MAX_NEW_TOKENS = 256

PROMPT_FILE = "/kaggle/working/llama_prompts_800.csv"
RETRIEVAL_FILE = "/kaggle/working/with_metadata_retrieval.csv"

OUTPUT_FILE = "/kaggle/working/final_boundary_stop_5question.csv"
CHUNKS_FILE = "/kaggle/working/final_boundary_stop_5question_chunks.csv"


TEST_IDS = [
    "ADANIENT_Q008",
    "ADANIPORTS_Q011",
    "DRREDDY_Q019",
    "AXISBANK_Q005",
    "SUNPHARMA_Q007",
]


# ============================================================
# LOAD DATA
# ============================================================

prompts_df = pd.read_csv(PROMPT_FILE)
retrieval_df = pd.read_csv(RETRIEVAL_FILE)

print("Prompts:", prompts_df.shape)
print("Retrieval:", retrieval_df.shape)

assert "question_id" in prompts_df.columns
assert "question_id" in retrieval_df.columns
assert "text" in retrieval_df.columns


# ============================================================
# BUILD CONTEXT DIRECTLY FROM RETRIEVAL DATA
# ============================================================

def get_contexts(question_id):

    rows = retrieval_df[
        retrieval_df["question_id"].astype(str) == str(question_id)
    ].copy()

    rows = rows.sort_values("rank").reset_index(drop=True)

    if len(rows) != 4:
        raise ValueError(
            f"{question_id}: expected exactly 4 retrieved chunks, "
            f"found {len(rows)}"
        )

    contexts = rows["text"].astype(str).tolist()

    return rows, contexts


# ============================================================
# FINAL PROMPT
# ============================================================

def build_prompt(question, contexts):

    context_text = "\n\n---\n\n".join(contexts)

    return f"""The following are excerpts from an earnings call transcript.

Answer the question using only information contained in the excerpts.
Give the direct answer first, followed by only the relevant supporting details.
Keep names, numbers, dates, and facts exactly as stated in the excerpts.
Do not introduce information that is not present in the excerpts.
Output only the answer.

### Transcript excerpts

{context_text}

### Question

{question}

### Answer
"""


# ============================================================
# GENERATION-ONLY BOUNDARY DETECTOR
#
# IMPORTANT:
# We ONLY inspect newly generated tokens.
# We never inspect the prompt.
# ============================================================

BOUNDARY_PATTERNS = [

    # Explicit QA markers
    r"\n\s*###\s*Question\b",
    r"\n\s*###\s*Answer\b",

    # Transcript question markers
    r"\n\s*Question\s*:",
    r"\n\s*Q\s*:",

    # Operators / moderator
    r"\n\s*Operator\b",
    r"\n\s*Moderator\b",

    # Common transcript analyst roles
    r"\n\s*Research Analyst\b",
    r"\n\s*Equity Research Analyst\b",
    r"\n\s*Senior Equity Research Analyst\b",
    r"\n\s*Analyst\b",

    # New speaker-type labels
    r"\n\s*CFO\b",
    r"\n\s*CEO\b",
    r"\n\s*Managing Director\b",
    r"\n\s*Chief Executive Officer\b",
]


class TranscriptBoundaryStoppingCriteria(StoppingCriteria):

    def __init__(self, tokenizer, prompt_length):
        self.tokenizer = tokenizer
        self.prompt_length = prompt_length
        self.triggered = False
        self.trigger_text = None

    def __call__(self, input_ids, scores, **kwargs):

        # Only generated tokens.
        generated_ids = input_ids[0, self.prompt_length:]

        if generated_ids.numel() == 0:
            return False

        generated_text = self.tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )

        # Check transcript-boundary patterns.
        for pattern in BOUNDARY_PATTERNS:

            match = re.search(
                pattern,
                generated_text,
                flags=re.IGNORECASE
            )

            if match is not None:
                self.triggered = True
                self.trigger_text = match.group(0)
                return True

        return False


# ============================================================
# CLEAN OUTPUT
# ============================================================

def clean_answer(text):

    text = text.strip()

    patterns = [
        r"\n\s*###\s*Question\b",
        r"\n\s*###\s*Answer\b",
        r"\n\s*Question\s*:",
        r"\n\s*Q\s*:",
        r"\n\s*Operator\b",
        r"\n\s*Moderator\b",
        r"\n\s*Research Analyst\b",
        r"\n\s*Equity Research Analyst\b",
        r"\n\s*Senior Equity Research Analyst\b",
        r"\n\s*Analyst\b",
    ]

    cut_positions = []

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if match:
            cut_positions.append(match.start())

    if cut_positions:
        text = text[:min(cut_positions)].strip()

    return text


# ============================================================
# GENERATION FUNCTION
# ============================================================

def generate_answer(prompt):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs["input_ids"].to(model.device)
    attention_mask = inputs["attention_mask"].to(model.device)

    prompt_length = input_ids.shape[1]

    stopper = TranscriptBoundaryStoppingCriteria(
        tokenizer=tokenizer,
        prompt_length=prompt_length
    )

    stopping = StoppingCriteriaList([stopper])

    with torch.no_grad():

        outputs = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,

            # IMPORTANT:
            # max_length intentionally omitted.
            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=True,
            temperature=TEMPERATURE,
            top_p=TOP_P,
            repetition_penalty=REPETITION_PENALTY,

            stopping_criteria=stopping,

            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )

    generated_ids = outputs[0, prompt_length:]

    raw_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()

    cleaned_text = clean_answer(raw_text)

    return (
        raw_text,
        cleaned_text,
        len(generated_ids),
        stopper.triggered,
        stopper.trigger_text
    )


# ============================================================
# RUN VALIDATION
# ============================================================

results = []
chunk_records = []

print("\n")
print("=" * 110)
print("5-QUESTION BOUNDARY-STOP VALIDATION")
print("=" * 110)


for qid in TEST_IDS:

    # --------------------------------------------------------
    # Question
    # --------------------------------------------------------

    qrows = prompts_df[
        prompts_df["question_id"].astype(str) == str(qid)
    ]

    if len(qrows) == 0:
        raise ValueError(f"{qid} not found in prompts dataframe")

    qrow = qrows.iloc[0]

    question = str(qrow["question"])
    company = str(qrow["company"])


    # --------------------------------------------------------
    # Retrieved chunks
    # --------------------------------------------------------

    rrows, contexts = get_contexts(qid)


    # --------------------------------------------------------
    # Prompt
    # --------------------------------------------------------

    prompt = build_prompt(
        question=question,
        contexts=contexts
    )


    # --------------------------------------------------------
    # Prompt tokens
    # --------------------------------------------------------

    prompt_tokens = len(
        tokenizer(
            prompt,
            add_special_tokens=True
        )["input_ids"]
    )


    # --------------------------------------------------------
    # Display
    # --------------------------------------------------------

    print("\n")
    print("=" * 110)
    print(f"{qid} | {company}")
    print("=" * 110)

    print("\nQUESTION:")
    print(question)

    print("\nRETRIEVED CHUNKS:")
    print("-" * 110)

    for rank, context in enumerate(contexts, start=1):

        row = rrows.iloc[rank - 1]

        similarity = row.get(
            "similarity_score",
            np.nan
        )

        retrieved_company = row.get(
            "retrieved_company",
            ""
        )

        print(f"\n--- CHUNK {rank} ---")
        print(
            f"Company: {retrieved_company} | "
            f"Similarity: {similarity}"
        )
        print(context)

        chunk_records.append({
            "question_id": qid,
            "company": company,
            "question": question,
            "rank": rank,
            "retrieved_company": retrieved_company,
            "similarity_score": similarity,
            "text": context,
        })


    # --------------------------------------------------------
    # Context window check
    # --------------------------------------------------------

    print("\nPROMPT TOKENS:", prompt_tokens)

    if prompt_tokens + MAX_NEW_TOKENS > 4096:
        print(
            "WARNING: prompt + requested generation exceeds "
            "4096-token Llama context window!"
        )
    else:
        print("Context window check: OK")


    # --------------------------------------------------------
    # Generate
    # --------------------------------------------------------

    print("\nGENERATING...")

    (
        raw_answer,
        cleaned_answer,
        generated_tokens,
        stopped,
        trigger
    ) = generate_answer(prompt)


    # --------------------------------------------------------
    # Output
    # --------------------------------------------------------

    print("\nRAW OUTPUT:")
    print(raw_answer)

    print("\nCLEANED ANSWER:")
    print(cleaned_answer)

    print("\nGenerated tokens:", generated_tokens)
    print("Boundary stop triggered:", stopped)
    print("Boundary trigger:", trigger)


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    results.append({
        "question_id": qid,
        "company": company,
        "question": question,
        "prompt_tokens": prompt_tokens,
        "generated_tokens": generated_tokens,
        "boundary_stop": stopped,
        "boundary_trigger": trigger,
        "raw_answer": raw_answer,
        "cleaned_answer": cleaned_answer,
        "n_contexts": len(contexts),
    })


# ============================================================
# SAVE
# ============================================================

results_df = pd.DataFrame(results)
chunks_df = pd.DataFrame(chunk_records)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

chunks_df.to_csv(
    CHUNKS_FILE,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n")
print("=" * 110)
print("FINAL SUMMARY")
print("=" * 110)

print(
    results_df[
        [
            "question_id",
            "prompt_tokens",
            "generated_tokens",
            "boundary_stop",
            "boundary_trigger"
        ]
    ].to_string(index=False)
)

print("\n")
print("OUTPUT FILE:")
print(OUTPUT_FILE)

print("\nCHUNKS FILE:")
print(CHUNKS_FILE)
```

```python
# ============================================================
# FINAL 800 GENERATION — EFFICIENT VERSION
#
# 400 WITHOUT_METADATA
# 400 WITH_METADATA
#
# Llama-2-7B BASE
# Early stopping when transcript continuation begins
# ============================================================

import os
import re
import hashlib
import numpy as np
import pandas as pd
import torch

from transformers import StoppingCriteria, StoppingCriteriaList


# ============================================================
# CONFIG
# ============================================================

MODEL_ID = "meta-llama/Llama-2-7b-hf"

TEMPERATURE = 0.2
TOP_P = 0.90
REPETITION_PENALTY = 1.15

# HARD SAFETY CAP
MAX_NEW_TOKENS = 256

GROUND_TRUTH_FILE = (
    "/kaggle/input/datasets/jasapologize/"
    "financial-replication-data/ground_truth_400.csv"
)

WITHOUT_RETRIEVAL_FILE = (
    "/kaggle/working/without_metadata_retrieval.csv"
)

WITH_RETRIEVAL_FILE = (
    "/kaggle/working/with_metadata_retrieval.csv"
)

OUTPUT_FILE = (
    "/kaggle/working/llama_final_800_generation.csv"
)

TEMP_OUTPUT_FILE = (
    "/kaggle/working/llama_final_800_generation.tmp.csv"
)


# ============================================================
# MODEL
# ============================================================

assert tokenizer is not None
assert model is not None

model.eval()

device = next(model.parameters()).device

print("=" * 100)
print("FINAL 800 GENERATION")
print("=" * 100)

print("Model:", MODEL_ID)
print("Device:", device)
print("Temperature:", TEMPERATURE)
print("Top-p:", TOP_P)
print("Repetition penalty:", REPETITION_PENALTY)
print("Max new tokens:", MAX_NEW_TOKENS)


# ============================================================
# LOAD DATA
# ============================================================

gt_df = pd.read_csv(GROUND_TRUTH_FILE)

without_df = pd.read_csv(
    WITHOUT_RETRIEVAL_FILE
)

with_df = pd.read_csv(
    WITH_RETRIEVAL_FILE
)


# ============================================================
# QUESTION TABLE
# ============================================================

question_df = (
    gt_df[
        [
            "question_id",
            "company",
            "ticker",
            "transcript",
            "question",
            "ground_truth_answer",
        ]
    ]
    .drop_duplicates("question_id")
    .reset_index(drop=True)
)

assert len(question_df) == 400


# ============================================================
# BUILD RETRIEVAL LOOKUPS
# ============================================================

def build_lookup(df):

    lookup = {}

    for qid, group in df.groupby("question_id"):

        group = group.copy()

        group["rank_num"] = pd.to_numeric(
            group["rank"],
            errors="coerce"
        )

        group = group.sort_values("rank_num")

        contexts = (
            group["text"]
            .astype(str)
            .tolist()
        )

        if len(contexts) != 4:
            raise ValueError(
                f"{qid}: expected 4 contexts, got "
                f"{len(contexts)}"
            )

        lookup[str(qid)] = contexts

    return lookup


without_lookup = build_lookup(without_df)
with_lookup = build_lookup(with_df)


# ============================================================
# FINAL PROMPT
# ============================================================

def build_prompt(question, contexts):

    context_text = "\n\n---\n\n".join(contexts)

    return f"""The following are excerpts from an earnings call transcript.

Answer the question using only information contained in the excerpts.
Give the direct answer first, followed by only the relevant supporting details.
Keep names, numbers, dates, and facts exactly as stated in the excerpts.
Do not introduce information that is not present in the excerpts.
Output only the answer.

### Transcript excerpts

{context_text}

### Question

{question}

### Answer
"""


# ============================================================
# FAST GENERATION-ONLY STOPPING
#
# IMPORTANT:
# We DO NOT stop at "### Answer".
#
# We ONLY stop when the model begins another question /
# transcript continuation.
# ============================================================

class AnswerBoundaryStoppingCriteria(StoppingCriteria):

    def __init__(
        self,
        tokenizer,
        prompt_length
    ):

        self.tokenizer = tokenizer
        self.prompt_length = prompt_length

        self.triggered = False
        self.trigger = ""

    def __call__(
        self,
        input_ids,
        scores,
        **kwargs
    ):

        generated_ids = (
            input_ids[0, self.prompt_length:]
        )

        if generated_ids.numel() == 0:
            return False

        # Only decode the generated continuation.
        generated_text = self.tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        )

        # We ONLY stop on an actual NEXT transcript/question
        # boundary.
        boundary_patterns = [

            r"\n\s*###\s*Question\b",

            r"\n\s*Question\s*:",

            r"\n\s*Operator\b",

            r"\n\s*Moderator\b",
        ]

        for pattern in boundary_patterns:

            match = re.search(
                pattern,
                generated_text,
                flags=re.IGNORECASE
            )

            if match:

                self.triggered = True
                self.trigger = match.group(0)

                return True

        return False


# ============================================================
# ANSWER EXTRACTION
#
# Only applied AFTER generation.
# Raw output remains untouched.
# ============================================================

def extract_answer(text):

    text = text.strip()

    boundary_patterns = [

        r"\n\s*###\s*Question\b",

        r"\n\s*Question\s*:",

        r"\n\s*Operator\b",

        r"\n\s*Moderator\b",
    ]

    positions = []

    for pattern in boundary_patterns:

        match = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if match:

            positions.append(
                match.start()
            )

    if positions:

        text = text[
            :min(positions)
        ].strip()

    return text


# ============================================================
# STABLE SEED
# ============================================================

def make_seed(
    question_id,
    condition
):

    key = (
        f"{question_id}|"
        f"{condition}|42"
    )

    digest = hashlib.sha256(
        key.encode()
    ).hexdigest()

    return int(
        digest[:8],
        16
    )


# ============================================================
# GENERATE ONE QUESTION
# ============================================================

def generate_one(
    prompt,
    seed
):

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed_all(
            seed
        )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs[
        "input_ids"
    ].to(device)

    attention_mask = inputs[
        "attention_mask"
    ].to(device)

    prompt_tokens = (
        input_ids.shape[1]
    )

    # Llama-2 context limit
    if (
        prompt_tokens +
        MAX_NEW_TOKENS
        > 4096
    ):

        raise ValueError(
            f"Prompt too long: "
            f"{prompt_tokens} + "
            f"{MAX_NEW_TOKENS} > 4096"
        )

    stopper = (
        AnswerBoundaryStoppingCriteria(
            tokenizer=tokenizer,
            prompt_length=prompt_tokens
        )
    )

    stopping_criteria = (
        StoppingCriteriaList(
            [stopper]
        )
    )

    with torch.no_grad():

        outputs = model.generate(

            input_ids=input_ids,

            attention_mask=attention_mask,

            # IMPORTANT:
            # No max_length.
            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=True,

            temperature=TEMPERATURE,

            top_p=TOP_P,

            repetition_penalty=(
                REPETITION_PENALTY
            ),

            stopping_criteria=(
                stopping_criteria
            ),

            pad_token_id=(
                tokenizer.eos_token_id
            ),

            eos_token_id=(
                tokenizer.eos_token_id
            ),
        )

    generated_ids = (
        outputs[
            0,
            prompt_tokens:
        ]
    )

    raw_output = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()

    answer_output = extract_answer(
        raw_output
    )

    generated_tokens = (
        len(generated_ids)
    )

    # Free temporary tensors
    del inputs
    del input_ids
    del attention_mask
    del outputs
    del generated_ids

    return {
        "prompt_tokens": prompt_tokens,
        "generated_tokens": generated_tokens,
        "raw_output": raw_output,
        "answer_output": answer_output,
        "boundary_stop": stopper.triggered,
        "boundary_trigger": stopper.trigger,
    }


# ============================================================
# RESUME
# ============================================================

if os.path.exists(OUTPUT_FILE):

    old_df = pd.read_csv(
        OUTPUT_FILE
    )

    completed = set()

    for _, row in old_df.iterrows():

        if str(row["status"]) == "OK":

            completed.add(
                (
                    str(row["question_id"]),
                    str(row["condition"])
                )
            )

    all_results = (
        old_df.to_dict("records")
    )

    print(
        "\nExisting output found:"
    )

    print(
        "Completed:",
        len(completed)
    )

else:

    all_results = []

    completed = set()

    print(
        "\nStarting new run."
    )


# ============================================================
# BUILD EXACTLY 800 JOBS
# ============================================================

jobs = []

for _, row in question_df.iterrows():

    qid = str(
        row["question_id"]
    )

    jobs.append({

        "question_id": qid,

        "condition":
            "WITHOUT_METADATA",

        "company":
            str(row["company"]),

        "ticker":
            str(row["ticker"]),

        "question":
            str(row["question"]),

        "ground_truth_answer":
            str(row["ground_truth_answer"]),

        "contexts":
            without_lookup[qid],
    })

    jobs.append({

        "question_id": qid,

        "condition":
            "WITH_METADATA",

        "company":
            str(row["company"]),

        "ticker":
            str(row["ticker"]),

        "question":
            str(row["question"]),

        "ground_truth_answer":
            str(row["ground_truth_answer"]),

        "contexts":
            with_lookup[qid],
    })


assert len(jobs) == 800


# ============================================================
# GENERATION LOOP
# ============================================================

for job_number, job in enumerate(
    jobs,
    start=1
):

    key = (
        job["question_id"],
        job["condition"]
    )

    # Resume support
    if key in completed:
        continue

    print("\n")
    print("=" * 100)

    print(
        f"[{job_number}/800] "
        f"{job['condition']} | "
        f"{job['question_id']} | "
        f"{job['company']}"
    )

    print("=" * 100)

    try:

        prompt = build_prompt(
            job["question"],
            job["contexts"]
        )

        seed = make_seed(
            job["question_id"],
            job["condition"]
        )

        result = generate_one(
            prompt,
            seed
        )

        status = "OK"
        error = ""

        print(
            "\nPrompt tokens:",
            result["prompt_tokens"]
        )

        print(
            "Generated tokens:",
            result["generated_tokens"]
        )

        print(
            "Boundary stop:",
            result["boundary_stop"]
        )

        print(
            "Boundary:",
            repr(
                result["boundary_trigger"]
            )
        )

        print(
            "\nANSWER OUTPUT:"
        )

        print(
            result["answer_output"]
        )

    except Exception as e:

        result = {

            "prompt_tokens": np.nan,

            "generated_tokens": np.nan,

            "raw_output": "",

            "answer_output": "",

            "boundary_stop": False,

            "boundary_trigger": "",
        }

        status = "ERROR"

        error = repr(e)

        print(
            "\nERROR:",
            error
        )


    # ========================================================
    # SAVE AFTER EVERY QUESTION
    # ========================================================

    all_results.append({

        "question_id":
            job["question_id"],

        "condition":
            job["condition"],

        "company":
            job["company"],

        "ticker":
            job["ticker"],

        "question":
            job["question"],

        "ground_truth_answer":
            job["ground_truth_answer"],

        "prompt_tokens":
            result["prompt_tokens"],

        "generated_tokens":
            result["generated_tokens"],

        "temperature":
            TEMPERATURE,

        "top_p":
            TOP_P,

        "repetition_penalty":
            REPETITION_PENALTY,

        "max_new_tokens":
            MAX_NEW_TOKENS,

        "seed":
            make_seed(
                job["question_id"],
                job["condition"]
            ),

        "raw_output":
            result["raw_output"],

        "answer_output":
            result["answer_output"],

        "boundary_stop":
            result["boundary_stop"],

        "boundary_trigger":
            result["boundary_trigger"],

        "status":
            status,

        "error":
            error,
    })


    save_df = pd.DataFrame(
        all_results
    )

    save_df.to_csv(
        TEMP_OUTPUT_FILE,
        index=False
    )

    os.replace(
        TEMP_OUTPUT_FILE,
        OUTPUT_FILE
    )

    if status == "OK":

        completed.add(key)

    print(
        f"\nSaved progress: "
        f"{len(completed)}/800"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

final_df = pd.read_csv(
    OUTPUT_FILE
)

print("\n")
print("=" * 100)
print("GENERATION FINISHED")
print("=" * 100)

print(
    "Rows:",
    len(final_df)
)

print(
    "\nStatus:"
)

print(
    final_df["status"]
    .value_counts()
)

print(
    "\nConditions:"
)

print(
    final_df.groupby(
        ["condition", "status"]
    ).size()
)

print(
    "\nGenerated token statistics:"
)

print(
    final_df[
        final_df["status"] == "OK"
    ]
    .groupby(
        "condition"
    )["generated_tokens"]
    .agg(
        [
            "count",
            "mean",
            "median",
            "min",
            "max"
        ]
    )
)

print(
    "\nSaved:",
    OUTPUT_FILE
)
```

```python
# ============================================================
# FINAL EVALUATION — RUN THIS AFTER THE 800 GENERATIONS FINISH
#
# Produces:
#   BERTScore
#   BARTScore
#   Jaro Similarity
#   LCSubsequence Similarity
#   LCSubsequence Word Count
#
# Final Table-4 style result:
#   Llama2-7b | Without Metadata
#   Llama2-7b | With Metadata
# ============================================================

import os
import re
import gc
import math
import numpy as np
import pandas as pd
import torch

from tqdm.auto import tqdm


# ============================================================
# CONFIG
# ============================================================

GENERATION_FILE = (
    "/kaggle/working/llama_final_800_generation.csv"
)

QUESTION_METRICS_FILE = (
    "/kaggle/working/llama2_7b_question_metrics.csv"
)

TABLE4_FILE = (
    "/kaggle/working/llama2_7b_final_table4.csv"
)

BERT_BATCH_SIZE = 8
BART_BATCH_SIZE = 4

BERT_MODEL = "roberta-large"
BART_MODEL = "facebook/bart-large-cnn"


# ============================================================
# WAIT / VALIDATE FINAL GENERATION FILE
# ============================================================

if not os.path.exists(GENERATION_FILE):
    raise FileNotFoundError(
        f"Generation file not found:\n{GENERATION_FILE}"
    )

gen_df = pd.read_csv(GENERATION_FILE)

print("=" * 110)
print("FINAL EVALUATION")
print("=" * 110)

print("Generation file shape:", gen_df.shape)

if "status" not in gen_df.columns:
    raise ValueError(
        "Generation file does not contain 'status'."
    )

successful = gen_df[
    gen_df["status"].astype(str).str.upper() == "OK"
].copy()

print("Successful rows:", len(successful))

if len(successful) != 800:
    raise ValueError(
        f"Expected 800 successful generations, "
        f"but found {len(successful)}.\n"
        f"Finish the generation run before evaluation."
    )

# Exactly 400 per condition
counts = (
    successful["condition"]
    .value_counts()
    .to_dict()
)

print("\nCondition counts:", counts)

if counts.get("WITHOUT_METADATA", 0) != 400:
    raise ValueError(
        "Expected 400 WITHOUT_METADATA rows."
    )

if counts.get("WITH_METADATA", 0) != 400:
    raise ValueError(
        "Expected 400 WITH_METADATA rows."
    )


# ============================================================
# IMPORTANT:
# EVALUATE THE ANSWER PORTION, NOT THE FULL TRANSCRIPT
#
# raw_output = untouched model output
# answer_output = deterministic answer extraction
#
# We preserve raw_output for audit, but the metric comparison
# uses answer_output.
# ============================================================

if "answer_output" not in successful.columns:
    raise ValueError(
        "Missing answer_output column."
    )

if "ground_truth_answer" not in successful.columns:
    raise ValueError(
        "Missing ground_truth_answer column."
    )

successful["prediction"] = (
    successful["answer_output"]
    .fillna("")
    .astype(str)
    .str.strip()
)

successful["reference"] = (
    successful["ground_truth_answer"]
    .fillna("")
    .astype(str)
    .str.strip()
)

print(
    "\nEmpty predictions:",
    (
        successful["prediction"].str.len() == 0
    ).sum()
)


# ============================================================
# FREE LLAMA GPU MEMORY
# ============================================================
# Generation is finished, so we do not need Llama anymore.
# This prevents BERT/BART evaluation from competing with it.
# ============================================================

print("\nReleasing Llama GPU memory...")

try:
    del model
except:
    pass

try:
    del tokenizer
except:
    pass

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

print("GPU memory released.")


# ============================================================
# 1. BERTSCORE
# ============================================================

print("\n")
print("=" * 110)
print("1. BERTScore")
print("=" * 110)

try:
    from bert_score import score as bert_score

except ImportError:
    raise ImportError(
        "bert-score is not installed.\n"
        "Run:\n"
        "!pip install -q bert-score"
    )

bert_device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

predictions_all = (
    successful["prediction"].tolist()
)

references_all = (
    successful["reference"].tolist()
)

print("Model:", BERT_MODEL)
print("Device:", bert_device)

# IMPORTANT:
# We use F1, without baseline rescaling.
P, R, F1 = bert_score(
    predictions_all,
    references_all,
    model_type=BERT_MODEL,
    lang="en",
    batch_size=BERT_BATCH_SIZE,
    device=bert_device,
    rescale_with_baseline=False,
    verbose=True,
)

successful["BERTScore"] = (
    F1.detach()
    .cpu()
    .numpy()
)

print(
    "Mean BERTScore:",
    successful["BERTScore"].mean()
)


# ============================================================
# FREE BERTSCORE MODEL
# ============================================================

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()


# ============================================================
# 2. BARTSCORE
# ============================================================

print("\n")
print("=" * 110)
print("2. BARTScore")
print("=" * 110)

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
)

import torch.nn.functional as F


bart_device = (
    torch.device("cuda")
    if torch.cuda.is_available()
    else torch.device("cpu")
)

print("Model:", BART_MODEL)
print("Device:", bart_device)


# ------------------------------------------------------------
# Load BART
# ------------------------------------------------------------

bart_tokenizer = AutoTokenizer.from_pretrained(
    BART_MODEL
)

bart_model = AutoModelForSeq2SeqLM.from_pretrained(
    BART_MODEL
)

bart_model = bart_model.to(
    bart_device
)

bart_model.eval()


# ------------------------------------------------------------
# BART log-likelihood scorer
#
# Average log probability per target token.
# ------------------------------------------------------------

def bart_directional_score(
    sources,
    targets,
    batch_size=4,
    max_length=1024
):

    scores = []

    for start in tqdm(
        range(
            0,
            len(sources),
            batch_size
        ),
        desc="BARTScore"
    ):

        src_batch = sources[
            start:start + batch_size
        ]

        tgt_batch = targets[
            start:start + batch_size
        ]

        src = bart_tokenizer(
            src_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt",
        )

        tgt = bart_tokenizer(
            tgt_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt",
        )

        src_input_ids = (
            src["input_ids"]
            .to(bart_device)
        )

        src_attention_mask = (
            src["attention_mask"]
            .to(bart_device)
        )

        tgt_input_ids = (
            tgt["input_ids"]
            .to(bart_device)
        )

        with torch.no_grad():

            outputs = bart_model(
                input_ids=src_input_ids,
                attention_mask=src_attention_mask,
                labels=tgt_input_ids,
            )

        logits = outputs.logits

        # Token-level negative log likelihood
        vocab_size = logits.shape[-1]

        token_losses = F.cross_entropy(
            logits.reshape(-1, vocab_size),
            tgt_input_ids.reshape(-1),
            ignore_index=bart_tokenizer.pad_token_id,
            reduction="none",
        )

        token_losses = token_losses.reshape(
            tgt_input_ids.shape
        )

        mask = (
            tgt_input_ids
            != bart_tokenizer.pad_token_id
        )

        token_count = (
            mask.sum(dim=1)
            .clamp(min=1)
        )

        # Mean LOG probability per target token
        batch_scores = (
            -token_losses.sum(dim=1)
            / token_count
        )

        scores.extend(
            batch_scores
            .detach()
            .cpu()
            .numpy()
            .tolist()
        )

        del src
        del tgt
        del src_input_ids
        del src_attention_mask
        del tgt_input_ids
        del outputs
        del logits
        del token_losses
        del token_count
        del batch_scores

    return np.asarray(
        scores,
        dtype=np.float64
    )


# ------------------------------------------------------------
# BARTScore directions
#
# P = reference -> hypothesis
# R = hypothesis -> reference
# Final = (P + R) / 2
# ------------------------------------------------------------

refs = references_all
hyps = predictions_all

bart_P = bart_directional_score(
    refs,
    hyps,
    batch_size=BART_BATCH_SIZE,
    max_length=1024,
)

bart_R = bart_directional_score(
    hyps,
    refs,
    batch_size=BART_BATCH_SIZE,
    max_length=1024,
)

bart_F = (
    bart_P + bart_R
) / 2.0

successful["BARTScore"] = bart_F

print(
    "Mean BARTScore:",
    successful["BARTScore"].mean()
)


# ============================================================
# FREE BART
# ============================================================

del bart_model
del bart_tokenizer

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()


# ============================================================
# 3. JARO SIMILARITY
# ============================================================

print("\n")
print("=" * 110)
print("3. Jaro Similarity")
print("=" * 110)

try:
    from rapidfuzz.distance import Jaro

except ImportError:
    raise ImportError(
        "rapidfuzz is not installed.\n"
        "Run:\n"
        "!pip install -q rapidfuzz"
    )


def jaro_score(
    prediction,
    reference
):

    # IMPORTANT:
    # No lowercasing, matching our replication notes.
    return Jaro.normalized_similarity(
        str(prediction),
        str(reference)
    )


successful["Jaro Similarity"] = [
    jaro_score(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

print(
    "Mean Jaro:",
    successful["Jaro Similarity"].mean()
)


# ============================================================
# 4 + 5. LCS METRICS
# ============================================================

print("\n")
print("=" * 110)
print("4. LCSubsequence Similarity")
print("5. LCSubsequence Word Count")
print("=" * 110)


def lcs_length(
    a,
    b
):

    if not a or not b:
        return 0

    # Make b the shorter sequence
    if len(b) > len(a):
        a, b = b, a

    previous = [
        0
    ] * (
        len(b) + 1
    )

    for x in a:

        current = [
            0
        ] * (
            len(b) + 1
        )

        for j, y in enumerate(
            b,
            start=1
        ):

            if x == y:

                current[j] = (
                    previous[j - 1]
                    + 1
                )

            else:

                current[j] = max(
                    previous[j],
                    current[j - 1]
                )

        previous = current

    return previous[-1]


def lcs_character_similarity(
    prediction,
    reference
):

    prediction = str(prediction)
    reference = str(reference)

    if not prediction and not reference:
        return 1.0

    if not prediction or not reference:
        return 0.0

    lcs = lcs_length(
        list(prediction),
        list(reference)
    )

    return (
        lcs /
        max(
            len(prediction),
            len(reference)
        )
    )


def lcs_word_count(
    prediction,
    reference
):

    pred_words = str(
        prediction
    ).split()

    ref_words = str(
        reference
    ).split()

    return lcs_length(
        pred_words,
        ref_words
    )


successful[
    "LCSubsequence Similarity"
] = [
    lcs_character_similarity(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

successful[
    "LCSubsequence Word Count"
] = [
    lcs_word_count(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]


print(
    "Mean LCS similarity:",
    successful[
        "LCSubsequence Similarity"
    ].mean()
)

print(
    "Mean LCS word count:",
    successful[
        "LCSubsequence Word Count"
    ].mean()
)


# ============================================================
# QUESTION-LEVEL METRICS FILE
# ============================================================

metric_columns = [
    "question_id",
    "condition",
    "company",
    "question",
    "prediction",
    "reference",
    "generated_tokens",
    "BERTScore",
    "BARTScore",
    "Jaro Similarity",
    "LCSubsequence Similarity",
    "LCSubsequence Word Count",
]

question_metrics = successful[
    [
        c
        for c in metric_columns
        if c in successful.columns
    ]
].copy()

question_metrics.to_csv(
    QUESTION_METRICS_FILE,
    index=False
)


# ============================================================
# FINAL TABLE 4 STYLE AGGREGATION
# ============================================================

rows = []

for condition in [
    "WITHOUT_METADATA",
    "WITH_METADATA",
]:

    subset = successful[
        successful["condition"]
        == condition
    ]

    assert len(subset) == 400

    rows.append({

        "Model":
            "Llama2-7b",

        "Condition":
            (
                "Without Metadata"
                if condition
                == "WITHOUT_METADATA"
                else
                "With Metadata"
            ),

        "BERTScore":
            subset[
                "BERTScore"
            ].mean(),

        "BARTScore":
            subset[
                "BARTScore"
            ].mean(),

        "Jaro Similarity":
            subset[
                "Jaro Similarity"
            ].mean(),

        "LCSubsequence Similarity":
            subset[
                "LCSubsequence Similarity"
            ].mean(),

        "LCSubsequence Word Count":
            subset[
                "LCSubsequence Word Count"
            ].mean(),
    })


table4 = pd.DataFrame(rows)


# ============================================================
# PRINT FINAL TABLE
# ============================================================

display_table = table4.copy()

display_table[
    "BERTScore"
] = display_table[
    "BERTScore"
].round(2)

display_table[
    "BARTScore"
] = display_table[
    "BARTScore"
].round(2)

display_table[
    "Jaro Similarity"
] = display_table[
    "Jaro Similarity"
].round(2)

display_table[
    "LCSubsequence Similarity"
] = display_table[
    "LCSubsequence Similarity"
].round(2)

display_table[
    "LCSubsequence Word Count"
] = (
    display_table[
        "LCSubsequence Word Count"
    ]
    .round(0)
    .astype(int)
)


print("\n\n")
print("=" * 110)
print("FINAL TABLE 4 — Llama2-7b")
print("=" * 110)

print(
    display_table.to_string(
        index=False
    )
)


# ============================================================
# SAVE
# ============================================================

table4.to_csv(
    TABLE4_FILE,
    index=False
)

print("\nSaved:")
print(QUESTION_METRICS_FILE)
print(TABLE4_FILE)

print("\n")
print("=" * 110)
print("EVALUATION COMPLETE")
print("=" * 110)
```

```python
# ============================================================
# LLAMA FINAL EVALUATION — FIX EMPTY answer_output
# ============================================================

import os
import re
import gc
import numpy as np
import pandas as pd
import torch
from tqdm.auto import tqdm

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

GENERATION_FILE = "/kaggle/working/llama_final_800_generation.csv"

QUESTION_METRICS_FILE = (
    "/kaggle/working/llama2_7b_question_metrics.csv"
)

TABLE4_FILE = (
    "/kaggle/working/llama2_7b_final_table4.csv"
)

BERT_BATCH_SIZE = 8
BART_BATCH_SIZE = 4

BERT_MODEL = "roberta-large"
BART_MODEL = "facebook/bart-large-cnn"

# ------------------------------------------------------------
# LOAD GENERATIONS
# ------------------------------------------------------------

gen_df = pd.read_csv(GENERATION_FILE)

print("=" * 110)
print("LLAMA-2-7B FINAL EVALUATION")
print("=" * 110)

print("Generation shape:", gen_df.shape)

# ------------------------------------------------------------
# VALIDATE 800 GENERATIONS
# ------------------------------------------------------------

if "status" not in gen_df.columns:
    raise ValueError("Missing 'status' column.")

successful = gen_df[
    gen_df["status"].astype(str).str.upper() == "OK"
].copy()

print("Successful rows:", len(successful))

if len(successful) != 800:
    raise ValueError(
        f"Expected 800 successful generations, got {len(successful)}"
    )

counts = successful["condition"].value_counts().to_dict()

print("Condition counts:", counts)

if counts.get("WITHOUT_METADATA", 0) != 400:
    raise ValueError("Expected 400 WITHOUT_METADATA rows.")

if counts.get("WITH_METADATA", 0) != 400:
    raise ValueError("Expected 400 WITH_METADATA rows.")

# ------------------------------------------------------------
# PREPARE PREDICTIONS
# ------------------------------------------------------------

successful["prediction"] = (
    successful["answer_output"]
    .fillna("")
    .astype(str)
    .str.strip()
)

successful["reference"] = (
    successful["ground_truth_answer"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------
# IMPORTANT FIX:
# One answer_output is missing, but raw_output contains the
# actual generated answer.
#
# Recover ONLY that missing answer_output from raw_output.
# We do NOT delete the question.
# ------------------------------------------------------------

missing_mask = successful["prediction"].eq("")

missing_count = missing_mask.sum()

print("\nMissing answer_output BEFORE FIX:", missing_count)

if missing_count > 0:

    raw_fallback = (
        successful.loc[missing_mask, "raw_output"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    successful.loc[missing_mask, "prediction"] = raw_fallback

    print(
        f"Recovered {missing_count} missing answer_output "
        "value(s) from raw_output."
    )

# ------------------------------------------------------------
# VERIFY NO EMPTY PREDICTIONS REMAIN
# ------------------------------------------------------------

remaining_empty = (
    successful["prediction"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("Empty predictions AFTER FIX:", remaining_empty)

if remaining_empty > 0:
    raise ValueError(
        f"{remaining_empty} predictions are still empty."
    )

print("✓ All 800 predictions are now non-empty.")

# ------------------------------------------------------------
# IMPORTANT:
# REMOVE LLAMA MODEL / TOKENIZER IF THEY STILL EXIST
# ------------------------------------------------------------

print("\nReleasing any remaining Llama GPU memory...")

try:
    del model
except:
    pass

try:
    del tokenizer
except:
    pass

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

print("GPU memory released.")

# ============================================================
# 1. BERTScore
# ============================================================

print("\n" + "=" * 110)
print("1. BERTScore")
print("=" * 110)

try:
    from bert_score import score as bert_score
except ImportError:
    raise ImportError(
        "bert-score is not installed.\n"
        "Run:\n"
        "!pip install -q bert-score"
    )

bert_device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

predictions_all = (
    successful["prediction"].tolist()
)

references_all = (
    successful["reference"].tolist()
)

print("Model:", BERT_MODEL)
print("Device:", bert_device)

P, R, F1 = bert_score(
    predictions_all,
    references_all,
    model_type=BERT_MODEL,
    lang="en",
    batch_size=BERT_BATCH_SIZE,
    device=bert_device,
    rescale_with_baseline=False,
    verbose=True,
)

successful["BERTScore"] = (
    F1.detach()
    .cpu()
    .numpy()
)

print(
    "Mean BERTScore:",
    successful["BERTScore"].mean()
)

# ------------------------------------------------------------
# Free BERTScore memory
# ------------------------------------------------------------

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

# ============================================================
# 2. BARTScore
# ============================================================

print("\n" + "=" * 110)
print("2. BARTScore")
print("=" * 110)

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
)

import torch.nn.functional as F

bart_device = (
    torch.device("cuda")
    if torch.cuda.is_available()
    else torch.device("cpu")
)

print("Model:", BART_MODEL)
print("Device:", bart_device)

bart_tokenizer = AutoTokenizer.from_pretrained(
    BART_MODEL
)

bart_model = AutoModelForSeq2SeqLM.from_pretrained(
    BART_MODEL
)

bart_model = bart_model.to(bart_device)
bart_model.eval()

# ------------------------------------------------------------
# BART directional score
# ------------------------------------------------------------

def bart_directional_score(
    sources,
    targets,
    batch_size=4,
    max_length=1024
):

    scores = []

    for start in tqdm(
        range(0, len(sources), batch_size),
        desc="BARTScore"
    ):

        src_batch = sources[
            start:start + batch_size
        ]

        tgt_batch = targets[
            start:start + batch_size
        ]

        src = bart_tokenizer(
            src_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt",
        )

        tgt = bart_tokenizer(
            tgt_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt",
        )

        src_input_ids = src["input_ids"].to(
            bart_device
        )

        src_attention_mask = src["attention_mask"].to(
            bart_device
        )

        tgt_input_ids = tgt["input_ids"].to(
            bart_device
        )

        with torch.no_grad():

            outputs = bart_model(
                input_ids=src_input_ids,
                attention_mask=src_attention_mask,
                labels=tgt_input_ids,
            )

        logits = outputs.logits

        vocab_size = logits.shape[-1]

        token_losses = F.cross_entropy(
            logits.reshape(-1, vocab_size),
            tgt_input_ids.reshape(-1),
            ignore_index=bart_tokenizer.pad_token_id,
            reduction="none",
        )

        token_losses = token_losses.reshape(
            tgt_input_ids.shape
        )

        mask = (
            tgt_input_ids
            != bart_tokenizer.pad_token_id
        )

        token_count = (
            mask.sum(dim=1)
            .clamp(min=1)
        )

        batch_scores = (
            -token_losses.sum(dim=1)
            / token_count
        )

        scores.extend(
            batch_scores
            .detach()
            .cpu()
            .numpy()
            .tolist()
        )

    return np.asarray(
        scores,
        dtype=np.float64
    )

# ------------------------------------------------------------
# BART directions
# ------------------------------------------------------------

refs = references_all
hyps = predictions_all

print("\nReference -> Hypothesis")

bart_P = bart_directional_score(
    refs,
    hyps,
    batch_size=BART_BATCH_SIZE,
    max_length=1024,
)

print("\nHypothesis -> Reference")

bart_R = bart_directional_score(
    hyps,
    refs,
    batch_size=BART_BATCH_SIZE,
    max_length=1024,
)

bart_F = (
    bart_P + bart_R
) / 2.0

successful["BARTScore"] = bart_F

print(
    "\nMean BARTScore:",
    successful["BARTScore"].mean()
)

# ------------------------------------------------------------
# Free BART
# ------------------------------------------------------------

del bart_model
del bart_tokenizer

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

# ============================================================
# 3. JARO SIMILARITY
# ============================================================

print("\n" + "=" * 110)
print("3. Jaro Similarity")
print("=" * 110)

try:
    from rapidfuzz.distance import Jaro
except ImportError:
    raise ImportError(
        "rapidfuzz is not installed.\n"
        "Run:\n"
        "!pip install -q rapidfuzz"
    )

def jaro_score(
    prediction,
    reference
):

    return Jaro.normalized_similarity(
        str(prediction),
        str(reference)
    )

successful["Jaro Similarity"] = [
    jaro_score(p, r)
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

print(
    "Mean Jaro:",
    successful["Jaro Similarity"].mean()
)

# ============================================================
# 4 + 5. LCS METRICS
# ============================================================

print("\n" + "=" * 110)
print("4. LCSubsequence Similarity")
print("5. LCSubsequence Word Count")
print("=" * 110)

def lcs_length(a, b):

    if not a or not b:
        return 0

    if len(b) > len(a):
        a, b = b, a

    previous = [0] * (len(b) + 1)

    for x in a:

        current = [0] * (len(b) + 1)

        for j, y in enumerate(
            b,
            start=1
        ):

            if x == y:

                current[j] = (
                    previous[j - 1] + 1
                )

            else:

                current[j] = max(
                    previous[j],
                    current[j - 1]
                )

        previous = current

    return previous[-1]


def lcs_character_similarity(
    prediction,
    reference
):

    prediction = str(prediction)
    reference = str(reference)

    if not prediction and not reference:
        return 1.0

    if not prediction or not reference:
        return 0.0

    lcs = lcs_length(
        list(prediction),
        list(reference)
    )

    return (
        lcs /
        max(
            len(prediction),
            len(reference)
        )
    )


def lcs_word_count(
    prediction,
    reference
):

    pred_words = str(
        prediction
    ).split()

    ref_words = str(
        reference
    ).split()

    return lcs_length(
        pred_words,
        ref_words
    )


successful[
    "LCSubsequence Similarity"
] = [
    lcs_character_similarity(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

successful[
    "LCSubsequence Word Count"
] = [
    lcs_word_count(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

print(
    "Mean LCS similarity:",
    successful[
        "LCSubsequence Similarity"
    ].mean()
)

print(
    "Mean LCS word count:",
    successful[
        "LCSubsequence Word Count"
    ].mean()
)

# ============================================================
# SAVE QUESTION-LEVEL METRICS
# ============================================================

metric_columns = [
    "question_id",
    "condition",
    "company",
    "question",
    "prediction",
    "reference",
    "generated_tokens",
    "BERTScore",
    "BARTScore",
    "Jaro Similarity",
    "LCSubsequence Similarity",
    "LCSubsequence Word Count",
]

question_metrics = successful[
    [
        c
        for c in metric_columns
        if c in successful.columns
    ]
].copy()

question_metrics.to_csv(
    QUESTION_METRICS_FILE,
    index=False
)

print(
    "\nSaved question-level metrics:"
)
print(QUESTION_METRICS_FILE)

# ============================================================
# FINAL TABLE 4 STYLE AGGREGATION
# ============================================================

rows = []

for condition in [
    "WITHOUT_METADATA",
    "WITH_METADATA",
]:

    subset = successful[
        successful["condition"] == condition
    ]

    print(
        f"\n{condition}: {len(subset)} rows"
    )

    if len(subset) != 400:
        raise ValueError(
            f"Expected 400 rows for {condition}, "
            f"found {len(subset)}"
        )

    rows.append({

        "Model":
            "Llama2-7b",

        "Condition":
            (
                "Without Metadata"
                if condition
                == "WITHOUT_METADATA"
                else
                "With Metadata"
            ),

        "BERTScore":
            subset["BERTScore"].mean(),

        "BARTScore":
            subset["BARTScore"].mean(),

        "Jaro Similarity":
            subset["Jaro Similarity"].mean(),

        "LCSubsequence Similarity":
            subset["LCSubsequence Similarity"].mean(),

        "LCSubsequence Word Count":
            subset["LCSubsequence Word Count"].mean(),

    })

table4 = pd.DataFrame(rows)

# ============================================================
# DISPLAY FINAL TABLE
# ============================================================

display_table = table4.copy()

display_table["BERTScore"] = (
    display_table["BERTScore"].round(2)
)

display_table["BARTScore"] = (
    display_table["BARTScore"].round(2)
)

display_table["Jaro Similarity"] = (
    display_table["Jaro Similarity"].round(2)
)

display_table["LCSubsequence Similarity"] = (
    display_table[
        "LCSubsequence Similarity"
    ].round(2)
)

display_table["LCSubsequence Word Count"] = (
    display_table[
        "LCSubsequence Word Count"
    ]
    .round(0)
    .astype(int)
)

print("\n\n")
print("=" * 110)
print("FINAL TABLE 4 — Llama2-7b")
print("=" * 110)

print(
    display_table.to_string(
        index=False
    )
)

# ============================================================
# SAVE FINAL TABLE
# ============================================================

table4.to_csv(
    TABLE4_FILE,
    index=False
)

print("\nSaved:")
print(QUESTION_METRICS_FILE)
print(TABLE4_FILE)

print("\n")
print("=" * 110)
print("LLAMA-2-7B EVALUATION COMPLETE")
print("=" * 110)
```
