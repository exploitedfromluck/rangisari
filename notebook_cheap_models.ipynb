```python
# ============================================================
# PHASE 2B — CREATE THE PAPER'S CHUNKER
# ============================================================

CHUNK_SIZE = 1024
CHUNK_OVERLAP = 0
SEPARATORS = ["\n\n", "\n", " ", ""]

splitter = RecursiveCharacterTextSplitter(
    chunk_size=CHUNK_SIZE,
    chunk_overlap=CHUNK_OVERLAP,
    separators=SEPARATORS
)

print("Chunker configuration:")
print("  chunk_size:", CHUNK_SIZE)
print("  chunk_overlap:", CHUNK_OVERLAP)
print("  separators:", SEPARATORS)
```

```python
# ============================================================
# PHASE 2C — TEST CHUNKING ON ONE TRANSCRIPT
# ============================================================

sample_path = transcript_files[0]

with open(
    sample_path,
    "r",
    encoding="utf-8",
    errors="replace"
) as f:
    sample_text = f.read()

sample_chunks = splitter.split_text(sample_text)

print("Transcript:", os.path.basename(sample_path))
print("Original characters:", len(sample_text))
print("Number of chunks:", len(sample_chunks))

print("\nChunk length statistics:")
chunk_lengths = [len(chunk) for chunk in sample_chunks]

print("Minimum:", min(chunk_lengths))
print("Maximum:", max(chunk_lengths))
print("Average:", sum(chunk_lengths) / len(chunk_lengths))

print("\nFirst 5 chunks:")
for i, chunk in enumerate(sample_chunks[:5]):
    print("\n" + "=" * 80)
    print(f"CHUNK {i}")
    print(f"Characters: {len(chunk)}")
    print("=" * 80)
    print(chunk)
```

```python
# ============================================================
# PHASE 2D — CHUNK ALL 50 TRANSCRIPTS
# ============================================================

all_chunks = []

for transcript_path in transcript_files:

    filename = os.path.basename(transcript_path)

    # Read transcript
    with open(
        transcript_path,
        "r",
        encoding="utf-8",
        errors="replace"
    ) as f:
        text = f.read()

    # Split using the validated paper configuration
    chunks = splitter.split_text(text)

    # Store every chunk with its source metadata
    for chunk_index, chunk_text in enumerate(chunks):

        all_chunks.append({
            "company": None,  # We will derive/validate this separately
            "transcript": filename,
            "chunk_id": f"{Path(filename).stem}_chunk_{chunk_index:04d}",
            "chunk_index": chunk_index,
            "text": chunk_text,
            "char_count": len(chunk_text)
        })

print("Total chunks created:", len(all_chunks))
```

```python
# ============================================================
# PHASE 2E — CHUNK VALIDATION
# ============================================================

chunks_df = pd.DataFrame(all_chunks)

print("Chunk dataframe shape:", chunks_df.shape)

print("\nChunk length statistics:")
print(chunks_df["char_count"].describe())

print("\nMaximum chunk length:", chunks_df["char_count"].max())
print("Minimum chunk length:", chunks_df["char_count"].min())

print("\nChunks exceeding 1024 characters:")
print((chunks_df["char_count"] > 1024).sum())

print("\nTranscripts represented:")
print(chunks_df["transcript"].nunique())
```

```python
# ============================================================
# PHASE 2F — CORPUS METADATA
# 2F-1: Build inventory of all available transcripts
# ============================================================

transcript_inventory = []

for transcript_path in transcript_files:

    filename = os.path.basename(transcript_path)

    transcript_inventory.append({
        "transcript": filename,
        "file_path": transcript_path
    })

transcript_inventory_df = pd.DataFrame(transcript_inventory)

print("Available transcript files:", len(transcript_inventory_df))

display(
    transcript_inventory_df.head(10)
)
```

```python
# ============================================================
# PHASE 2F-2 — PARSE COMPANY METADATA FROM FILENAMES
# ============================================================

import re

def parse_transcript_metadata(filename):
    """
    Convert:
        01_Reliance_Industries_Q1_FY24.txt

    into:
        company = Reliance Industries
        transcript = original filename
    """

    stem = os.path.splitext(filename)[0]

    # Remove the numeric prefix, e.g. "01_"
    stem = re.sub(r"^\d+_", "", stem)

    # Remove the quarter/year suffix
    company_name = re.sub(r"_Q1_FY24$", "", stem)

    # Convert underscores to spaces
    company_name = company_name.replace("_", " ")

    return company_name


transcript_inventory_df["company"] = (
    transcript_inventory_df["transcript"]
    .apply(parse_transcript_metadata)
)

print("Transcript metadata created:")
display(transcript_inventory_df.head(15))
```

```python
# ============================================================
# PHASE 2F-3 — CHECK COMPANY/TICKER COVERAGE
# ============================================================

# Build a unique company → ticker reference from ground truth.
company_ticker_reference = (
    ground_truth[["company", "ticker"]]
    .drop_duplicates()
    .sort_values("company")
    .reset_index(drop=True)
)

print("Unique companies with ticker information in ground truth:",
      len(company_ticker_reference))

print("\nCompanies in ground truth:")
display(company_ticker_reference)
```

```python
# ============================================================
# PHASE 2F-3B — FIND THE 3 COMPANIES WITHOUT TICKERS
# ============================================================

corpus_companies = set(
    transcript_inventory_df["company"]
)

known_companies = set(
    company_ticker_reference["company"]
)

companies_without_ticker = sorted(
    corpus_companies - known_companies
)

print(
    "Corpus companies without ticker information in ground truth:",
    len(companies_without_ticker)
)

for company in companies_without_ticker:
    print(" -", company)
```

```python
# ============================================================
# PHASE 2F-4 — COMPLETE CORPUS COMPANY/TICKER METADATA
# ============================================================

# Explicit ticker mapping for the 3 transcripts that have
# no questions in ground_truth_400.csv.
#
# These are corpus metadata only; they are not being inferred
# from the 400-question dataset.

manual_ticker_mapping = {
    "Power Grid Corporation": "POWERGRID",
    "Reliance Industries": "RELIANCE",
    "Tata Motors": "TATAMOTORS",
}

# Start with the 42 company/ticker pairs available in ground truth.
company_ticker_map = dict(
    zip(
        company_ticker_reference["company"],
        company_ticker_reference["ticker"]
    )
)

# Add the 3 missing corpus companies explicitly.
company_ticker_map.update(manual_ticker_mapping)

# Attach ticker to the 45-transcript inventory.
transcript_inventory_df["ticker"] = (
    transcript_inventory_df["company"]
    .map(company_ticker_map)
)

print("Total corpus transcripts:", len(transcript_inventory_df))
print(
    "Transcripts with missing company:",
    transcript_inventory_df["company"].isna().sum()
)
print(
    "Transcripts with missing ticker:",
    transcript_inventory_df["ticker"].isna().sum()
)

print("\nComplete corpus metadata:")
display(
    transcript_inventory_df[
        ["transcript", "company", "ticker"]
    ]
)
```

```python
# ============================================================
# PHASE 2F-5 — FINAL CORPUS METADATA VALIDATION
# ============================================================

print("Total transcripts:", len(transcript_inventory_df))
print(
    "Unique transcripts:",
    transcript_inventory_df["transcript"].nunique()
)
print(
    "Unique companies:",
    transcript_inventory_df["company"].nunique()
)
print(
    "Unique tickers:",
    transcript_inventory_df["ticker"].nunique()
)

print("\nMissing values:")
print(
    transcript_inventory_df[
        ["transcript", "company", "ticker"]
    ].isna().sum()
)

print("\nDuplicate transcript entries:")
print(
    transcript_inventory_df[
        transcript_inventory_df["transcript"].duplicated(keep=False)
    ]
)
```

```python
# ============================================================
# PHASE 2F-6 — ATTACH METADATA TO ALL CHUNKS
# ============================================================

# Start from the clean chunks produced in Phase 2D.
chunks_df = pd.DataFrame(all_chunks).copy()

# Remove the placeholder company column created during chunking.
chunks_df = chunks_df.drop(columns=["company"])

# Keep only the metadata needed for the corpus.
metadata_for_chunks = transcript_inventory_df[
    ["transcript", "company", "ticker"]
].copy()

# Attach company and ticker using the transcript filename.
chunks_df = chunks_df.merge(
    metadata_for_chunks,
    on="transcript",
    how="left",
    validate="many_to_one"
)

print("Final chunk dataframe shape:", chunks_df.shape)

print("\nColumns:")
print(chunks_df.columns.tolist())

print("\nMissing metadata:")
print(
    chunks_df[
        ["transcript", "company", "ticker"]
    ].isna().sum()
)

print("\nFirst 5 chunks:")
display(chunks_df.head())
```

```python
# ============================================================
# PHASE 2F-7 — FINAL CHUNK METADATA INTEGRITY CHECK
# ============================================================

print("Total chunks:", len(chunks_df))
print("Unique transcripts:", chunks_df["transcript"].nunique())
print("Unique companies:", chunks_df["company"].nunique())
print("Unique tickers:", chunks_df["ticker"].nunique())

print("\nMissing values:")
print(
    chunks_df[
        ["transcript", "company", "ticker"]
    ].isna().sum()
)

# Verify that every transcript has exactly one company/ticker.
metadata_consistency = (
    chunks_df
    .groupby("transcript")
    .agg(
        companies=("company", "nunique"),
        tickers=("ticker", "nunique")
    )
    .reset_index()
)

print("\nTranscripts with inconsistent metadata:")
display(
    metadata_consistency[
        (metadata_consistency["companies"] != 1) |
        (metadata_consistency["tickers"] != 1)
    ]
)
```

```python
# ================================================================
# PHASE 3.4 — GENERATING QUESTION EMBEDDINGS
# ================================================================

import os
import numpy as np
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer

print("=" * 70)
print("PHASE 3.4 — GENERATING QUESTION EMBEDDINGS")
print("=" * 70)

# ------------------------------------------------
# Paths
# ------------------------------------------------

DATA_DIR = "/kaggle/input/datasets/ramnarenkakulapati/emo-bros"

GROUND_TRUTH_PATH = os.path.join(
    DATA_DIR,
    "ground_truth_400.csv"
)

QUESTION_EMBEDDINGS_PATH = (
    "/kaggle/working/question_embeddings.npy"
)

QUESTION_METADATA_PATH = (
    "/kaggle/working/question_embedding_metadata.csv"
)

EMBEDDING_MODEL = "sentence-transformers/all-mpnet-base-v2"

# ------------------------------------------------
# Load ground truth
# ------------------------------------------------

ground_truth_df = pd.read_csv(GROUND_TRUTH_PATH)

print(f"Ground truth shape: {ground_truth_df.shape}")

assert len(ground_truth_df) == 400
assert "question_id" in ground_truth_df.columns
assert "question" in ground_truth_df.columns
assert "company" in ground_truth_df.columns
assert "ticker" in ground_truth_df.columns

questions = ground_truth_df["question"].astype(str).tolist()

print(f"Questions:           {len(questions)}")

# ------------------------------------------------
# Load embedding model
# ------------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Embedding model:     {EMBEDDING_MODEL}")
print(f"Device:              {device}")

question_embedder = SentenceTransformer(
    EMBEDDING_MODEL,
    device=device
)

embedding_dimension = (
    question_embedder.get_sentence_embedding_dimension()
)

print(f"Embedding dimension: {embedding_dimension}")

# ------------------------------------------------
# Generate embeddings
# ------------------------------------------------

print("\nGenerating question embeddings...")

question_embeddings = question_embedder.encode(
    questions,
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True
).astype(np.float32)

print(f"Raw shape: {question_embeddings.shape}")

# ------------------------------------------------
# Normalize for cosine similarity
# ------------------------------------------------

norms = np.linalg.norm(
    question_embeddings,
    axis=1,
    keepdims=True
)

question_embeddings = (
    question_embeddings / (norms + 1e-8)
)

# ------------------------------------------------
# Save
# ------------------------------------------------

np.save(
    QUESTION_EMBEDDINGS_PATH,
    question_embeddings
)

ground_truth_df[
    ["question_id", "company", "ticker"]
].to_csv(
    QUESTION_METADATA_PATH,
    index=False
)

# ------------------------------------------------
# Validation
# ------------------------------------------------

saved_embeddings = np.load(
    QUESTION_EMBEDDINGS_PATH
)

assert saved_embeddings.shape == (400, 768)
assert np.isfinite(saved_embeddings).all()

norms_check = np.linalg.norm(
    saved_embeddings,
    axis=1
)

assert np.allclose(
    norms_check,
    1.0,
    atol=1e-5
)

metadata_check = pd.read_csv(
    QUESTION_METADATA_PATH
)

assert (
    metadata_check["question_id"].tolist()
    == ground_truth_df["question_id"].tolist()
)

print("\n" + "=" * 70)
print("PHASE 3.4 CHECK PASSED")
print("=" * 70)

print("✓ 400 questions embedded")
print(f"✓ Shape: {saved_embeddings.shape}")
print("✓ Dimension: 768")
print("✓ No NaN/Inf values")
print("✓ Embeddings normalized")
print("✓ Question ordering verified")
print("✓ question_embeddings.npy saved")
print("=" * 70)
```

```python
# ================================================================
# PHASE 4.2 — MMR CONFIGURATION + FUNCTION
# ================================================================

import numpy as np

print("=" * 70)
print("PHASE 4.2 — MMR CONFIGURATION")
print("=" * 70)

# ------------------------------------------------
# Retrieval parameters
# ------------------------------------------------

FETCH_K = 20       # Paper: retrieve 20 candidates
FINAL_K = 4        # Paper: use 4 chunks in final context
MMR_LAMBDA = 0.5   # Implementation assumption
                   # Paper does not specify lambda

print(f"Candidate chunks (fetch_k): {FETCH_K}")
print(f"Final chunks (k):           {FINAL_K}")
print(f"MMR lambda:                 {MMR_LAMBDA}")
print("MMR lambda = 0.5 is an implementation assumption.")
print("The paper specifies 20 candidates and 4 final chunks.")

# ------------------------------------------------
# MMR function
# ------------------------------------------------

def maximal_marginal_relevance(
    query_embedding,
    candidate_embeddings,
    candidate_indices,
    lambda_mult=MMR_LAMBDA,
    k=FINAL_K
):
    """
    Select k chunks using Maximal Marginal Relevance.

    Relevance:
        similarity between query and candidate

    Diversity:
        similarity between candidate and already-selected chunks

    MMR:
        lambda * relevance
        - (1 - lambda) * diversity
    """

    candidate_indices = np.asarray(candidate_indices)

    if len(candidate_indices) == 0:
        return []

    k = min(k, len(candidate_indices))

    # ------------------------------------------------
    # Query-candidate relevance
    # ------------------------------------------------

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32
    )

    candidate_embeddings = np.asarray(
        candidate_embeddings,
        dtype=np.float32
    )

    relevance_scores = (
        candidate_embeddings @ query_embedding
    )

    # ------------------------------------------------
    # Candidate-candidate similarity
    # ------------------------------------------------

    candidate_similarity = (
        candidate_embeddings @ candidate_embeddings.T
    )

    # ------------------------------------------------
    # First selection = highest relevance
    # ------------------------------------------------

    selected_positions = [
        int(np.argmax(relevance_scores))
    ]

    # ------------------------------------------------
    # Iteratively select remaining chunks
    # ------------------------------------------------

    while len(selected_positions) < k:

        best_position = None
        best_score = -np.inf

        for position in range(len(candidate_indices)):

            if position in selected_positions:
                continue

            relevance = relevance_scores[position]

            # Similarity to the most similar
            # already-selected candidate
            diversity_penalty = max(
                candidate_similarity[
                    position,
                    selected_positions
                ]
            )

            mmr_score = (
                lambda_mult * relevance
                - (1 - lambda_mult) * diversity_penalty
            )

            if mmr_score > best_score:
                best_score = mmr_score
                best_position = position

        selected_positions.append(best_position)

    # Convert local candidate positions
    # back to global chunk indices.
    selected_global_indices = [
        int(candidate_indices[position])
        for position in selected_positions
    ]

    return selected_global_indices


# ------------------------------------------------
# Smoke test
# ------------------------------------------------

test_question_embedding = question_embeddings[0]

# Use the first 20 chunks simply to test the function.
# This is NOT the actual retrieval.
test_candidate_indices = np.arange(
    min(FETCH_K, len(chunks_df))
)

test_candidate_embeddings = (
    chunk_embeddings[test_candidate_indices]
)

test_selected = maximal_marginal_relevance(
    query_embedding=test_question_embedding,
    candidate_embeddings=test_candidate_embeddings,
    candidate_indices=test_candidate_indices,
    lambda_mult=MMR_LAMBDA,
    k=FINAL_K
)

# ------------------------------------------------
# Validation
# ------------------------------------------------

assert len(test_selected) == FINAL_K
assert len(set(test_selected)) == FINAL_K

for idx in test_selected:
    assert 0 <= idx < len(chunks_df)

print("\n" + "=" * 70)
print("PHASE 4.2 CHECK PASSED")
print("=" * 70)

print(f"✓ MMR function defined")
print(f"✓ fetch_k = {FETCH_K}")
print(f"✓ final_k = {FINAL_K}")
print(f"✓ MMR lambda = {MMR_LAMBDA}")
print(f"✓ Smoke test selected {len(test_selected)} unique chunks")
print("=" * 70)
```

```python
# ================================================================
# PHASE 4.3 — RETRIEVAL: WITHOUT_METADATA vs WITH_METADATA
# ================================================================

import os
import numpy as np
import pandas as pd

print("=" * 70)
print("PHASE 4.3 — RETRIEVAL: WITHOUT_METADATA vs WITH_METADATA")
print("=" * 70)

# ----------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------

CHUNKS_PATH = "/kaggle/working/chunks.csv"
CHUNK_METADATA_PATH = "/kaggle/working/chunk_metadata.csv"
CHUNK_EMBEDDINGS_PATH = "/kaggle/working/chunk_embeddings.npy"
QUESTION_EMBEDDINGS_PATH = "/kaggle/working/question_embeddings.npy"
GROUND_TRUTH_PATH = (
    "/kaggle/input/datasets/ramnarenkakulapati/"
    "emo-bros/ground_truth_400.csv"
)

WITHOUT_METADATA_PATH = "/kaggle/working/without_metadata_retrieval.csv"
WITH_METADATA_PATH = "/kaggle/working/with_metadata_retrieval.csv"

# ----------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------

chunks_df = pd.read_csv(CHUNKS_PATH)
chunk_metadata_df = pd.read_csv(CHUNK_METADATA_PATH)
chunk_embeddings = np.load(CHUNK_EMBEDDINGS_PATH)
question_embeddings = np.load(QUESTION_EMBEDDINGS_PATH)
ground_truth_df = pd.read_csv(GROUND_TRUTH_PATH)

print(f"Chunks:              {chunks_df.shape}")
print(f"Chunk metadata:      {chunk_metadata_df.shape}")
print(f"Chunk embeddings:    {chunk_embeddings.shape}")
print(f"Question embeddings: {question_embeddings.shape}")
print(f"Ground truth:        {ground_truth_df.shape}")

# ----------------------------------------------------------------
# VALIDATION
# ----------------------------------------------------------------

assert len(chunks_df) == len(chunk_embeddings)
assert len(chunks_df) == len(chunk_metadata_df)
assert len(ground_truth_df) == len(question_embeddings)

assert chunk_embeddings.shape[1] == question_embeddings.shape[1]
assert "question_id" in ground_truth_df.columns
assert "company" in ground_truth_df.columns
assert "ticker" in ground_truth_df.columns
assert "chunk_id" in chunks_df.columns
assert "company" in chunks_df.columns
assert "ticker" in chunks_df.columns

assert chunks_df["chunk_id"].is_unique
assert ground_truth_df["question_id"].is_unique

# ----------------------------------------------------------------
# NORMALIZE EMBEDDINGS
# ----------------------------------------------------------------
# After normalization, dot product = cosine similarity.

chunk_norms = np.linalg.norm(
    chunk_embeddings, axis=1, keepdims=True
)

question_norms = np.linalg.norm(
    question_embeddings, axis=1, keepdims=True
)

chunk_embeddings_normalized = (
    chunk_embeddings / (chunk_norms + 1e-8)
)

question_embeddings_normalized = (
    question_embeddings / (question_norms + 1e-8)
)

assert np.isfinite(chunk_embeddings_normalized).all()
assert np.isfinite(question_embeddings_normalized).all()

# ----------------------------------------------------------------
# RETRIEVE TOP-20 BY COSINE SIMILARITY
# ----------------------------------------------------------------

def retrieve_candidates(
    query_embedding,
    candidate_indices,
    fetch_k=20
):
    """
    Retrieve the top fetch_k chunks using cosine similarity.

    candidate_indices:
        Chunks allowed to participate in retrieval.
        For WITHOUT_METADATA: all chunks.
        For WITH_METADATA: only chunks belonging to
        the question's company.
    """

    candidate_indices = np.asarray(
        candidate_indices,
        dtype=np.int64
    )

    candidate_vectors = chunk_embeddings_normalized[
        candidate_indices
    ]

    # Because both vectors are normalized:
    # dot product = cosine similarity.
    similarities = candidate_vectors @ query_embedding

    top_n = min(fetch_k, len(candidate_indices))

    top_positions = np.argsort(
        -similarities
    )[:top_n]

    top_indices = candidate_indices[top_positions]
    top_scores = similarities[top_positions]

    return top_indices, top_scores


# ----------------------------------------------------------------
# RUN BOTH RETRIEVAL CONDITIONS
# ----------------------------------------------------------------

FETCH_K = 20
FINAL_K = 4

without_metadata_rows = []
with_metadata_rows = []

all_chunk_indices = np.arange(len(chunks_df))

print("\nStarting retrieval...")
print(f"Questions: {len(ground_truth_df)}")
print(f"Candidate chunks: {FETCH_K}")
print(f"Final chunks: {FINAL_K}")

# ----------------------------------------------------------------
# QUESTION-BY-QUESTION RETRIEVAL
# ----------------------------------------------------------------

for q_idx, row in ground_truth_df.iterrows():

    question_id = row["question_id"]
    company = row["company"]
    ticker = row["ticker"]

    query_embedding = question_embeddings_normalized[q_idx]

    # ============================================================
    # CONDITION 1 — WITHOUT_METADATA
    # ============================================================

    # No company restriction.
    # All transcript chunks are eligible.

    candidate_indices = all_chunk_indices

    top20_indices, top20_scores = retrieve_candidates(
        query_embedding=query_embedding,
        candidate_indices=candidate_indices,
        fetch_k=FETCH_K
    )

    top20_embeddings = chunk_embeddings_normalized[
        top20_indices
    ]

    # MMR selects the final 4 from the 20 candidates.
    selected_positions = maximal_marginal_relevance(
        query_embedding=query_embedding,
        candidate_embeddings=top20_embeddings,
        candidate_indices=np.arange(len(top20_indices)),
        lambda_mult=MMR_LAMBDA,
        k=FINAL_K
    )

    selected_indices = top20_indices[selected_positions]

    # Save final 4
    for rank, chunk_idx in enumerate(selected_indices, start=1):

        chunk_row = chunks_df.iloc[chunk_idx]

        # Similarity of this final chunk to the question.
        # This is NOT the MMR score.
        similarity_score = float(
            chunk_embeddings_normalized[chunk_idx]
            @ query_embedding
        )

        without_metadata_rows.append({
            "question_id": question_id,
            "company": company,
            "ticker": ticker,
            "condition": "WITHOUT_METADATA",
            "rank": rank,
            "chunk_id": chunk_row["chunk_id"],
            "chunk_index": chunk_row["chunk_index"],
            "transcript": chunk_row["transcript"],
            "retrieved_company": chunk_row["company"],
            "retrieved_ticker": chunk_row["ticker"],
            "similarity_score": similarity_score,
            "text": chunk_row["text"]
        })

    # ============================================================
    # CONDITION 2 — WITH_METADATA
    # ============================================================

    # Metadata filter:
    # only chunks belonging to the company associated
    # with the question are allowed into retrieval.

    metadata_mask = (
        chunks_df["company"] == company
    )

    metadata_candidate_indices = np.flatnonzero(
        metadata_mask.to_numpy()
    )

    assert len(metadata_candidate_indices) > 0, (
        f"No chunks found for company: {company}"
    )

    # Similarity search is performed ONLY inside the
    # metadata-filtered company documents.
    top20_indices, top20_scores = retrieve_candidates(
        query_embedding=query_embedding,
        candidate_indices=metadata_candidate_indices,
        fetch_k=FETCH_K
    )

    top20_embeddings = chunk_embeddings_normalized[
        top20_indices
    ]

    # MMR selects the final 4 from those 20.
    selected_positions = maximal_marginal_relevance(
        query_embedding=query_embedding,
        candidate_embeddings=top20_embeddings,
        candidate_indices=np.arange(len(top20_indices)),
        lambda_mult=MMR_LAMBDA,
        k=FINAL_K
    )

    selected_indices = top20_indices[selected_positions]

    # Critical validation:
    # metadata condition must not retrieve another company.
    retrieved_companies = chunks_df.iloc[
        selected_indices
    ]["company"].unique()

    assert len(retrieved_companies) == 1
    assert retrieved_companies[0] == company

    # Save final 4
    for rank, chunk_idx in enumerate(selected_indices, start=1):

        chunk_row = chunks_df.iloc[chunk_idx]

        similarity_score = float(
            chunk_embeddings_normalized[chunk_idx]
            @ query_embedding
        )

        with_metadata_rows.append({
            "question_id": question_id,
            "company": company,
            "ticker": ticker,
            "condition": "WITH_METADATA",
            "rank": rank,
            "chunk_id": chunk_row["chunk_id"],
            "chunk_index": chunk_row["chunk_index"],
            "transcript": chunk_row["transcript"],
            "retrieved_company": chunk_row["company"],
            "retrieved_ticker": chunk_row["ticker"],
            "similarity_score": similarity_score,
            "text": chunk_row["text"]
        })


# ----------------------------------------------------------------
# CREATE DATAFRAMES
# ----------------------------------------------------------------

without_metadata_df = pd.DataFrame(
    without_metadata_rows
)

with_metadata_df = pd.DataFrame(
    with_metadata_rows
)

# ----------------------------------------------------------------
# SAVE RESULTS
# ----------------------------------------------------------------

without_metadata_df.to_csv(
    WITHOUT_METADATA_PATH,
    index=False
)

with_metadata_df.to_csv(
    WITH_METADATA_PATH,
    index=False
)

# ----------------------------------------------------------------
# VALIDATION
# ----------------------------------------------------------------

expected_rows = len(ground_truth_df) * FINAL_K

assert len(without_metadata_df) == expected_rows
assert len(with_metadata_df) == expected_rows

assert (
    without_metadata_df.groupby("question_id")
    .size()
    .eq(FINAL_K)
    .all()
)

assert (
    with_metadata_df.groupby("question_id")
    .size()
    .eq(FINAL_K)
    .all()
)

assert set(without_metadata_df["rank"]) == {1, 2, 3, 4}
assert set(with_metadata_df["rank"]) == {1, 2, 3, 4}

# WITH_METADATA must have zero cross-company retrieval.
with_cross_company = (
    with_metadata_df["retrieved_company"]
    != with_metadata_df["company"]
).sum()

assert with_cross_company == 0

# WITHOUT_METADATA leakage analysis.
without_cross_company = (
    without_metadata_df["retrieved_company"]
    != without_metadata_df["company"]
).sum()

# ----------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 4.3 CHECK PASSED")
print("=" * 70)

print(f"✓ Questions processed:              {len(ground_truth_df)}")
print(f"✓ WITHOUT_METADATA rows:            {len(without_metadata_df)}")
print(f"✓ WITH_METADATA rows:               {len(with_metadata_df)}")
print(f"✓ Candidates per question:          {FETCH_K}")
print(f"✓ Final chunks per question:        {FINAL_K}")
print(f"✓ MMR lambda:                       {MMR_LAMBDA}")

print("\nRetrieval pipeline:")
print("  WITHOUT_METADATA:")
print("    All chunks → similarity top-20 → MMR → final 4")

print("\n  WITH_METADATA:")
print("    Company filter → similarity top-20 → MMR → final 4")

print("\nMetadata validation:")
print(f"  WITH_METADATA cross-company chunks: {with_cross_company}")

print("\nWITHOUT_METADATA leakage:")
print(f"  Cross-company chunks:               {without_cross_company}")

print("\nSaved:")
print(f"  ✓ {WITHOUT_METADATA_PATH}")
print(f"  ✓ {WITH_METADATA_PATH}")

print("=" * 70)
```

```python
# ================================================================
# PHASE 4.4 — RETRIEVAL VALIDATION & ANALYSIS
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("PHASE 4.4 — RETRIEVAL VALIDATION & ANALYSIS")
print("=" * 70)

# ----------------------------------------------------------------
# LOAD RETRIEVAL RESULTS
# ----------------------------------------------------------------

WITHOUT_METADATA_PATH = "/kaggle/working/without_metadata_retrieval.csv"
WITH_METADATA_PATH = "/kaggle/working/with_metadata_retrieval.csv"

without_df = pd.read_csv(WITHOUT_METADATA_PATH)
with_df = pd.read_csv(WITH_METADATA_PATH)

print(f"WITHOUT_METADATA: {without_df.shape}")
print(f"WITH_METADATA:    {with_df.shape}")

# ----------------------------------------------------------------
# VALIDATE BASIC STRUCTURE
# ----------------------------------------------------------------

assert len(without_df) == 1600
assert len(with_df) == 1600

assert without_df["question_id"].nunique() == 400
assert with_df["question_id"].nunique() == 400

assert set(without_df["rank"]) == {1, 2, 3, 4}
assert set(with_df["rank"]) == {1, 2, 3, 4}

# ----------------------------------------------------------------
# CORRECT COMPANY / TICKER CHECK
# ----------------------------------------------------------------

without_df["correct_company"] = (
    without_df["retrieved_company"]
    == without_df["company"]
)

with_df["correct_company"] = (
    with_df["retrieved_company"]
    == with_df["company"]
)

without_df["correct_ticker"] = (
    without_df["retrieved_ticker"]
    == without_df["ticker"]
)

with_df["correct_ticker"] = (
    with_df["retrieved_ticker"]
    == with_df["ticker"]
)

# ----------------------------------------------------------------
# CHUNK-LEVEL RETRIEVAL ACCURACY
# ----------------------------------------------------------------

without_company_accuracy = (
    without_df["correct_company"].mean()
)

with_company_accuracy = (
    with_df["correct_company"].mean()
)

without_ticker_accuracy = (
    without_df["correct_ticker"].mean()
)

with_ticker_accuracy = (
    with_df["correct_ticker"].mean()
)

# ----------------------------------------------------------------
# QUESTION-LEVEL RETRIEVAL RECALL
# ----------------------------------------------------------------
# Did at least ONE of the final 4 chunks come from
# the correct company?

without_question_recall = (
    without_df
    .groupby("question_id")["correct_company"]
    .any()
    .mean()
)

with_question_recall = (
    with_df
    .groupby("question_id")["correct_company"]
    .any()
    .mean()
)

# ----------------------------------------------------------------
# RANK-1 ACCURACY
# ----------------------------------------------------------------

without_rank1 = (
    without_df[without_df["rank"] == 1]["correct_company"].mean()
)

with_rank1 = (
    with_df[with_df["rank"] == 1]["correct_company"].mean()
)

# ----------------------------------------------------------------
# RANK-WISE ACCURACY
# ----------------------------------------------------------------

rank_results = []

for rank in [1, 2, 3, 4]:

    without_rank_accuracy = (
        without_df[without_df["rank"] == rank]
        ["correct_company"]
        .mean()
    )

    with_rank_accuracy = (
        with_df[with_df["rank"] == rank]
        ["correct_company"]
        .mean()
    )

    rank_results.append({
        "rank": rank,
        "without_metadata": without_rank_accuracy,
        "with_metadata": with_rank_accuracy
    })

rank_df = pd.DataFrame(rank_results)

# ----------------------------------------------------------------
# CROSS-COMPANY LEAKAGE
# ----------------------------------------------------------------

without_leakage_chunks = (
    (~without_df["correct_company"]).sum()
)

with_leakage_chunks = (
    (~with_df["correct_company"]).sum()
)

without_leakage_rate = (
    without_leakage_chunks / len(without_df)
)

with_leakage_rate = (
    with_leakage_chunks / len(with_df)
)

# ----------------------------------------------------------------
# QUESTIONS WITH AT LEAST ONE WRONG-COMPANY CHUNK
# ----------------------------------------------------------------

without_question_leakage = (
    without_df
    .groupby("question_id")["correct_company"]
    .apply(lambda x: (~x).any())
    .sum()
)

with_question_leakage = (
    with_df
    .groupby("question_id")["correct_company"]
    .apply(lambda x: (~x).any())
    .sum()
)

# ----------------------------------------------------------------
# QUESTIONS WITH ALL FOUR CORRECT
# ----------------------------------------------------------------

without_all_four_correct = (
    without_df
    .groupby("question_id")["correct_company"]
    .all()
    .sum()
)

with_all_four_correct = (
    with_df
    .groupby("question_id")["correct_company"]
    .all()
    .sum()
)

# ----------------------------------------------------------------
# SAVE RANK-WISE ANALYSIS
# ----------------------------------------------------------------

RANK_RESULTS_PATH = "/kaggle/working/retrieval_rank_analysis.csv"

rank_df.to_csv(
    RANK_RESULTS_PATH,
    index=False
)

# ----------------------------------------------------------------
# PRINT RESULTS
# ----------------------------------------------------------------

print("\n" + "=" * 70)
print("RETRIEVAL VALIDATION RESULTS")
print("=" * 70)

print("\n1. CHUNK-LEVEL CORRECT COMPANY")

print(
    f"WITHOUT_METADATA: "
    f"{without_company_accuracy:.4f} "
    f"({without_company_accuracy * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_company_accuracy:.4f} "
    f"({with_company_accuracy * 100:.2f}%)"
)

print("\n2. CHUNK-LEVEL CORRECT TICKER")

print(
    f"WITHOUT_METADATA: "
    f"{without_ticker_accuracy:.4f} "
    f"({without_ticker_accuracy * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_ticker_accuracy:.4f} "
    f"({with_ticker_accuracy * 100:.2f}%)"
)

print("\n3. QUESTION-LEVEL RECALL")
print("At least one correct-company chunk in final 4:")

print(
    f"WITHOUT_METADATA: "
    f"{without_question_recall:.4f} "
    f"({without_question_recall * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_question_recall:.4f} "
    f"({with_question_recall * 100:.2f}%)"
)

print("\n4. RANK-1 CORRECT COMPANY")

print(
    f"WITHOUT_METADATA: "
    f"{without_rank1:.4f} "
    f"({without_rank1 * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_rank1:.4f} "
    f"({with_rank1 * 100:.2f}%)"
)

print("\n5. CROSS-COMPANY LEAKAGE")

print(
    f"WITHOUT_METADATA: "
    f"{without_leakage_chunks} / {len(without_df)} "
    f"({without_leakage_rate * 100:.2f}%)"
)

print(
    f"WITH_METADATA:    "
    f"{with_leakage_chunks} / {len(with_df)} "
    f"({with_leakage_rate * 100:.2f}%)"
)

print("\n6. QUESTIONS WITH CROSS-COMPANY LEAKAGE")

print(
    f"WITHOUT_METADATA: "
    f"{without_question_leakage} / 400"
)

print(
    f"WITH_METADATA:    "
    f"{with_question_leakage} / 400"
)

print("\n7. QUESTIONS WHERE ALL 4 CHUNKS ARE CORRECT COMPANY")

print(
    f"WITHOUT_METADATA: "
    f"{without_all_four_correct} / 400"
)

print(
    f"WITH_METADATA:    "
    f"{with_all_four_correct} / 400"
)

print("\n8. RANK-WISE CORRECT COMPANY")

print(rank_df.to_string(index=False))

print("\n" + "=" * 70)
print("PHASE 4.4 CHECK PASSED")
print("=" * 70)

print("✓ 1,600 rows validated per condition")
print("✓ 400 questions validated per condition")
print("✓ Rank 1–4 validated")
print("✓ Company-level retrieval accuracy calculated")
print("✓ Question-level recall calculated")
print("✓ Cross-company leakage calculated")
print("✓ Rank-wise analysis saved")
print(f"✓ Saved: {RANK_RESULTS_PATH}")

print("=" * 70)
```

```python
# ============================================================
# PHASE 5.6 — LOAD MPNet FOR FLAN CONTEXT COMPRESSION
# ============================================================

from sentence_transformers import SentenceTransformer
import re
import numpy as np

EMBED_MODEL = "sentence-transformers/all-mpnet-base-v2"

print("=" * 70)
print("PHASE 5.6 — LOADING MPNet")
print("=" * 70)

embedder = SentenceTransformer(
    EMBED_MODEL,
    device="cuda"
)

print("Model:", EMBED_MODEL)
print("Device:", embedder.device)
print("Embedding dimension:", embedder.get_embedding_dimension())

assert embedder.get_embedding_dimension() == 768

print("\n✓ MPNet ready for FLAN context compression.")
```

```python
# ============================================================
# PHASE 5.7 — COVERAGE-AWARE EXTRACTIVE COMPRESSION
# ============================================================

def split_into_sentences(text):
    """
    Split retrieved transcript text into sentences while
    preserving the original sentence wording.
    """
    text = str(text).strip()

    if not text:
        return []

    sentences = re.split(
        r'(?<=[.!?])\s+',
        text
    )

    return [
        s.strip()
        for s in sentences
        if s.strip()
    ]


def get_fixed_prompt_tokens(question):
    """
    Count tokens used by everything except the retrieved
    evidence context.
    """
    fixed_text = (
        "Answer the following question based ONLY on the provided context. "
        "Do not use knowledge outside the context.\n\n"
        "Context:\n"
        "\n\nQuestion: "
        + question
        + "\n\nAnswer:"
    )

    return len(
        flan_tokenizer(
            fixed_text,
            add_special_tokens=True,
            truncation=False
        )["input_ids"]
    )


def compress_context_for_flan(
    question,
    contexts,
    max_total_tokens=512,
    top_candidates_per_chunk=3
):
    """
    Coverage-aware extractive compression for FLAN-T5-base.

    Each of the 4 retrieved chunks receives representation
    priority. Sentences are selected using question-sentence
    cosine similarity with MPNet.

    No generated summarization or paraphrasing is performed.
    """

    assert len(contexts) == 4, (
        f"Expected 4 retrieved chunks, got {len(contexts)}"
    )

    # --------------------------------------------------------
    # 1. Calculate available evidence-token budget
    # --------------------------------------------------------

    fixed_tokens = get_fixed_prompt_tokens(question)

    available_tokens = max_total_tokens - fixed_tokens

    if available_tokens <= 0:
        raise ValueError(
            f"Fixed prompt requires {fixed_tokens} tokens."
        )

    # --------------------------------------------------------
    # 2. Split each retrieved chunk into sentences
    # --------------------------------------------------------

    records = []

    for chunk_rank, chunk_text in enumerate(contexts, start=1):

        sentences = split_into_sentences(chunk_text)

        for sentence_index, sentence in enumerate(sentences):

            token_count = len(
                flan_tokenizer(
                    sentence,
                    add_special_tokens=False,
                    truncation=False
                )["input_ids"]
            )

            records.append({
                "chunk_rank": chunk_rank,
                "sentence_index": sentence_index,
                "text": sentence,
                "token_count": token_count
            })

    if not records:
        return "", {
            "fixed_tokens": fixed_tokens,
            "available_context_tokens": available_tokens,
            "used_context_tokens": 0,
            "total_sentences": 0,
            "selected_sentences": 0,
            "chunks_represented": 0
        }

    # --------------------------------------------------------
    # 3. Embed all candidate sentences + the question
    # --------------------------------------------------------

    sentence_texts = [
        r["text"]
        for r in records
    ]

    sentence_embeddings = embedder.encode(
        sentence_texts,
        batch_size=32,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    )

    question_embedding = embedder.encode(
        question,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    )

    # --------------------------------------------------------
    # 4. Question-sentence cosine similarity
    # --------------------------------------------------------

    similarities = sentence_embeddings @ question_embedding

    for record, score in zip(records, similarities):
        record["similarity"] = float(score)

    # --------------------------------------------------------
    # 5. Build candidate pool within EACH retrieved chunk
    # --------------------------------------------------------

    candidates = []

    for chunk_rank in range(1, 5):

        chunk_candidates = [
            r
            for r in records
            if r["chunk_rank"] == chunk_rank
        ]

        chunk_candidates.sort(
            key=lambda x: x["similarity"],
            reverse=True
        )

        candidates.extend(
            chunk_candidates[:top_candidates_per_chunk]
        )

    # --------------------------------------------------------
    # 6. First priority:
    #    select the best sentence from every chunk
    # --------------------------------------------------------

    selected = []
    used_tokens = 0

    for chunk_rank in range(1, 5):

        chunk_candidates = [
            r
            for r in candidates
            if r["chunk_rank"] == chunk_rank
        ]

        if not chunk_candidates:
            continue

        best = max(
            chunk_candidates,
            key=lambda x: x["similarity"]
        )

        separator_tokens = 2 if selected else 0

        if (
            used_tokens
            + separator_tokens
            + best["token_count"]
            <= available_tokens
        ):
            selected.append(best)

            used_tokens += (
                separator_tokens
                + best["token_count"]
            )

    # --------------------------------------------------------
    # 7. Fill remaining budget with globally relevant sentences
    # --------------------------------------------------------

    remaining = [
        r for r in candidates
        if not any(
            r["chunk_rank"] == s["chunk_rank"]
            and r["sentence_index"] == s["sentence_index"]
            for s in selected
        )
    ]

    remaining.sort(
        key=lambda x: x["similarity"],
        reverse=True
    )

    for record in remaining:

        separator_tokens = 2

        if (
            used_tokens
            + separator_tokens
            + record["token_count"]
            <= available_tokens
        ):
            selected.append(record)

            used_tokens += (
                separator_tokens
                + record["token_count"]
            )

    # --------------------------------------------------------
    # 8. Restore original transcript order
    # --------------------------------------------------------

    selected.sort(
        key=lambda x: (
            x["chunk_rank"],
            x["sentence_index"]
        )
    )

    # --------------------------------------------------------
    # 9. Preserve original wording exactly
    # --------------------------------------------------------

    compressed_context = "\n\n".join(
        r["text"]
        for r in selected
    )

    metadata = {
        "fixed_tokens": fixed_tokens,
        "available_context_tokens": available_tokens,
        "used_context_tokens": used_tokens,
        "total_sentences": len(records),
        "selected_sentences": len(selected),
        "chunks_represented": len(
            set(r["chunk_rank"] for r in selected)
        )
    }

    return compressed_context, metadata


print("=" * 70)
print("COVERAGE-AWARE COMPRESSION FUNCTION READY")
print("=" * 70)
print("✓ MPNet question-sentence similarity")
print("✓ Coverage across all 4 retrieved chunks")
print("✓ Original transcript wording preserved")
print("✓ FLAN-specific 512-token budget")
```

```python
# ============================================================
# PHASE 5.8 — APPLY COMPRESSION TO ALL 800 FLAN PROMPTS
# ============================================================

print("=" * 70)
print("PHASE 5.8 — COMPRESSING 800 FLAN-T5 PROMPTS")
print("=" * 70)

compressed_prompts = []
compression_records = []

for _, row in tqdm(
    flan_prompts_df.iterrows(),
    total=len(flan_prompts_df),
    desc="Compressing"
):

    question_id = row["question_id"]
    condition = row["condition"]
    question = row["question"]

    # Use already-computed Phase 4 retrievals
    source_df = (
        without_metadata_df
        if condition == "WITHOUT_METADATA"
        else with_metadata_df
    )

    retrieved = (
        source_df[
            source_df["question_id"] == question_id
        ]
        .sort_values("rank")
    )

    contexts = retrieved["text"].tolist()

    assert len(contexts) == 4

    # Question-aware, coverage-aware extraction
    compressed_context, metadata = (
        compress_context_for_flan(
            question=question,
            contexts=contexts,
            max_total_tokens=512
        )
    )

    # Final FLAN prompt
    final_prompt = (
        "Answer the following question based ONLY on the provided context. "
        "Do not use knowledge outside the context.\n\n"
        "Context:\n"
        + compressed_context
        + f"\n\nQuestion: {question}\n\n"
        "Answer:"
    )

    # Verify final token count
    final_tokens = len(
        flan_tokenizer(
            final_prompt,
            add_special_tokens=True,
            truncation=False
        )["input_ids"]
    )

    assert final_tokens <= 512, (
        f"{question_id}/{condition}: "
        f"{final_tokens} tokens > 512"
    )

    compressed_prompts.append(final_prompt)

    compression_records.append({
        "question_id": question_id,
        "condition": condition,
        "original_prompt_tokens": int(row["prompt_tokens"]),
        "final_prompt_tokens": int(final_tokens),
        **metadata
    })


# Store final prompts
flan_prompts_df["flan_prompt"] = compressed_prompts

# Store compression metadata
compression_df = pd.DataFrame(
    compression_records
)

print("\n" + "=" * 70)
print("COMPRESSION COMPLETE")
print("=" * 70)

print("Total prompts:", len(compression_df))

print(
    "Mean original tokens:",
    compression_df["original_prompt_tokens"].mean()
)

print(
    "Mean final tokens:",
    compression_df["final_prompt_tokens"].mean()
)

print(
    "Mean chunks represented:",
    compression_df["chunks_represented"].mean()
)

print(
    "\nPrompts <= 512:",
    (compression_df["final_prompt_tokens"] <= 512).sum(),
    "/",
    len(compression_df)
)

assert len(compression_df) == 800
assert compression_df["final_prompt_tokens"].max() <= 512

print("\n✓ All 800 prompts fit within FLAN-T5's 512-token limit.")
```

```python
# ============================================================
# PHASE 5.14 — FLAN-T5 METRIC EVALUATION
#
# Metrics:
#   1. BERTScore       — roberta-large
#   2. BARTScore       — facebook/bart-large-cnn
#   3. Jaro
#   4. LCSubsequence similarity
#   5. LCSubsequence word count
#
# Evaluation set:
#   796 generations
#   398 WITHOUT_METADATA
#   398 WITH_METADATA
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import torch

from rapidfuzz.distance import Jaro

print("=" * 70)
print("PHASE 5.14 — FLAN-T5 METRIC EVALUATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. LOAD CLEAN FLAN DATA
# ------------------------------------------------------------

EVAL_FILE = "/kaggle/working/flan_796_evaluation.csv"

df = pd.read_csv(EVAL_FILE)

print("\nLoaded:")
print(EVAL_FILE)

print(f"Rows: {len(df)}")

print("\nCondition counts:")
print(df["condition"].value_counts())

assert len(df) == 796
assert df["condition"].value_counts()["WITHOUT_METADATA"] == 398
assert df["condition"].value_counts()["WITH_METADATA"] == 398


# ------------------------------------------------------------
# 2. BASIC TEXT CLEANING
# ------------------------------------------------------------

def clean_text(x):
    if pd.isna(x):
        return ""
    
    x = str(x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()


df["reference"] = df["ground_truth_answer"].apply(clean_text)
df["candidate"] = df["answer_output"].apply(clean_text)

assert df["reference"].str.len().gt(0).all()
assert df["candidate"].str.len().gt(0).all()

print("\n✓ All evaluated rows contain reference and candidate text")


# ============================================================
# 3. JARO
# ============================================================

print("\n" + "=" * 70)
print("3. CALCULATING JARO")
print("=" * 70)

df["jaro"] = [
    Jaro.normalized_similarity(candidate, reference)
    for candidate, reference
    in zip(df["candidate"], df["reference"])
]

print("✓ Jaro calculated")


# ============================================================
# 4. CHARACTER-LEVEL LCS
# ============================================================

print("\n" + "=" * 70)
print("4. CALCULATING CHARACTER-LEVEL LCS")
print("=" * 70)

def lcs_length_chars(a, b):
    """
    Character-level Longest Common Subsequence.
    Memory-efficient two-row dynamic programming.
    """
    
    if not a or not b:
        return 0
    
    # Keep shorter string in b to reduce memory
    if len(a) < len(b):
        a, b = b, a
    
    prev = [0] * (len(b) + 1)
    
    for ca in a:
        curr = [0] * (len(b) + 1)
        
        for j, cb in enumerate(b, start=1):
            if ca == cb:
                curr[j] = prev[j - 1] + 1
            else:
                curr[j] = max(prev[j], curr[j - 1])
        
        prev = curr
    
    return prev[-1]


def lcs_similarity(a, b):
    """
    LCS similarity normalized by maximum text length.
    """
    
    if not a or not b:
        return 0.0
    
    lcs = lcs_length_chars(a, b)
    return lcs / max(len(a), len(b))


df["lcsubsequence_similarity"] = [
    lcs_similarity(candidate, reference)
    for candidate, reference
    in zip(df["candidate"], df["reference"])
]

print("✓ Character-level LCS calculated")


# ============================================================
# 5. WORD-LEVEL LCS
# ============================================================

print("\n" + "=" * 70)
print("5. CALCULATING WORD-LEVEL LCS")
print("=" * 70)

def lcs_length_words(a, b):
    """
    Word-level Longest Common Subsequence.
    """
    
    a_words = a.split()
    b_words = b.split()
    
    if not a_words or not b_words:
        return 0
    
    if len(a_words) < len(b_words):
        a_words, b_words = b_words, a_words
    
    prev = [0] * (len(b_words) + 1)
    
    for wa in a_words:
        curr = [0] * (len(b_words) + 1)
        
        for j, wb in enumerate(b_words, start=1):
            if wa == wb:
                curr[j] = prev[j - 1] + 1
            else:
                curr[j] = max(prev[j], curr[j - 1])
        
        prev = curr
    
    return prev[-1]


df["lcsubsequence_word_count"] = [
    lcs_length_words(candidate, reference)
    for candidate, reference
    in zip(df["candidate"], df["reference"])
]

print("✓ Word-level LCS calculated")


# ============================================================
# 6. BERTSCORE
# ============================================================

print("\n" + "=" * 70)
print("6. CALCULATING BERTSCORE")
print("=" * 70)

print("Model: roberta-large")
print("Metric: F1")
print("Rescale with baseline: False")

# Install only if necessary
try:
    from bert_score import score as bertscore
except ImportError:
    print("\nInstalling bert-score...")
    !pip install -q bert-score
    from bert_score import score as bertscore


candidates = df["candidate"].tolist()
references = df["reference"].tolist()

device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"\nDevice: {device}")

P, R, F1 = bertscore(
    candidates,
    references,
    model_type="roberta-large",
    lang="en",
    device=device,
    batch_size=8,
    rescale_with_baseline=False,
    verbose=True
)

df["bertscore_precision"] = P.cpu().numpy()
df["bertscore_recall"] = R.cpu().numpy()
df["bertscore_f1"] = F1.cpu().numpy()

# Free GPU memory
del P, R, F1

if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("\n✓ BERTScore calculated")


# ============================================================
# 7. BARTSCORE
# ============================================================

print("\n" + "=" * 70)
print("7. CALCULATING BARTSCORE")
print("=" * 70)

print("Model: facebook/bart-large-cnn")
print("Direction: candidate → reference")
print("Direction: reference → candidate")
print("Final score: average of both directions")


from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

BART_MODEL = "facebook/bart-large-cnn"

bart_tokenizer = AutoTokenizer.from_pretrained(BART_MODEL)

bart_model = AutoModelForSeq2SeqLM.from_pretrained(
    BART_MODEL
)

bart_model = bart_model.to(device)
bart_model.eval()

print("✓ BART model loaded")


# ------------------------------------------------------------
# BART log-likelihood scorer
# ------------------------------------------------------------

def bart_score_batch(
    source_texts,
    target_texts,
    batch_size=4,
    max_length=1024
):
    """
    Computes average token log-probability of target
    conditioned on source.

    This is the log-likelihood form used for our
    BARTScore-style evaluation.
    """

    scores = []

    for start in range(0, len(source_texts), batch_size):

        batch_sources = source_texts[start:start + batch_size]
        batch_targets = target_texts[start:start + batch_size]

        inputs = bart_tokenizer(
            batch_sources,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=max_length
        ).to(device)

        targets = bart_tokenizer(
            batch_targets,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=max_length
        )

        labels = targets["input_ids"].to(device)

        # Ignore padding tokens in loss
        labels[labels == bart_tokenizer.pad_token_id] = -100

        with torch.no_grad():

            outputs = bart_model(
                input_ids=inputs["input_ids"],
                attention_mask=inputs["attention_mask"],
                labels=labels
            )

            logits = outputs.logits

            log_probs = torch.log_softmax(
                logits,
                dim=-1
            )

            safe_labels = labels.clone()
            safe_labels[safe_labels == -100] = 0

            token_log_probs = torch.gather(
                log_probs,
                2,
                safe_labels.unsqueeze(-1)
            ).squeeze(-1)

            valid_mask = labels != -100

            token_log_probs = (
                token_log_probs * valid_mask
            )

            lengths = valid_mask.sum(dim=1).clamp(min=1)

            sequence_scores = (
                token_log_probs.sum(dim=1)
                / lengths
            )

            scores.extend(
                sequence_scores.cpu().numpy().tolist()
            )

    return np.array(scores)


# ------------------------------------------------------------
# Candidate → Reference
# ------------------------------------------------------------

print("\nCalculating candidate → reference...")

bart_candidate_to_reference = bart_score_batch(
    candidates,
    references,
    batch_size=4,
    max_length=1024
)

print("✓ Candidate → reference complete")


# ------------------------------------------------------------
# Reference → Candidate
# ------------------------------------------------------------

print("\nCalculating reference → candidate...")

bart_reference_to_candidate = bart_score_batch(
    references,
    candidates,
    batch_size=4,
    max_length=1024
)

print("✓ Reference → candidate complete")


# ------------------------------------------------------------
# Bidirectional BARTScore
# ------------------------------------------------------------

df["bartscore_candidate_to_reference"] = (
    bart_candidate_to_reference
)

df["bartscore_reference_to_candidate"] = (
    bart_reference_to_candidate
)

df["bartscore"] = (
    bart_candidate_to_reference
    + bart_reference_to_candidate
) / 2


# Free BART model
del bart_model
del bart_tokenizer

if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("✓ BARTScore calculated")


# ============================================================
# 8. SAVE PER-QUESTION METRICS
# ============================================================

METRICS_FILE = "/kaggle/working/flan_796_question_metrics.csv"

df.to_csv(
    METRICS_FILE,
    index=False
)

print("\n" + "=" * 70)
print("PER-QUESTION METRICS SAVED")
print("=" * 70)

print(METRICS_FILE)


# ============================================================
# 9. TABLE-4 STYLE AGGREGATION
# ============================================================

metric_columns = [
    "bertscore_f1",
    "bartscore",
    "jaro",
    "lcsubsequence_similarity",
    "lcsubsequence_word_count"
]

summary = (
    df
    .groupby("condition")[metric_columns]
    .mean()
    .reset_index()
)

print("\n" + "=" * 70)
print("FLAN-T5 TABLE-4 STYLE RESULTS")
print("=" * 70)

print(
    summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# 10. SAVE FINAL TABLE
# ============================================================

FINAL_TABLE_FILE = "/kaggle/working/flan_796_final_table4.csv"

summary.to_csv(
    FINAL_TABLE_FILE,
    index=False
)

print("\n✓ Final Table-4-style results saved:")
print(FINAL_TABLE_FILE)

print("\n" + "=" * 70)
print("✓ PHASE 5.14 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 6.6 — PYTHIA FULL 800 GENERATION
# ============================================================

import torch
import pandas as pd
import re
import time
import gc

print("=" * 70)
print("PHASE 6.6 — PYTHIA FULL 800 GENERATION")
print("=" * 70)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

PYTHIA_MAX_NEW_TOKENS = 1024

# Stop generation when the model clearly enters
# a pathological repetition loop.
REPETITION_WINDOW = 30
REPETITION_THRESHOLD = 0.60

# ------------------------------------------------------------
# Verify input
# ------------------------------------------------------------

assert len(pythia_prompts) == 800, (
    f"Expected 800 prompts, got {len(pythia_prompts)}"
)

assert set(pythia_prompts["condition"]) == {
    "WITHOUT_METADATA",
    "WITH_METADATA"
}

print(f"\nTotal prompts: {len(pythia_prompts)}")
print(
    pythia_prompts["condition"]
    .value_counts()
    .to_string()
)

# ------------------------------------------------------------
# Repetition detector
# ------------------------------------------------------------

def repetition_ratio(text, window=REPETITION_WINDOW):

    words = text.split()

    if len(words) < window * 2:
        return 0.0

    recent = words[-window:]

    # Compare the latest window against the preceding window
    previous = words[-2 * window:-window]

    if len(previous) != len(recent):
        return 0.0

    matches = sum(
        a.lower() == b.lower()
        for a, b in zip(previous, recent)
    )

    return matches / window


def has_repetition(text):

    ratio = repetition_ratio(text)

    return ratio >= REPETITION_THRESHOLD


# ------------------------------------------------------------
# Generation function
# ------------------------------------------------------------

def generate_pythia_full(prompt):

    inputs = pythia_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs["input_ids"].to(pythia_model.device)
    attention_mask = inputs["attention_mask"].to(pythia_model.device)

    with torch.no_grad():

        output_ids = pythia_model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,

            max_new_tokens=PYTHIA_MAX_NEW_TOKENS,

            # Deterministic baseline
            do_sample=False,

            # Prevent padding warnings
            pad_token_id=pythia_tokenizer.pad_token_id,
            eos_token_id=pythia_tokenizer.eos_token_id
        )

    generated_ids = output_ids[0][input_ids.shape[1]:]

    generated_text = pythia_tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()

    return generated_text, len(generated_ids)


# ------------------------------------------------------------
# Run all 800
# ------------------------------------------------------------

results = []

start_time = time.time()

for i, row in pythia_prompts.iterrows():

    question_id = row["question_id"]
    condition = row["condition"]

    print(
        f"\n[{len(results)+1:03d}/800] "
        f"{question_id} | {condition}"
    )

    try:

        answer, generated_tokens = generate_pythia_full(
            row["prompt"]
        )

        # --------------------------------------------
        # Detect pathological repetition AFTER generation
        # --------------------------------------------

        repeated = has_repetition(answer)

        if repeated:

            status = "repetition_detected"

        elif generated_tokens >= PYTHIA_MAX_NEW_TOKENS:

            status = "max_tokens_reached"

        else:

            status = "success"

        print(
            f"tokens={generated_tokens} | "
            f"status={status}"
        )

        print(
            "preview:",
            answer[:250].replace("\n", " ")
        )

        results.append({

            "question_id": question_id,
            "condition": condition,

            "generated_answer": answer,
            "generated_tokens": generated_tokens,

            "repetition_detected": repeated,

            "status": status

        })

    except Exception as e:

        print(
            f"ERROR: {type(e).__name__}: {e}"
        )

        results.append({

            "question_id": question_id,
            "condition": condition,

            "generated_answer": "",

            "generated_tokens": 0,

            "repetition_detected": False,

            "status": (
                f"error: "
                f"{type(e).__name__}: {e}"
            )

        })

    # --------------------------------------------
    # Periodic checkpoint
    # --------------------------------------------

    if len(results) % 50 == 0:

        checkpoint = pd.DataFrame(results)

        checkpoint.to_csv(
            "/kaggle/working/"
            "pythia_800_generation_checkpoint.csv",
            index=False
        )

        elapsed = time.time() - start_time

        print(
            f"\nCHECKPOINT SAVED "
            f"({len(results)}/800)"
        )

        print(
            f"Elapsed: {elapsed/60:.2f} minutes"
        )


# ------------------------------------------------------------
# Final dataframe
# ------------------------------------------------------------

pythia_generations = pd.DataFrame(results)

OUTPUT_FILE = (
    "/kaggle/working/"
    "pythia_800_generations.csv"
)

pythia_generations.to_csv(
    OUTPUT_FILE,
    index=False
)


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

elapsed = time.time() - start_time

print("\n" + "=" * 70)
print("PYTHIA GENERATION SUMMARY")
print("=" * 70)

print(f"Total rows          : {len(pythia_generations)}")

print(
    f"Successful          : "
    f"{(pythia_generations['status'] == 'success').sum()}"
)

print(
    f"Repetition detected : "
    f"{pythia_generations['repetition_detected'].sum()}"
)

print(
    f"Max tokens reached  : "
    f"{(pythia_generations['status'] == 'max_tokens_reached').sum()}"
)

print(
    f"Errors              : "
    f"{pythia_generations['status'].astype(str).str.startswith('error').sum()}"
)

print(
    f"\nMean generated tokens: "
    f"{pythia_generations['generated_tokens'].mean():.2f}"
)

print(
    f"Median generated tokens: "
    f"{pythia_generations['generated_tokens'].median():.2f}"
)

print(
    "\nSTATUS COUNTS:"
)

print(
    pythia_generations["status"]
    .value_counts()
    .to_string()
)

print(
    "\nCONDITION × STATUS:"
)

print(
    pd.crosstab(
        pythia_generations["condition"],
        pythia_generations["status"]
    ).to_string()
)

print(
    f"\n✓ Saved: {OUTPUT_FILE}"
)

print(
    f"✓ Runtime: {elapsed/60:.2f} minutes"
)

print("\n" + "=" * 70)
print("✓ PHASE 6.6 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 6.8 — PYTHIA-1.4B FINAL GENERATION
# 128 MAX NEW TOKENS
# ============================================================

import os
import gc
import re
import time
import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

MODEL_NAME = "EleutherAI/pythia-1.4b"

# Paper configuration
CONTEXT_SIZE = 2048

# IMPORTANT:
# Paper specifies 1024 as maximum output tokens.
# We use 128 here because Pythia does not reliably emit EOS
# and 1024 causes extremely long transcript-like generations.
MAX_NEW_TOKENS = 128

# Decoding configuration selected from Phase 6.4–6.7 diagnostics
DO_SAMPLE = True
TEMPERATURE = 0.7
TOP_P = 0.9
REPETITION_PENALTY = 1.1

# Checkpoint frequency
CHECKPOINT_EVERY = 50

# Input files
PROMPTS_FILE = "/kaggle/working/pythia_800_prompts.csv"

# Output files
CHECKPOINT_FILE = "/kaggle/working/pythia_800_generation_checkpoint_128.csv"
FINAL_FILE = "/kaggle/working/pythia_800_generations_128.csv"

# ------------------------------------------------------------
# 2. LOAD PROMPTS
# ------------------------------------------------------------

print("=" * 70)
print("PHASE 6.8 — PYTHIA FINAL GENERATION")
print("=" * 70)

prompts_df = pd.read_csv(PROMPTS_FILE)

print(f"Prompts loaded: {len(prompts_df)}")
print(f"Columns: {list(prompts_df.columns)}")

assert len(prompts_df) == 800, (
    f"Expected 800 prompts, found {len(prompts_df)}"
)

assert set(prompts_df["condition"].unique()) == {
    "WITHOUT_METADATA",
    "WITH_METADATA"
}

print("\nCondition counts:")
print(prompts_df["condition"].value_counts())

# ------------------------------------------------------------
# 3. LOAD TOKENIZER
# ------------------------------------------------------------

print("\nLoading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    revision="main"
)

# Pythia has a padding token, but set it explicitly for safety
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("Tokenizer loaded.")
print("EOS token:", repr(tokenizer.eos_token))
print("PAD token:", repr(tokenizer.pad_token))
print("Vocab size:", tokenizer.vocab_size)

# ------------------------------------------------------------
# 4. LOAD MODEL
# ------------------------------------------------------------

print("\nLoading Pythia-1.4B...")

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16 if device.type == "cuda" else torch.float32,
    revision="main"
)

model = model.to(device)
model.eval()

print("Model loaded.")
print("Parameters:", sum(p.numel() for p in model.parameters()))
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print(
        "GPU memory allocated:",
        round(torch.cuda.memory_allocated() / 1024**3, 2),
        "GB"
    )

# ------------------------------------------------------------
# 5. GENERATION FUNCTION
# ------------------------------------------------------------

def generate_answer(prompt):
    """
    Generate one Pythia response.

    We preserve the complete raw generation and separately
    extract the answer portion.

    IMPORTANT:
    The 128-token limit is an implementation decision made
    because Pythia frequently fails to terminate naturally.
    """

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False
    )

    input_ids = inputs["input_ids"].to(device)
    attention_mask = inputs["attention_mask"].to(device)

    input_token_count = input_ids.shape[1]

    # Safety check against Pythia context window
    if input_token_count >= CONTEXT_SIZE:
        return {
            "generated_answer": "",
            "raw_generated_text": "",
            "generated_tokens": 0,
            "input_tokens": input_token_count,
            "eos_generated": False,
            "extraction_status": "INPUT_TOO_LONG",
            "status": "error",
            "error": f"Input has {input_token_count} tokens"
        }

    try:

        with torch.no_grad():

            outputs = model.generate(
                input_ids=input_ids,
                attention_mask=attention_mask,

                # We only count newly generated tokens
                max_new_tokens=MAX_NEW_TOKENS,

                # Sampling configuration
                do_sample=DO_SAMPLE,
                temperature=TEMPERATURE,
                top_p=TOP_P,
                repetition_penalty=REPETITION_PENALTY,

                # Stop token
                eos_token_id=tokenizer.eos_token_id,
                pad_token_id=tokenizer.pad_token_id,

                # Prevent HF warning / unwanted behavior
                use_cache=True
            )

        # ----------------------------------------------------
        # Extract ONLY newly generated tokens
        # ----------------------------------------------------

        generated_ids = outputs[0][input_ids.shape[1]:]

        generated_tokens = len(generated_ids)

        raw_generated_text = tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        ).strip()

        # ----------------------------------------------------
        # Check whether EOS was actually generated
        # ----------------------------------------------------

        eos_generated = (
            tokenizer.eos_token_id in generated_ids.tolist()
        )

        # ----------------------------------------------------
        # Extract answer before newly generated Question block
        #
        # Pythia often continues by generating another QA pair.
        # We stop extraction at the first new "Question:" marker.
        # ----------------------------------------------------

        generated_answer = raw_generated_text

        extraction_status = "FULL_GENERATION"

        question_patterns = [
            r"\nQuestion\s*:",
            r"\nQ\s*:",
            r"\nQuestion\s*-\s*"
        ]

        earliest_position = None

        for pattern in question_patterns:

            match = re.search(
                pattern,
                generated_answer,
                flags=re.IGNORECASE
            )

            if match:

                if (
                    earliest_position is None
                    or match.start() < earliest_position
                ):
                    earliest_position = match.start()

        if earliest_position is not None:

            generated_answer = generated_answer[
                :earliest_position
            ].strip()

            extraction_status = "TRUNCATED_AT_NEW_QUESTION"

        # ----------------------------------------------------
        # Remove obvious repeated prompt suffix if generated
        # ----------------------------------------------------

        if "\nAnswer:" in generated_answer:

            parts = re.split(
                r"\nAnswer\s*:",
                generated_answer,
                maxsplit=1,
                flags=re.IGNORECASE
            )

            if len(parts) > 1:

                # Only use the latter part if the generated text
                # actually appears to reproduce the QA structure.
                possible_answer = parts[-1].strip()

                if len(possible_answer) > 0:
                    generated_answer = possible_answer
                    extraction_status = (
                        extraction_status + "|ANSWER_SECTION"
                    )

        # ----------------------------------------------------
        # Basic status
        # ----------------------------------------------------

        if generated_tokens == 0:

            status = "empty"

        else:

            status = "success"

        return {
            "generated_answer": generated_answer,
            "raw_generated_text": raw_generated_text,
            "generated_tokens": generated_tokens,
            "input_tokens": input_token_count,
            "eos_generated": eos_generated,
            "extraction_status": extraction_status,
            "status": status,
            "error": ""
        }

    except Exception as e:

        return {
            "generated_answer": "",
            "raw_generated_text": "",
            "generated_tokens": 0,
            "input_tokens": input_token_count,
            "eos_generated": False,
            "extraction_status": "GENERATION_ERROR",
            "status": "error",
            "error": repr(e)
        }


# ------------------------------------------------------------
# 6. RUN 800 GENERATIONS
# ------------------------------------------------------------

results = []

start_time = time.time()

print("\n" + "=" * 70)
print("STARTING GENERATION")
print("=" * 70)

print(f"Total prompts: {len(prompts_df)}")
print(f"Max new tokens: {MAX_NEW_TOKENS}")
print(f"Temperature: {TEMPERATURE}")
print(f"Top-p: {TOP_P}")
print(f"Repetition penalty: {REPETITION_PENALTY}")
print("=" * 70)

for idx, row in prompts_df.iterrows():

    question_id = row["question_id"]
    condition = row["condition"]
    prompt = row["prompt"]

    print(
        f"\n[{idx + 1:03d}/800] "
        f"{question_id} | {condition}",
        flush=True
    )

    result = generate_answer(prompt)

    record = {
        "question_id": question_id,
        "condition": condition,
        "generated_answer": result["generated_answer"],
        "raw_generated_text": result["raw_generated_text"],
        "generated_tokens": result["generated_tokens"],
        "input_tokens": result["input_tokens"],
        "eos_generated": result["eos_generated"],
        "extraction_status": result["extraction_status"],
        "status": result["status"],
        "error": result["error"]
    }

    results.append(record)

    print(
        f"  status={result['status']} "
        f"| generated_tokens={result['generated_tokens']} "
        f"| input_tokens={result['input_tokens']} "
        f"| eos={result['eos_generated']} "
        f"| extraction={result['extraction_status']}",
        flush=True
    )

    # --------------------------------------------------------
    # CHECKPOINT
    # --------------------------------------------------------

    if (
        (idx + 1) % CHECKPOINT_EVERY == 0
        or idx == len(prompts_df) - 1
    ):

        checkpoint_df = pd.DataFrame(results)

        checkpoint_df.to_csv(
            CHECKPOINT_FILE,
            index=False
        )

        elapsed = time.time() - start_time

        print("\n" + "-" * 70)
        print(
            f"CHECKPOINT SAVED: {idx + 1}/800"
        )
        print(
            f"Elapsed: {elapsed / 60:.2f} minutes"
        )
        print(
            f"File: {CHECKPOINT_FILE}"
        )
        print("-" * 70)

    # Free temporary CUDA memory
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ------------------------------------------------------------
# 7. SAVE FINAL FILE
# ------------------------------------------------------------

generation_df = pd.DataFrame(results)

generation_df.to_csv(
    FINAL_FILE,
    index=False
)

elapsed_total = time.time() - start_time

print("\n" + "=" * 70)
print("GENERATION COMPLETE")
print("=" * 70)

print(f"Rows generated: {len(generation_df)}")
print(f"Total time: {elapsed_total / 60:.2f} minutes")
print(f"Final file: {FINAL_FILE}")

# ------------------------------------------------------------
# 8. SUMMARY
# ------------------------------------------------------------

print("\nSTATUS COUNTS")
print("-" * 50)
print(
    generation_df["status"]
    .value_counts(dropna=False)
)

print("\nEXTRACTION STATUS")
print("-" * 50)
print(
    generation_df["extraction_status"]
    .value_counts(dropna=False)
)

print("\nCONDITION × STATUS")
print("-" * 50)
print(
    pd.crosstab(
        generation_df["condition"],
        generation_df["status"]
    )
)

print("\nCONDITION × EOS")
print("-" * 50)
print(
    pd.crosstab(
        generation_df["condition"],
        generation_df["eos_generated"]
    )
)

print("\nGENERATED TOKEN STATISTICS")
print("-" * 50)

print(
    generation_df["generated_tokens"]
    .describe()
)

print("\nMEAN TOKENS BY CONDITION")
print("-" * 50)

print(
    generation_df
    .groupby("condition")["generated_tokens"]
    .agg(["count", "mean", "median", "min", "max"])
)

# ------------------------------------------------------------
# 9. CHECK FOR EXACTLY 800 ROWS
# ------------------------------------------------------------

assert len(generation_df) == 800

assert (
    generation_df["question_id"].nunique() == 400
)

assert (
    generation_df["condition"].nunique() == 2
)

print("\n✓ 800 generations confirmed.")
print("✓ 400 unique questions confirmed.")
print("✓ Both metadata conditions confirmed.")
print("✓ Generation file saved successfully.")

# ------------------------------------------------------------
# 10. SHOW SAMPLE OUTPUTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SAMPLE GENERATED ANSWERS")
print("=" * 70)

for _, sample in generation_df.head(10).iterrows():

    print("\n" + "-" * 70)
    print(
        sample["question_id"],
        "|",
        sample["condition"]
    )

    print(
        "Tokens:",
        sample["generated_tokens"]
    )

    print(
        "EOS:",
        sample["eos_generated"]
    )

    print(
        "Extraction:",
        sample["extraction_status"]
    )

    print("\nANSWER:")
    print(sample["generated_answer"][:1000])

print("\n" + "=" * 70)
print("DONE")
print("=" * 70)
```

```python
# ============================================================
# PYTHIA-1.4B FINAL EVALUATION
# ============================================================
#
# Evaluates:
#   1. BERTScore
#   2. BARTScore
#   3. Jaro Similarity
#   4. LCSubsequence Similarity
#   5. LCSubsequence Word Count
#
# Conditions:
#   WITHOUT_METADATA = 400
#   WITH_METADATA    = 400
#
# Generation setup:
#   Pythia-1.4B
#   max_new_tokens = 128
#   temperature = 0.7
#   top_p = 0.9
#   repetition_penalty = 1.1
# ============================================================

import os
import gc
import numpy as np
import pandas as pd
import torch
from tqdm.auto import tqdm

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

GENERATION_FILE = (
    "/kaggle/working/pythia_800_generations_128.csv"
)

GROUND_TRUTH_FILE = (
    "/kaggle/input/datasets/ramnarenkakulapati/"
    "emo-bros/ground_truth_400.csv"
)

QUESTION_METRICS_FILE = (
    "/kaggle/working/pythia_1_4b_question_metrics.csv"
)

TABLE4_FILE = (
    "/kaggle/working/pythia_1_4b_final_table4.csv"
)

BERT_BATCH_SIZE = 8
BART_BATCH_SIZE = 4

BERT_MODEL = "roberta-large"
BART_MODEL = "facebook/bart-large-cnn"


# ============================================================
# 1. LOAD GENERATIONS
# ============================================================

print("=" * 110)
print("PYTHIA-1.4B FINAL EVALUATION")
print("=" * 110)

if not os.path.exists(GENERATION_FILE):
    raise FileNotFoundError(
        f"Generation file not found:\n{GENERATION_FILE}"
    )

gen_df = pd.read_csv(GENERATION_FILE)

print("Generation file:", GENERATION_FILE)
print("Generation shape:", gen_df.shape)
print("Generation columns:")
print(list(gen_df.columns))


# ============================================================
# 2. VALIDATE GENERATIONS
# ============================================================

if "status" not in gen_df.columns:
    raise ValueError(
        "Generation file does not contain 'status'."
    )

successful = gen_df[
    gen_df["status"]
    .astype(str)
    .str.lower()
    .isin(["success", "ok"])
].copy()

print("\nSuccessful rows:", len(successful))

if len(successful) != 800:
    raise ValueError(
        f"Expected 800 successful generations, "
        f"but found {len(successful)}."
    )

# ------------------------------------------------------------
# Condition counts
# ------------------------------------------------------------

condition_counts = (
    successful["condition"]
    .value_counts()
    .to_dict()
)

print("\nCondition counts:", condition_counts)

if condition_counts.get("WITHOUT_METADATA", 0) != 400:
    raise ValueError(
        "Expected 400 WITHOUT_METADATA rows."
    )

if condition_counts.get("WITH_METADATA", 0) != 400:
    raise ValueError(
        "Expected 400 WITH_METADATA rows."
    )


# ============================================================
# 3. LOAD GROUND TRUTH
# ============================================================

print("\nLoading ground truth...")

gt_df = pd.read_csv(
    GROUND_TRUTH_FILE
)

print("Ground-truth shape:", gt_df.shape)

required_gt_columns = [
    "question_id",
    "company",
    "question",
    "ground_truth_answer"
]

for col in required_gt_columns:
    if col not in gt_df.columns:
        raise ValueError(
            f"Ground truth is missing column: {col}"
        )

print(
    "Ground-truth question count:",
    gt_df["question_id"].nunique()
)


# ============================================================
# 4. MERGE GENERATIONS WITH GROUND TRUTH
# ============================================================

# Keep only the columns we need from ground truth.
gt_small = gt_df[
    [
        "question_id",
        "company",
        "question",
        "ground_truth_answer"
    ]
].copy()

# Check uniqueness
if gt_small["question_id"].duplicated().any():
    raise ValueError(
        "Ground truth contains duplicate question_id values."
    )

successful = successful.merge(
    gt_small,
    on="question_id",
    how="left",
    validate="many_to_one"
)

print("\nAfter ground-truth merge:", successful.shape)

# Check unmatched questions
missing_gt = successful[
    successful["ground_truth_answer"].isna()
]

print(
    "Generations without ground truth:",
    len(missing_gt)
)

if len(missing_gt) > 0:
    print(
        missing_gt[
            ["question_id", "condition"]
        ].head(20)
    )

    raise ValueError(
        "Some Pythia generations could not be matched "
        "to the ground truth."
    )


# ============================================================
# 5. PREPARE PREDICTIONS + REFERENCES
# ============================================================

# Pythia generation code used generated_answer.
# Keep a fallback to raw_generated_text in case one answer
# is missing after generation.

if "generated_answer" not in successful.columns:
    raise ValueError(
        "Missing 'generated_answer' column."
    )

successful["prediction"] = (
    successful["generated_answer"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------
# Fallback to raw generated output if needed
# ------------------------------------------------------------

if "raw_generated_text" in successful.columns:

    missing_mask = (
        successful["prediction"] == ""
    )

    missing_count = missing_mask.sum()

    print(
        "\nEmpty generated_answer values BEFORE fallback:",
        missing_count
    )

    if missing_count > 0:

        fallback = (
            successful.loc[
                missing_mask,
                "raw_generated_text"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
        )

        successful.loc[
            missing_mask,
            "prediction"
        ] = fallback

        print(
            f"Recovered {missing_count} "
            "prediction(s) from raw_generated_text."
        )

else:

    print(
        "\nraw_generated_text column not present."
    )

# ------------------------------------------------------------
# Reference answers
# ------------------------------------------------------------

successful["reference"] = (
    successful["ground_truth_answer"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------
# Final empty check
# ------------------------------------------------------------

empty_predictions = (
    successful["prediction"]
    .fillna("")
    .astype(str)
    .str.strip()
    == ""
)

empty_count = empty_predictions.sum()

print(
    "Empty predictions AFTER fallback:",
    empty_count
)

if empty_count > 0:

    print(
        "\nEmpty prediction rows:"
    )

    display(
        successful.loc[
            empty_predictions,
            [
                "question_id",
                "condition",
                "company"
            ]
        ]
    )

    # We cannot score an empty answer with BERTScore.
    # Remove only truly empty generations.
    successful = successful[
        ~empty_predictions
    ].copy()

    print(
        f"\nRemoved {empty_count} empty prediction(s) "
        "from metric evaluation."
    )

print(
    "\nRows remaining for evaluation:",
    len(successful)
)


# ============================================================
# 6. RELEASE PYTHIA GPU MEMORY
# ============================================================

print("\nReleasing Pythia GPU memory...")

try:
    del model
except:
    pass

try:
    del tokenizer
except:
    pass

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

print("GPU memory released.")


# ============================================================
# 7. BERTSCORE
# ============================================================

print("\n" + "=" * 110)
print("1. BERTScore")
print("=" * 110)

try:
    from bert_score import score as bert_score
except ImportError:
    raise ImportError(
        "bert-score is not installed.\n"
        "Run:\n"
        "!pip install -q bert-score"
    )

bert_device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

predictions_all = (
    successful["prediction"].tolist()
)

references_all = (
    successful["reference"].tolist()
)

print("Model:", BERT_MODEL)
print("Device:", bert_device)
print("Number of predictions:", len(predictions_all))

P, R, F1 = bert_score(
    predictions_all,
    references_all,
    model_type=BERT_MODEL,
    lang="en",
    batch_size=BERT_BATCH_SIZE,
    device=bert_device,
    rescale_with_baseline=False,
    verbose=True,
)

successful["BERTScore"] = (
    F1.detach()
    .cpu()
    .numpy()
)

print(
    "\nMean BERTScore:",
    successful["BERTScore"].mean()
)

# Free BERT memory
gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()


# ============================================================
# 8. BARTSCORE
# ============================================================

print("\n" + "=" * 110)
print("2. BARTScore")
print("=" * 110)

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
)

import torch.nn.functional as F

bart_device = (
    torch.device("cuda")
    if torch.cuda.is_available()
    else torch.device("cpu")
)

print("Model:", BART_MODEL)
print("Device:", bart_device)

bart_tokenizer = AutoTokenizer.from_pretrained(
    BART_MODEL
)

bart_model = AutoModelForSeq2SeqLM.from_pretrained(
    BART_MODEL
)

bart_model = bart_model.to(bart_device)
bart_model.eval()


# ------------------------------------------------------------
# BART directional scorer
# ------------------------------------------------------------

def bart_directional_score(
    sources,
    targets,
    batch_size=4,
    max_length=1024
):

    scores = []

    for start in tqdm(
        range(0, len(sources), batch_size),
        desc="BARTScore"
    ):

        src_batch = sources[
            start:start + batch_size
        ]

        tgt_batch = targets[
            start:start + batch_size
        ]

        src = bart_tokenizer(
            src_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        )

        tgt = bart_tokenizer(
            tgt_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        )

        src_input_ids = (
            src["input_ids"]
            .to(bart_device)
        )

        src_attention_mask = (
            src["attention_mask"]
            .to(bart_device)
        )

        tgt_input_ids = (
            tgt["input_ids"]
            .to(bart_device)
        )

        with torch.no_grad():

            outputs = bart_model(
                input_ids=src_input_ids,
                attention_mask=src_attention_mask,
                labels=tgt_input_ids
            )

        logits = outputs.logits

        vocab_size = logits.shape[-1]

        token_losses = F.cross_entropy(
            logits.reshape(-1, vocab_size),
            tgt_input_ids.reshape(-1),
            ignore_index=bart_tokenizer.pad_token_id,
            reduction="none"
        )

        token_losses = token_losses.reshape(
            tgt_input_ids.shape
        )

        mask = (
            tgt_input_ids
            != bart_tokenizer.pad_token_id
        )

        token_count = (
            mask.sum(dim=1)
            .clamp(min=1)
        )

        batch_scores = (
            -token_losses.sum(dim=1)
            / token_count
        )

        scores.extend(
            batch_scores
            .detach()
            .cpu()
            .numpy()
            .tolist()
        )

        # Cleanup
        del src
        del tgt
        del src_input_ids
        del src_attention_mask
        del tgt_input_ids
        del outputs
        del logits
        del token_losses
        del token_count
        del batch_scores

    return np.asarray(
        scores,
        dtype=np.float64
    )


# ------------------------------------------------------------
# BART directions
# ------------------------------------------------------------

refs = references_all
hyps = predictions_all

print("\nReference -> Hypothesis")

bart_P = bart_directional_score(
    refs,
    hyps,
    batch_size=BART_BATCH_SIZE,
    max_length=1024
)

print("\nHypothesis -> Reference")

bart_R = bart_directional_score(
    hyps,
    refs,
    batch_size=BART_BATCH_SIZE,
    max_length=1024
)

bart_F = (
    bart_P + bart_R
) / 2.0

successful["BARTScore"] = bart_F

print(
    "\nMean BARTScore:",
    successful["BARTScore"].mean()
)


# ------------------------------------------------------------
# Free BART
# ------------------------------------------------------------

del bart_model
del bart_tokenizer

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()


# ============================================================
# 9. JARO SIMILARITY
# ============================================================

print("\n" + "=" * 110)
print("3. Jaro Similarity")
print("=" * 110)

try:
    from rapidfuzz.distance import Jaro
except ImportError:
    raise ImportError(
        "rapidfuzz is not installed.\n"
        "Run:\n"
        "!pip install -q rapidfuzz"
    )


def jaro_score(
    prediction,
    reference
):

    return Jaro.normalized_similarity(
        str(prediction),
        str(reference)
    )


successful["Jaro Similarity"] = [
    jaro_score(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

print(
    "Mean Jaro:",
    successful["Jaro Similarity"].mean()
)


# ============================================================
# 10. LC SUBSEQUENCE METRICS
# ============================================================

print("\n" + "=" * 110)
print("4. LCSubsequence Similarity")
print("5. LCSubsequence Word Count")
print("=" * 110)


def lcs_length(a, b):

    if not a or not b:
        return 0

    # Make b shorter
    if len(b) > len(a):
        a, b = b, a

    previous = [
        0
    ] * (
        len(b) + 1
    )

    for x in a:

        current = [
            0
        ] * (
            len(b) + 1
        )

        for j, y in enumerate(
            b,
            start=1
        ):

            if x == y:

                current[j] = (
                    previous[j - 1] + 1
                )

            else:

                current[j] = max(
                    previous[j],
                    current[j - 1]
                )

        previous = current

    return previous[-1]


def lcs_character_similarity(
    prediction,
    reference
):

    prediction = str(prediction)
    reference = str(reference)

    if not prediction and not reference:
        return 1.0

    if not prediction or not reference:
        return 0.0

    lcs = lcs_length(
        list(prediction),
        list(reference)
    )

    return (
        lcs /
        max(
            len(prediction),
            len(reference)
        )
    )


def lcs_word_count(
    prediction,
    reference
):

    pred_words = str(
        prediction
    ).split()

    ref_words = str(
        reference
    ).split()

    return lcs_length(
        pred_words,
        ref_words
    )


successful[
    "LCSubsequence Similarity"
] = [
    lcs_character_similarity(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

successful[
    "LCSubsequence Word Count"
] = [
    lcs_word_count(
        p,
        r
    )
    for p, r in zip(
        successful["prediction"],
        successful["reference"]
    )
]

print(
    "Mean LCS similarity:",
    successful[
        "LCSubsequence Similarity"
    ].mean()
)

print(
    "Mean LCS word count:",
    successful[
        "LCSubsequence Word Count"
    ].mean()
)


# ============================================================
# 11. QUESTION-LEVEL METRICS
# ============================================================

metric_columns = [
    "question_id",
    "condition",
    "company",
    "question",
    "prediction",
    "reference",
    "generated_tokens",
    "eos_generated",
    "extraction_status",
    "BERTScore",
    "BARTScore",
    "Jaro Similarity",
    "LCSubsequence Similarity",
    "LCSubsequence Word Count",
]

question_metrics = successful[
    [
        c
        for c in metric_columns
        if c in successful.columns
    ]
].copy()

question_metrics.to_csv(
    QUESTION_METRICS_FILE,
    index=False
)

print(
    "\nSaved question-level metrics:"
)
print(QUESTION_METRICS_FILE)


# ============================================================
# 12. FINAL TABLE-4 STYLE AGGREGATION
# ============================================================

rows = []

for condition in [
    "WITHOUT_METADATA",
    "WITH_METADATA"
]:

    subset = successful[
        successful["condition"]
        == condition
    ]

    print(
        f"\n{condition}: {len(subset)} rows"
    )

    # We expect 400 unless an actual empty prediction had to
    # be removed.
    if len(subset) not in [399, 400]:
        raise ValueError(
            f"Unexpected row count for {condition}: "
            f"{len(subset)}"
        )

    rows.append({

        "Model":
            "Pythia-1.4b",

        "Condition":
            (
                "Without Metadata"
                if condition
                == "WITHOUT_METADATA"
                else
                "With Metadata"
            ),

        "BERTScore":
            subset[
                "BERTScore"
            ].mean(),

        "BARTScore":
            subset[
                "BARTScore"
            ].mean(),

        "Jaro Similarity":
            subset[
                "Jaro Similarity"
            ].mean(),

        "LCSubsequence Similarity":
            subset[
                "LCSubsequence Similarity"
            ].mean(),

        "LCSubsequence Word Count":
            subset[
                "LCSubsequence Word Count"
            ].mean(),
    })


table4 = pd.DataFrame(rows)


# ============================================================
# 13. DISPLAY FINAL RESULTS
# ============================================================

display_table = table4.copy()

display_table[
    "BERTScore"
] = display_table[
    "BERTScore"
].round(2)

display_table[
    "BARTScore"
] = display_table[
    "BARTScore"
].round(2)

display_table[
    "Jaro Similarity"
] = display_table[
    "Jaro Similarity"
].round(2)

display_table[
    "LCSubsequence Similarity"
] = display_table[
    "LCSubsequence Similarity"
].round(2)

display_table[
    "LCSubsequence Word Count"
] = (
    display_table[
        "LCSubsequence Word Count"
    ]
    .round(0)
    .astype(int)
)


print("\n\n")
print("=" * 110)
print("FINAL TABLE 4 — Pythia-1.4b")
print("=" * 110)

print(
    display_table.to_string(
        index=False
    )
)


# ============================================================
# 14. SAVE FINAL TABLE
# ============================================================

table4.to_csv(
    TABLE4_FILE,
    index=False
)

print("\nSaved:")
print(QUESTION_METRICS_FILE)
print(TABLE4_FILE)

print("\n")
print("=" * 110)
print("PYTHIA-1.4B EVALUATION COMPLETE")
print("=" * 110)
```

```python
# ============================================================
# BLOOMZ-7B1 — FINAL 800 GENERATION RUN
# ============================================================
#
# Model:
#   bigscience/bloomz-7b1
#
# Conditions:
#   WITHOUT_METADATA = 400
#   WITH_METADATA    = 400
#
# Generation:
#   max_new_tokens   = 128
#   greedy decoding
#   8-bit loading
#
# Output:
#   /kaggle/working/bloomz_800_generations_final.csv
# ============================================================

import os
import gc
import re
import time
import torch
import pandas as pd

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig
)

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

MODEL_NAME = "bigscience/bloomz-7b1"

PROMPTS_FILE = (
    "/kaggle/working/pythia_800_prompts.csv"
)

OUTPUT_FILE = (
    "/kaggle/working/bloomz_800_generations_final.csv"
)

CHECKPOINT_FILE = (
    "/kaggle/working/bloomz_800_generation_checkpoint_final.csv"
)

MAX_NEW_TOKENS = 128

# BLOOMZ behaved well with greedy decoding
DO_SAMPLE = False

BATCH_SIZE = 2

CHECKPOINT_EVERY = 50

# ------------------------------------------------------------
# LOAD PROMPTS
# ------------------------------------------------------------

print("=" * 110)
print("BLOOMZ-7B1 FINAL GENERATION")
print("=" * 110)

if not os.path.exists(PROMPTS_FILE):
    raise FileNotFoundError(
        f"Prompt file not found:\n{PROMPTS_FILE}"
    )

prompts_df = pd.read_csv(PROMPTS_FILE)

print("Prompt file:", PROMPTS_FILE)
print("Prompt shape:", prompts_df.shape)
print("Columns:", list(prompts_df.columns))

if len(prompts_df) != 800:
    raise ValueError(
        f"Expected 800 prompts, found {len(prompts_df)}"
    )

if "question_id" not in prompts_df.columns:
    raise ValueError("Missing question_id column.")

if "condition" not in prompts_df.columns:
    raise ValueError("Missing condition column.")

if "prompt" not in prompts_df.columns:
    raise ValueError("Missing prompt column.")

print("\nCondition counts:")
print(prompts_df["condition"].value_counts())

# ------------------------------------------------------------
# LOAD TOKENIZER
# ------------------------------------------------------------

print("\nLoading BLOOMZ tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

# Left padding is preferred for batched causal-LM generation
tokenizer.padding_side = "left"

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("Tokenizer loaded.")
print("EOS:", repr(tokenizer.eos_token))
print("PAD:", repr(tokenizer.pad_token))
print("Vocab:", tokenizer.vocab_size)

# ------------------------------------------------------------
# LOAD MODEL IN 8-BIT
# ------------------------------------------------------------

print("\nLoading BLOOMZ-7B1 in 8-bit...")

if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA GPU is required for the BLOOMZ run."
    )

quant_config = BitsAndBytesConfig(
    load_in_8bit=True
)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quant_config,
    device_map="auto"
)

model.eval()

print("Model loaded.")
print(
    "Parameters:",
    sum(p.numel() for p in model.parameters())
)

print(
    "GPU:",
    torch.cuda.get_device_name(0)
)

print(
    "GPU memory allocated:",
    round(
        torch.cuda.memory_allocated() / 1024**3,
        2
    ),
    "GB"
)

# ------------------------------------------------------------
# MODEL INPUT DEVICE
# ------------------------------------------------------------

model_input_device = next(
    model.parameters()
).device

print(
    "Model input device:",
    model_input_device
)

# ------------------------------------------------------------
# ANSWER EXTRACTION
# ------------------------------------------------------------

def extract_answer(raw_text):

    text = str(raw_text).strip()

    if not text:
        return "", "EMPTY"

    # Stop if BLOOMZ starts generating another QA item
    patterns = [
        r"\nQuestion\s*:",
        r"\nQ\s*:",
        r"\nQuestion\s*-\s*"
    ]

    earliest = None

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if match:

            if (
                earliest is None
                or match.start() < earliest
            ):
                earliest = match.start()

    if earliest is not None:

        answer = text[:earliest].strip()

        if answer:
            return answer, "TRUNCATED_AT_NEW_QUESTION"

    return text, "FULL_GENERATION"


# ------------------------------------------------------------
# GENERATION FUNCTION
# ------------------------------------------------------------

def generate_batch(batch_prompts):

    encoded = tokenizer(
        batch_prompts,
        padding=True,
        truncation=False,
        return_tensors="pt"
    )

    input_ids = encoded["input_ids"].to(
        model_input_device
    )

    attention_mask = encoded["attention_mask"].to(
        model_input_device
    )

    input_width = input_ids.shape[1]

    with torch.no_grad():

        outputs = model.generate(

            input_ids=input_ids,
            attention_mask=attention_mask,

            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=DO_SAMPLE,

            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.pad_token_id,

            use_cache=True
        )

    batch_results = []

    for i in range(len(batch_prompts)):

        generated_ids = outputs[i][
            input_width:
        ]

        generated_tokens = len(
            generated_ids
        )

        raw_text = tokenizer.decode(
            generated_ids,
            skip_special_tokens=True
        ).strip()

        eos_generated = (
            tokenizer.eos_token_id
            in generated_ids.tolist()
        )

        answer, extraction_status = (
            extract_answer(raw_text)
        )

        batch_results.append({

            "generated_answer":
                answer,

            "raw_generated_text":
                raw_text,

            "generated_tokens":
                generated_tokens,

            "eos_generated":
                eos_generated,

            "extraction_status":
                extraction_status,

            "status":
                "success"

        })

    return batch_results


# ------------------------------------------------------------
# GENERATION LOOP
# ------------------------------------------------------------

results = []

start_time = time.time()

print("\n" + "=" * 110)
print("STARTING 800 GENERATIONS")
print("=" * 110)

print("Model:", MODEL_NAME)
print("Max new tokens:", MAX_NEW_TOKENS)
print("Greedy decoding:", not DO_SAMPLE)
print("Batch size:", BATCH_SIZE)
print("=" * 110)

total = len(prompts_df)

for start in range(
    0,
    total,
    BATCH_SIZE
):

    end = min(
        start + BATCH_SIZE,
        total
    )

    batch_df = prompts_df.iloc[
        start:end
    ]

    batch_prompts = (
        batch_df["prompt"]
        .tolist()
    )

    batch_results = generate_batch(
        batch_prompts
    )

    for (_, row), gen_result in zip(
        batch_df.iterrows(),
        batch_results
    ):

        record = {

            "question_id":
                row["question_id"],

            "condition":
                row["condition"],

            # Preserve useful ground-truth metadata
            "company":
                row["company"]
                if "company" in row
                else "",

            "question":
                row["question"]
                if "question" in row
                else "",

            "generated_answer":
                gen_result[
                    "generated_answer"
                ],

            "raw_generated_text":
                gen_result[
                    "raw_generated_text"
                ],

            "generated_tokens":
                gen_result[
                    "generated_tokens"
                ],

            "eos_generated":
                gen_result[
                    "eos_generated"
                ],

            "extraction_status":
                gen_result[
                    "extraction_status"
                ],

            "status":
                gen_result[
                    "status"
                ]
        }

        # Preserve GT if it already exists in prompt file
        if "ground_truth_answer" in row:
            record[
                "ground_truth_answer"
            ] = row[
                "ground_truth_answer"
            ]

        results.append(record)

    completed = end

    # --------------------------------------------------------
    # PROGRESS
    # --------------------------------------------------------

    latest = results[-len(batch_results):]

    mean_tokens = sum(
        r["generated_tokens"]
        for r in latest
    ) / len(latest)

    print(
        f"[{completed:03d}/{total}] "
        f"completed | "
        f"last batch mean tokens={mean_tokens:.1f}",
        flush=True
    )

    # --------------------------------------------------------
    # CHECKPOINT
    # --------------------------------------------------------

    if (
        completed % CHECKPOINT_EVERY == 0
        or completed == total
    ):

        checkpoint_df = pd.DataFrame(
            results
        )

        checkpoint_df.to_csv(
            CHECKPOINT_FILE,
            index=False
        )

        elapsed = (
            time.time() - start_time
        ) / 60

        print(
            f"  CHECKPOINT SAVED → "
            f"{completed}/{total} | "
            f"{elapsed:.2f} min"
        )

    # CUDA cleanup
    torch.cuda.empty_cache()


# ------------------------------------------------------------
# SAVE FINAL OUTPUT
# ------------------------------------------------------------

final_df = pd.DataFrame(
    results
)

final_df.to_csv(
    OUTPUT_FILE,
    index=False
)

elapsed_total = (
    time.time() - start_time
) / 60


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 110)
print("BLOOMZ-7B1 GENERATION COMPLETE")
print("=" * 110)

print(
    "Generation file:",
    OUTPUT_FILE
)

print(
    "Shape:",
    final_df.shape
)

print(
    "Successful:",
    (
        final_df["status"]
        .astype(str)
        .str.lower()
        .eq("success")
        .sum()
    )
)

print("\nCondition counts:")
print(
    final_df["condition"].value_counts()
)

print("\nExtraction status:")
print(
    final_df[
        "extraction_status"
    ].value_counts()
)

print("\nEOS counts:")
print(
    final_df[
        "eos_generated"
    ].value_counts()
)

print("\nGenerated-token statistics:")
print(
    final_df[
        "generated_tokens"
    ].describe()
)

print("\nMean tokens by condition:")
print(
    final_df.groupby(
        "condition"
    )[
        "generated_tokens"
    ].agg(
        [
            "count",
            "mean",
            "median",
            "min",
            "max"
        ]
    )
)

print(
    f"\nTotal generation time: "
    f"{elapsed_total:.2f} minutes"
)


# ------------------------------------------------------------
# SAMPLE OUTPUTS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("SAMPLE OUTPUTS")
print("=" * 110)

for _, row in final_df.head(10).iterrows():

    print("\n" + "-" * 80)

    print(
        row["question_id"],
        "|",
        row["condition"]
    )

    print(
        "Tokens:",
        row["generated_tokens"]
    )

    print(
        "EOS:",
        row["eos_generated"]
    )

    print(
        "Extraction:",
        row["extraction_status"]
    )

    print("\nAnswer:")

    print(
        row["generated_answer"][:1000]
    )


# ------------------------------------------------------------
# CLEANUP
# ------------------------------------------------------------

del model
del tokenizer

gc.collect()
torch.cuda.empty_cache()
torch.cuda.ipc_collect()

print("\nGPU memory released.")

print("\n" + "=" * 110)
print("DONE — BLOOMZ IS THE LAST MODEL")
print("=" * 110)
```

```python
# ============================================================
# BLOOMZ HOTFIX + FINAL 800 GENERATION
# NO KERNEL RESTART REQUIRED
# ============================================================

import os
import gc
import re
import time
import torch
import pandas as pd

from transformers import AutoTokenizer, AutoModelForCausalLM
from transformers import BitsAndBytesConfig

# ------------------------------------------------------------
# 1. PATCH THE ALREADY-LOADED TRANSFORMERS BNB CHECK
# ------------------------------------------------------------

import bitsandbytes as bnb
import transformers

from transformers.quantizers import quantizer_bnb_8bit

print("=" * 100)
print("BLOOMZ HOTFIX")
print("=" * 100)

print("bitsandbytes:", bnb.__version__)
print("CUDA:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0))

# The quantizer module was loaded before bitsandbytes became
# available. Patch its stale availability function directly.
quantizer_bnb_8bit.is_bitsandbytes_available = lambda: True

print(
    "Quantizer sees bitsandbytes:",
    quantizer_bnb_8bit.is_bitsandbytes_available()
)

# ------------------------------------------------------------
# 2. CONFIG
# ------------------------------------------------------------

MODEL_NAME = "bigscience/bloomz-7b1"

PROMPTS_FILE = (
    "/kaggle/working/pythia_800_prompts.csv"
)

OUTPUT_FILE = (
    "/kaggle/working/bloomz_800_generations_final.csv"
)

CHECKPOINT_FILE = (
    "/kaggle/working/bloomz_800_generation_checkpoint_final.csv"
)

MAX_NEW_TOKENS = 128
BATCH_SIZE = 2
CHECKPOINT_EVERY = 50

# ------------------------------------------------------------
# 3. LOAD PROMPTS
# ------------------------------------------------------------

prompts_df = pd.read_csv(PROMPTS_FILE)

print("\nPrompt shape:", prompts_df.shape)

if len(prompts_df) != 800:
    raise ValueError(
        f"Expected 800 prompts, got {len(prompts_df)}"
    )

print(
    prompts_df["condition"].value_counts()
)

# ------------------------------------------------------------
# 4. LOAD TOKENIZER
# ------------------------------------------------------------

print("\nLoading BLOOMZ tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

tokenizer.padding_side = "left"

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("Tokenizer ready.")

# ------------------------------------------------------------
# 5. LOAD BLOOMZ-7B1 IN 8-BIT
# ------------------------------------------------------------

print("\nLoading BLOOMZ-7B1 in 8-bit...")

quant_config = BitsAndBytesConfig(
    load_in_8bit=True
)

# Clean any stale object if present
try:
    del model
except:
    pass

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quant_config,
    device_map="auto"
)

model.eval()

print("\n✓ BLOOMZ MODEL LOADED")

print(
    "Parameters:",
    sum(p.numel() for p in model.parameters())
)

print(
    "GPU memory:",
    round(
        torch.cuda.memory_allocated() / 1024**3,
        2
    ),
    "GB"
)

model_input_device = next(
    model.parameters()
).device

print(
    "Input device:",
    model_input_device
)

# ------------------------------------------------------------
# 6. ANSWER EXTRACTION
# ------------------------------------------------------------

def extract_answer(text):

    text = str(text).strip()

    if not text:
        return "", "EMPTY"

    patterns = [
        r"\nQuestion\s*:",
        r"\nQ\s*:",
        r"\nQuestion\s*-\s*"
    ]

    earliest = None

    for pattern in patterns:

        m = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if m is not None:

            if (
                earliest is None
                or m.start() < earliest
            ):
                earliest = m.start()

    if earliest is not None:

        ans = text[:earliest].strip()

        if ans:
            return (
                ans,
                "TRUNCATED_AT_NEW_QUESTION"
            )

    return text, "FULL_GENERATION"


# ------------------------------------------------------------
# 7. GENERATION
# ------------------------------------------------------------

def generate_batch(prompts):

    enc = tokenizer(
        prompts,
        padding=True,
        truncation=False,
        return_tensors="pt"
    )

    input_ids = enc["input_ids"].to(
        model_input_device
    )

    attention_mask = enc["attention_mask"].to(
        model_input_device
    )

    input_width = input_ids.shape[1]

    with torch.no_grad():

        outputs = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,

            max_new_tokens=MAX_NEW_TOKENS,

            # Greedy — same successful BLOOMZ setup
            do_sample=False,

            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.pad_token_id,

            use_cache=True
        )

    batch_results = []

    for i in range(len(prompts)):

        new_ids = outputs[i][
            input_width:
        ]

        generated_tokens = len(new_ids)

        raw_text = tokenizer.decode(
            new_ids,
            skip_special_tokens=True
        ).strip()

        eos_generated = (
            tokenizer.eos_token_id
            in new_ids.tolist()
        )

        answer, extraction_status = (
            extract_answer(raw_text)
        )

        batch_results.append({
            "generated_answer": answer,
            "raw_generated_text": raw_text,
            "generated_tokens": generated_tokens,
            "eos_generated": eos_generated,
            "extraction_status": extraction_status,
            "status": "success"
        })

    return batch_results


# ------------------------------------------------------------
# 8. RUN ALL 800
# ------------------------------------------------------------

results = []

start_time = time.time()

print("\n" + "=" * 100)
print("STARTING BLOOMZ 800-GENERATION RUN")
print("=" * 100)

for start in range(
    0,
    len(prompts_df),
    BATCH_SIZE
):

    end = min(
        start + BATCH_SIZE,
        len(prompts_df)
    )

    batch_df = prompts_df.iloc[
        start:end
    ]

    batch_results = generate_batch(
        batch_df["prompt"].tolist()
    )

    for (_, row), result in zip(
        batch_df.iterrows(),
        batch_results
    ):

        results.append({

            "question_id":
                row["question_id"],

            "condition":
                row["condition"],

            "question":
                row["question"],

            "generated_answer":
                result["generated_answer"],

            "raw_generated_text":
                result["raw_generated_text"],

            "generated_tokens":
                result["generated_tokens"],

            "eos_generated":
                result["eos_generated"],

            "extraction_status":
                result["extraction_status"],

            "status":
                result["status"]
        })

    completed = end

    print(
        f"[{completed:03d}/800] complete",
        flush=True
    )

    # --------------------------------------------------------
    # CHECKPOINT
    # --------------------------------------------------------

    if (
        completed % CHECKPOINT_EVERY == 0
        or completed == 800
    ):

        pd.DataFrame(results).to_csv(
            CHECKPOINT_FILE,
            index=False
        )

        elapsed = (
            time.time() - start_time
        ) / 60

        print(
            f"  CHECKPOINT SAVED | "
            f"{completed}/800 | "
            f"{elapsed:.2f} min"
        )

    torch.cuda.empty_cache()


# ------------------------------------------------------------
# 9. FINAL SAVE
# ------------------------------------------------------------

final_df = pd.DataFrame(results)

final_df.to_csv(
    OUTPUT_FILE,
    index=False
)

elapsed_total = (
    time.time() - start_time
) / 60


# ============================================================
# 10. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("BLOOMZ-7B1 GENERATION COMPLETE")
print("=" * 100)

print(
    "Shape:",
    final_df.shape
)

print(
    "Successful:",
    (
        final_df["status"]
        .eq("success")
        .sum()
    )
)

print("\nCondition counts:")
print(
    final_df["condition"].value_counts()
)

print("\nExtraction:")
print(
    final_df[
        "extraction_status"
    ].value_counts()
)

print("\nEOS:")
print(
    final_df[
        "eos_generated"
    ].value_counts()
)

print("\nGenerated tokens:")
print(
    final_df[
        "generated_tokens"
    ].describe()
)

print("\nMean tokens by condition:")
print(
    final_df.groupby(
        "condition"
    )["generated_tokens"].agg(
        [
            "count",
            "mean",
            "median",
            "min",
            "max"
        ]
    )
)

print(
    f"\nTotal time: "
    f"{elapsed_total:.2f} minutes"
)

print(
    "\nSaved:",
    OUTPUT_FILE
)

# ------------------------------------------------------------
# CLEANUP
# ------------------------------------------------------------

del model
del tokenizer

gc.collect()
torch.cuda.empty_cache()
torch.cuda.ipc_collect()

print("\n✓ GPU memory released.")
print("✓ BLOOMZ FINAL GENERATION DONE.")
```

```python
# ============================================================
# PHASE 7 — BLOOMZ-7B1 EVALUATION
# Evaluate all 800 generations using the same metrics/config
# used for Llama-2-7B, Pythia and FLAN-T5.
# ============================================================

import os
import re
import math
import numpy as np
import pandas as pd
import torch

from rapidfuzz.distance import Jaro
from transformers import BartForConditionalGeneration, BartTokenizer
from bert_score import score as bertscore_score

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

GEN_PATH = "/kaggle/working/bloomz_800_generations_final.csv"
GT_PATH = "/kaggle/input/datasets/ramnarenkakulapati/emo-bro/ground_truth_400.csv"

OUT_METRICS = "/kaggle/working/bloomz_7b1_question_metrics.csv"
OUT_TABLE = "/kaggle/working/bloomz_7b1_final_table4.csv"

# ------------------------------------------------------------
# 2. LOAD GENERATIONS + GROUND TRUTH
# ------------------------------------------------------------

gen = pd.read_csv(GEN_PATH)
gt = pd.read_csv(GT_PATH)

print("Generation shape:", gen.shape)
print("Generation columns:", gen.columns.tolist())

print("\nGround-truth shape:", gt.shape)
print("Ground-truth columns:", gt.columns.tolist())

# ------------------------------------------------------------
# 3. RECOVER ANSWER TEXT
#    Use answer_output, fallback to raw_output if needed.
# ------------------------------------------------------------

if "answer_output" in gen.columns:
    gen["prediction"] = gen["answer_output"]

    if "raw_output" in gen.columns:
        mask = gen["prediction"].isna() | (
            gen["prediction"].astype(str).str.strip() == ""
        )
        gen.loc[mask, "prediction"] = gen.loc[mask, "raw_output"]

elif "raw_output" in gen.columns:
    gen["prediction"] = gen["raw_output"]

else:
    raise ValueError("Neither answer_output nor raw_output exists.")

gen["prediction"] = gen["prediction"].fillna("").astype(str)

# ------------------------------------------------------------
# 4. MERGE GROUND TRUTH
# ------------------------------------------------------------

required_gt = ["question_id", "ground_truth_answer"]

for c in required_gt:
    if c not in gt.columns:
        raise ValueError(f"Missing GT column: {c}")

eval_df = gen.merge(
    gt[["question_id", "ground_truth_answer"]],
    on="question_id",
    how="left",
    validate="one_to_one"
)

eval_df["ground_truth_answer"] = (
    eval_df["ground_truth_answer"]
    .fillna("")
    .astype(str)
)

print("\nMerged shape:", eval_df.shape)
print("Missing GT:", eval_df["ground_truth_answer"].eq("").sum())

# ------------------------------------------------------------
# 5. BASIC VALIDATION
# ------------------------------------------------------------

assert len(eval_df) == 800, f"Expected 800 rows, got {len(eval_df)}"
assert eval_df["question_id"].nunique() == 400
assert set(eval_df["condition"].unique()) == {
    "WITHOUT_METADATA",
    "WITH_METADATA"
}

print("\nCondition counts:")
print(eval_df["condition"].value_counts())

# ------------------------------------------------------------
# 6. BERTSCORE
#    Same setup used previously:
#    model = roberta-large
#    F1
#    rescale_with_baseline = False
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("BERTSCORE")
print("=" * 80)

preds = eval_df["prediction"].tolist()
refs = eval_df["ground_truth_answer"].tolist()

P, R, F1 = bertscore_score(
    preds,
    refs,
    model_type="roberta-large",
    lang="en",
    rescale_with_baseline=False,
    verbose=True
)

eval_df["bertscore_f1"] = F1.cpu().numpy()

print(
    "Mean BERTScore F1:",
    eval_df["bertscore_f1"].mean()
)

# ------------------------------------------------------------
# 7. JARO SIMILARITY
# ------------------------------------------------------------

def jaro_similarity(a, b):
    return Jaro.normalized_similarity(str(a), str(b))

eval_df["jaro"] = [
    jaro_similarity(p, r)
    for p, r in zip(
        eval_df["prediction"],
        eval_df["ground_truth_answer"]
    )
]

print("\nMean Jaro:", eval_df["jaro"].mean())

# ------------------------------------------------------------
# 8. CHARACTER-LEVEL LCS
#    LCS similarity = LCS length / max(len(pred), len(ref))
# ------------------------------------------------------------

def lcs_length_chars(a, b):
    a = str(a)
    b = str(b)

    if not a or not b:
        return 0

    # Keep memory manageable
    if len(a) < len(b):
        a, b = b, a

    prev = [0] * (len(b) + 1)

    for ca in a:
        curr = [0] * (len(b) + 1)

        for j, cb in enumerate(b, start=1):
            if ca == cb:
                curr[j] = prev[j - 1] + 1
            else:
                curr[j] = max(prev[j], curr[j - 1])

        prev = curr

    return prev[-1]


def lcs_similarity_chars(a, b):
    a = str(a)
    b = str(b)

    denom = max(len(a), len(b))

    if denom == 0:
        return 1.0

    return lcs_length_chars(a, b) / denom


eval_df["lcs_similarity"] = [
    lcs_similarity_chars(p, r)
    for p, r in zip(
        eval_df["prediction"],
        eval_df["ground_truth_answer"]
    )
]

print(
    "Mean LCS similarity:",
    eval_df["lcs_similarity"].mean()
)

# ------------------------------------------------------------
# 9. WORD-LEVEL LCS WORD COUNT
# ------------------------------------------------------------

def lcs_length_words(a, b):
    a = str(a).split()
    b = str(b).split()

    if not a or not b:
        return 0

    if len(a) < len(b):
        a, b = b, a

    prev = [0] * (len(b) + 1)

    for wa in a:
        curr = [0] * (len(b) + 1)

        for j, wb in enumerate(b, start=1):
            if wa == wb:
                curr[j] = prev[j - 1] + 1
            else:
                curr[j] = max(prev[j], curr[j - 1])

        prev = curr

    return prev[-1]


eval_df["lcs_word_count"] = [
    lcs_length_words(p, r)
    for p, r in zip(
        eval_df["prediction"],
        eval_df["ground_truth_answer"]
    )
]

print(
    "Mean LCS word count:",
    eval_df["lcs_word_count"].mean()
)

# ------------------------------------------------------------
# 10. BARTSCORE
#     Same bidirectional average log-likelihood implementation
#     used in our previous model evaluations.
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("BARTSCORE")
print("=" * 80)

BART_MODEL = "facebook/bart-large-cnn"

bart_tokenizer = BartTokenizer.from_pretrained(BART_MODEL)

bart_model = BartForConditionalGeneration.from_pretrained(
    BART_MODEL,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32
)

bart_device = "cuda" if torch.cuda.is_available() else "cpu"
bart_model = bart_model.to(bart_device)
bart_model.eval()

print("BART device:", bart_device)


def bart_directional_score(
    sources,
    targets,
    batch_size=4,
    max_length=1024
):
    scores = []

    for start in range(0, len(sources), batch_size):

        src_batch = sources[start:start + batch_size]
        tgt_batch = targets[start:start + batch_size]

        enc = bart_tokenizer(
            src_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        )

        labels = bart_tokenizer(
            tgt_batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        )["input_ids"]

        enc = {k: v.to(bart_device) for k, v in enc.items()}
        labels = labels.to(bart_device)

        labels[labels == bart_tokenizer.pad_token_id] = -100

        with torch.no_grad():
            outputs = bart_model(
                **enc,
                labels=labels
            )

            logits = outputs.logits

        log_probs = torch.log_softmax(logits, dim=-1)

        safe_labels = labels.clone()
        safe_labels[safe_labels == -100] = 0

        token_log_probs = log_probs.gather(
            dim=-1,
            index=safe_labels.unsqueeze(-1)
        ).squeeze(-1)

        mask = (labels != -100)

        seq_scores = (
            token_log_probs * mask
        ).sum(dim=1) / mask.sum(dim=1).clamp(min=1)

        scores.extend(
            seq_scores.float().cpu().numpy().tolist()
        )

    return np.array(scores)


# candidate -> reference
score_pred_to_ref = bart_directional_score(
    eval_df["prediction"].tolist(),
    eval_df["ground_truth_answer"].tolist()
)

# reference -> candidate
score_ref_to_pred = bart_directional_score(
    eval_df["ground_truth_answer"].tolist(),
    eval_df["prediction"].tolist()
)

eval_df["bartscore"] = (
    score_pred_to_ref + score_ref_to_pred
) / 2.0

print(
    "Mean BARTScore:",
    eval_df["bartscore"].mean()
)

# ------------------------------------------------------------
# 11. CONDITION-LEVEL SUMMARY
# ------------------------------------------------------------

summary = (
    eval_df
    .groupby("condition")
    .agg(
        BERTScore=("bertscore_f1", "mean"),
        BARTScore=("bartscore", "mean"),
        Jaro=("jaro", "mean"),
        LCS_Similarity=("lcs_similarity", "mean"),
        LCS_Word_Count=("lcs_word_count", "mean"),
        N=("question_id", "count")
    )
    .reset_index()
)

print("\n" + "=" * 80)
print("BLOOMZ CONDITION RESULTS")
print("=" * 80)
print(summary.to_string(index=False))

# ------------------------------------------------------------
# 12. CREATE PAPER-STYLE TABLE 4
# ------------------------------------------------------------

without = summary[
    summary["condition"] == "WITHOUT_METADATA"
].iloc[0]

with_meta = summary[
    summary["condition"] == "WITH_METADATA"
].iloc[0]

table4 = pd.DataFrame({
    "Model": ["BLOOMZ-7b1", "BLOOMZ-7b1"],
    "Metadata": ["Without", "With"],
    "BERTScore": [
        without["BERTScore"],
        with_meta["BERTScore"]
    ],
    "BARTScore": [
        without["BARTScore"],
        with_meta["BARTScore"]
    ],
    "Jaro": [
        without["Jaro"],
        with_meta["Jaro"]
    ],
    "LCSubsequence": [
        without["LCS_Similarity"],
        with_meta["LCS_Similarity"]
    ],
    "LCSubsequence Word Count": [
        without["LCS_Word_Count"],
        with_meta["LCS_Word_Count"]
    ]
})

# Rounded paper-style display
table4_rounded = table4.copy()

for c in ["BERTScore", "BARTScore", "Jaro", "LCSubsequence"]:
    table4_rounded[c] = table4[c].round(2)

table4_rounded["LCSubsequence Word Count"] = (
    table4["LCSubsequence Word Count"]
    .round(0)
    .astype(int)
)

print("\n" + "=" * 80)
print("TABLE 4 — BLOOMZ-7B1")
print("=" * 80)
print(table4_rounded.to_string(index=False))

# ------------------------------------------------------------
# 13. SAVE EVERYTHING
# ------------------------------------------------------------

eval_df.to_csv(OUT_METRICS, index=False)
table4.to_csv(OUT_TABLE, index=False)

print("\nSaved:")
print(OUT_METRICS)
print(OUT_TABLE)

# ------------------------------------------------------------
# 14. IMPROVEMENT WITH METADATA
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("METADATA EFFECT")
print("=" * 80)

for metric in [
    "BERTScore",
    "BARTScore",
    "Jaro",
    "LCS_Similarity",
    "LCS_Word_Count"
]:

    w = float(with_meta[metric])
    wo = float(without[metric])

    print(
        f"{metric:25s}: "
        f"{wo:.6f} -> {w:.6f}   "
        f"Delta = {w - wo:+.6f}"
    )

print("\n✓ BLOOMZ EVALUATION COMPLETE")
```
